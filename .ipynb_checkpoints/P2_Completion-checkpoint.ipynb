# Paper 2 — Experiment Completion

This notebook finishes the experimental programme. Two questions remain open after the v3 runs, and this
notebook answers both, then consolidates every number the manuscript restructure needs.

| Step | Question | Cost |
|---|---|---|
| **1** | Does the educational-utility advantage survive against **DKVMN**, the strongest baseline on Junyi and the model that beats PCDT at retention there? | seconds — reads saved predictions, no retraining |
| **2** | Is the Junyi PCDT-vs-DKVMN retention result **seed-robust**, or an artefact of seed 20260202? | ~160 min — one full Junyi run at seed 20260203 |
| **3** | Cross-seed comparison and explicit verdict | seconds |
| **4** | Consolidated `manuscript_numbers.json` — every figure needed to rewrite the paper | seconds |

**Why these two.** Step 1 decides how the educational claim must be *worded*: if the advantage exists
only against a session-resetting control, the paper cannot claim it against the state of the art. Step 2
decides whether the study's most consequential negative finding is real; it currently rests on a single
seed, and a referee would ask.

**Run order.** Set `RUN_STEP2 = False` first and execute — Step 1 finishes in seconds and may change the
wording of the educational claim immediately. Then set `RUN_STEP2 = True` and leave it running.

**No number in the manuscript may come from anywhere but the files this notebook writes.**

## 0. Configuration — EDIT THIS CELL

```python
import os, sys, json, time, glob, warnings
import numpy as np, pandas as pd
warnings.filterwarnings("ignore", category=FutureWarning)

# ---- paths -------------------------------------------------------------------------------------
CACHE       = "/home/hamed/Downloads/Research/P02/PCDT_P2_cache_rev_V3"
PIPELINE_NB = "/home/hamed/Downloads/Research/P02/P2_Revision_Pipeline_v3_junyi.ipynb"   # the v3 notebook
DATASETS    = ["assistments_2021", "ednet_flat", "junyi"]

# ---- what to run -------------------------------------------------------------------------------
RUN_STEP1 = True      # DKVMN utility               (seconds)
RUN_STEP2 = False     # Junyi seed confirmation     (~160 min)  <-- flip to True for the long run
RUN_STEP3 = True      # cross-seed comparison       (seconds, needs Step 2 to have run once)
RUN_STEP4 = True      # consolidated manuscript numbers

# ---- analysis constants (unchanged from the pre-registration) -----------------------------------
BUCKETS    = [("within",0,20*60),("sameday",20*60,86400),("d1",86400,7*86400),
              ("d7",7*86400,28*86400),("d28",28*86400,np.inf)]
THRESHOLDS = (0.3, 0.4, 0.5, 0.6, 0.7)
MIN_POINTS = 2_000
MMDE       = 0.010          # pre-registered minimum meaningful ΔAUC

# verdict containers -- defined up front so Step 4 works even when Steps 1-3 are skipped
VERDICT_UTILITY, VERDICT_SEED = {}, {}
UTIL = pd.DataFrame()

RES = os.path.join(CACHE, "results")
OUT = os.path.join(RES, "COMPLETION"); os.makedirs(OUT, exist_ok=True)
pd.set_option("display.width", 220); pd.set_option("display.max_columns", 40)

print("cache :", CACHE)
print("output:", OUT)
for d in DATASETS:
    p = os.path.join(RES, f"REV_{d}", "predictions_test.npz")
    print(f"  {d:20s} predictions: {'found' if os.path.exists(p) else 'MISSING'}")
```

## Step 1 — Educational utility against DKVMN

The gap in the v3 results: utility was computed against the session-resetting control and against DKT-F,
but never against DKVMN. On Junyi, DKVMN is both the strongest baseline *and* the model that beats PCDT
at retention on every horizon — and Junyi is the one dataset where the utility advantage over DKT-F
holds. So the single dataset carrying the educational claim against a strong baseline is the one with an
even stronger, untested comparator.

Decision rule fixed before looking: an operating point counts as a **strict** win only if PCDT flags more
learners who genuinely needed review **and** raises no additional false alarms.

```python
from sklearn.metrics import confusion_matrix

def utility(p, y, thr):
    """Flag for review when predicted success probability < thr."""
    flag = (p < thr).astype(int); need = (y == 0).astype(int)
    tn, fp, fn, tp = confusion_matrix(need, flag, labels=[0,1]).ravel()
    prec = tp/(tp+fp) if tp+fp else np.nan
    rec  = tp/(tp+fn) if tp+fn else np.nan
    spec = tn/(tn+fp) if tn+fp else np.nan
    f1   = 2*prec*rec/(prec+rec) if (prec and rec and prec+rec) else np.nan
    return dict(correctly_flagged=int(tp), false_alarms=int(fp), missed=int(fn),
                precision=prec, recall=rec, f1=f1,
                balanced_accuracy=(rec+spec)/2 if (rec and spec) else np.nan)

def utility_table(dataset, comparators=("Control","DKT-F","DKVMN")):
    z = np.load(os.path.join(RES, f"REV_{dataset}", "predictions_test.npz"))
    have = [k for k in z.files if k not in ("targets","mask","bucket")]
    Y, M, B = z["targets"], z["mask"] > 0, z["bucket"]
    rows = []
    for comp in comparators:
        if comp not in have:
            print(f"  [{dataset}] {comp} not in cache (have {have}) -- skipped"); continue
        for bid,(name,lo,hi) in enumerate(BUCKETS):
            sel = (B == bid) & M
            if sel.sum() < MIN_POINTS: continue
            for thr in THRESHOLDS:
                up = utility(z["PCDT"][sel], Y[sel], thr)
                uc = utility(z[comp][sel],   Y[sel], thr)
                rows.append(dict(dataset=dataset, horizon=name, comparator=comp, threshold=thr,
                    n_points=int(sel.sum()),
                    extra_correctly_flagged=up["correctly_flagged"]-uc["correctly_flagged"],
                    extra_false_alarms=up["false_alarms"]-uc["false_alarms"],
                    d_precision=up["precision"]-uc["precision"],
                    d_recall=up["recall"]-uc["recall"],
                    d_f1=up["f1"]-uc["f1"],
                    d_balanced_acc=up["balanced_accuracy"]-uc["balanced_accuracy"]))
    return pd.DataFrame(rows)

if RUN_STEP1:
    UTIL = pd.concat([utility_table(d) for d in DATASETS], ignore_index=True)
    UTIL.to_csv(f"{OUT}/table6b_utility_all_comparators.csv", index=False)
    with open(f"{OUT}/table6b_utility_all_comparators.tex","w") as fh:
        fh.write(UTIL.to_latex(index=False, float_format="%.4f", escape=True))
    real = UTIL[UTIL.horizon.isin(["d1","d7","d28"])]
    print("\n===== PCDT vs DKVMN, real horizons =====")
    dk = real[real.comparator=="DKVMN"]
    print(dk[["dataset","horizon","threshold","extra_correctly_flagged","extra_false_alarms",
              "d_precision","d_f1"]].round(4).to_string(index=False) if len(dk) else "  none")
```

```python
# ---- explicit verdict on the educational claim -------------------------------------------------
if RUN_STEP1 and len(UTIL):
    real = UTIL[UTIL.horizon.isin(["d1","d7","d28"])]
    for comp in ["Control","DKT-F","DKVMN"]:
        sub = real[real.comparator==comp]
        if not len(sub): continue
        strict = sub[(sub.extra_correctly_flagged>0) & (sub.extra_false_alarms<=0)]
        VERDICT_UTILITY[comp] = dict(
            n_cells=len(sub), n_strict_wins=len(strict),
            datasets_with_strict_win=sorted(strict.dataset.unique().tolist()))
        print(f"vs {comp:8s}: {len(strict):>2}/{len(sub):>2} operating points are STRICT wins "
              f"(more correct flags AND no extra false alarms) "
              f"| datasets: {sorted(strict.dataset.unique().tolist()) or 'none'}")

    dk = VERDICT_UTILITY.get("DKVMN", {})
    print("\n---- what the manuscript may claim ----")
    if dk.get("n_strict_wins", 0) == 0:
        print("NO strict win against DKVMN on any dataset or horizon.")
        print("=> Word the educational claim as: 'time-aware modelling improves review targeting over a")
        print("   session-resetting baseline'. Do NOT claim an advantage over the strongest baseline.")
    else:
        print(f"Strict wins against DKVMN on: {dk['datasets_with_strict_win']}")
        print("=> The educational claim MAY be stated against the strongest baseline, but only for")
        print("   those datasets and horizons. Name them explicitly; do not generalise.")
    json.dump(VERDICT_UTILITY, open(f"{OUT}/verdict_utility.json","w"), indent=2)
```

## Step 2 — Junyi seed confirmation

Executes the v3 pipeline for Junyi at `seed_index = 1` (seed 20260203) by running its code cells in
order, with only the dataset and seed patched. Results land in `REV_junyi_seed2`, so the primary Junyi
run is untouched and both survive for comparison.

Roughly 160 minutes. Everything else — split logic, the eight models, capacity matching, thresholds —
is identical to the primary run, so any difference in the result is attributable to the seed alone.

If you would rather run it by hand, open the v3 notebook, set `CFG.dataset="junyi"` and
`CFG.seed_index=1`, and Run All. This cell is a convenience, not a requirement.

```python
if RUN_STEP2:
    t0 = time.time()
    nbj = json.load(open(PIPELINE_NB))
    src = "\n".join("".join(c["source"]) for c in nbj["cells"] if c["cell_type"] == "code")

    # patch ONLY the dataset and the seed; nothing else may differ from the primary run
    import re as _re
    before = src
    src = _re.sub(r'dataset:\s*str\s*=\s*"[^"]+"', 'dataset: str = "junyi"', src, count=1)
    src = _re.sub(r'seed_index:\s*int\s*=\s*\d+',   'seed_index: int = 1',   src, count=1)
    assert src != before, "config patch failed -- check the v3 notebook's Config field names"
    # strip any pip magics that would break exec
    src = "\n".join(l for l in src.split("\n") if not l.strip().startswith("!"))

    print("running the v3 pipeline for junyi at seed_index=1 ...")
    ns = {"__name__": "__main__"}
    exec(compile(src, "<v3-pipeline-seed2>", "exec"), ns)
    print(f"\nSTEP 2 complete in {(time.time()-t0)/60:.1f} min")
else:
    print("RUN_STEP2 is False -- skipping the 160-minute confirmation run")
```

## Step 3 — Cross-seed comparison and verdict

Compares the two Junyi runs on the numbers that matter. The decision rule is fixed here rather than
after seeing the output: the result is **confirmed** if every PCDT-vs-DKVMN retention advantage keeps its
sign across both seeds and no confidence interval crosses zero in either run.

```python
def load_persistence(tag=""):
    p = os.path.join(RES, f"REV_junyi{tag}", "tables", "table5_persistence.csv")
    return pd.read_csv(p) if os.path.exists(p) else None

if RUN_STEP3:
    a, b = load_persistence(""), load_persistence("_seed2")
    if a is None or b is None:
        print("seed-2 results not found -- run Step 2 first (set RUN_STEP2 = True)")
    else:
        k = ["horizon","comparator"]
        m = a[k+["retention_advantage","ci_lo","ci_hi","adequate"]].merge(
            b[k+["retention_advantage","ci_lo","ci_hi","adequate"]], on=k, suffixes=("_s1","_s2"))
        m["sign_agrees"] = np.sign(m.retention_advantage_s1) == np.sign(m.retention_advantage_s2)
        m["both_exclude_zero"] = ((m.ci_lo_s1*m.ci_hi_s1 > 0) & (m.ci_lo_s2*m.ci_hi_s2 > 0))
        m["abs_change"] = (m.retention_advantage_s2 - m.retention_advantage_s1).abs()
        m.to_csv(f"{OUT}/seed_comparison_junyi.csv", index=False)
        print(m.round(4).to_string(index=False))

        dk = m[(m.comparator=="DKVMN") & m.adequate_s1.fillna(False) & m.adequate_s2.fillna(False)]
        confirmed = bool(len(dk)) and bool(dk.sign_agrees.all() and dk.both_exclude_zero.all())
        VERDICT_SEED = dict(n_cells=int(len(dk)), confirmed=confirmed,
                            max_abs_change=float(dk.abs_change.max()) if len(dk) else None)
        print("\n---- verdict on the DKVMN retention result ----")
        if confirmed:
            print("CONFIRMED across both seeds: every PCDT-vs-DKVMN cell keeps its sign and excludes zero.")
            print(f"Largest change between seeds: {dk.abs_change.max():.4f}")
            print("=> Report as a robust negative result. Psi does not outperform a memory-augmented")
            print("   tracer at retention on Junyi.")
        else:
            print("NOT CONFIRMED -- signs or significance disagree between seeds.")
            print("=> Do NOT report the DKVMN comparison as settled. Run a third seed (seed_index=2)")
            print("   before drawing any conclusion; one of the two runs is anomalous.")
        json.dump(VERDICT_SEED, open(f"{OUT}/verdict_seed.json","w"), indent=2)
```

## Step 4 — Consolidated manuscript numbers

One file containing every figure needed to rewrite the paper, pulled from the run outputs rather than
retyped. If a number is not in here, it does not go in the manuscript.

```python
if RUN_STEP4:
    NUM = {"generated": time.strftime("%Y-%m-%d %H:%M"), "datasets": {}}
    for d in DATASETS:
        base = os.path.join(RES, f"REV_{d}")
        entry = {}
        rp = os.path.join(base, "reproducibility.json")
        if os.path.exists(rp):
            r = json.load(open(rp))
            entry["provenance"] = {k: r.get(k) for k in
                ("dataset","seed","seed_index","strongest_baseline","runtime_minutes",
                 "n_learners","n_interactions","n_kcs","split","horizons_real","horizons_inadequate")}
        t2 = os.path.join(base,"tables","table2_predictive.csv")
        if os.path.exists(t2):
            t = pd.read_csv(t2)
            entry["within_session"] = t[["model","params","auc","acc","rmse","nll","ece"]].round(4).to_dict("records")
            base_models = ["DKT","SAKT","AKT","DKVMN","DKT-F"]
            bb = t[t.model.isin(base_models)].sort_values("auc", ascending=False)
            if len(bb):
                pcdt = float(t[t.model=="PCDT"].auc.iloc[0])
                entry["H1"] = dict(strongest_baseline=bb.iloc[0]["model"],
                                   baseline_auc=round(float(bb.iloc[0]["auc"]),4),
                                   pcdt_auc=round(pcdt,4),
                                   delta=round(pcdt-float(bb.iloc[0]["auc"]),4),
                                   meaningful=bool(abs(pcdt-float(bb.iloc[0]["auc"])) >= MMDE),
                                   mmde=MMDE)
        for name,key in [("table5_persistence","persistence"),("table7_ablation","ablation"),
                         ("tableA_temporal_profile","temporal"),("table3_statistics","statistics")]:
            p = os.path.join(base,"tables",f"{name}.csv")
            if os.path.exists(p):
                entry[key] = pd.read_csv(p).round(4).to_dict("records")
        NUM["datasets"][d] = entry

    if len(UTIL):     NUM["educational_utility"] = UTIL.round(4).to_dict("records")
    if VERDICT_UTILITY: NUM["verdict_utility"] = VERDICT_UTILITY
    if VERDICT_SEED:    NUM["verdict_seed"] = VERDICT_SEED

    with open(f"{OUT}/manuscript_numbers.json","w") as fh: json.dump(NUM, fh, indent=2, default=str)
    print("wrote", f"{OUT}/manuscript_numbers.json")
    for d in DATASETS:
        h = NUM["datasets"].get(d,{}).get("H1")
        if h: print(f"  {d:20s} H1: PCDT {h['pcdt_auc']} vs {h['strongest_baseline']} "
                    f"{h['baseline_auc']} = {h['delta']:+.4f} -> "
                    f"{'MEANINGFUL' if h['meaningful'] else 'below MMDE (H1 not supported)'}")
```
