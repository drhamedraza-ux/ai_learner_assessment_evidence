# Experiment P2-04 — Representational Stability and Drift of the PCDT State

**Paper 2:** *A Persistent Multidimensional State Representation for Cognitive Digital Twins*
**Research question:** RQ2.4
**Produces:** Manuscript **Table 6** (state drift under repeated inputs, vs baseline hidden states) and
**Table 7** (hyperparameter sensitivity of drift)
**Depends on:** Experiment P2-02 (the trained models and their capacity-matched configuration)

---

> **RQ2.4 —** Does the incremental update mechanism produce longitudinally stable state trajectories
> under repeated near-identical inputs, and how sensitive is that stability to the update mechanism's
> hyperparameters?

Section 4.4 of the manuscript is explicit that this is a **design-property evaluation**, not a headline
accuracy claim: its significance derives from the requirements of downstream generative inference, and
the KT literature has not measured this property at all. This notebook keeps that framing — it measures
a property of the representation, and it does not dress a low drift number up as a predictive result.


## 0. Four things to read before running

### 0.1 "Drift" and "recovery" are not standard metrics — they must be defined before they are measured

There is no off-the-shelf "state drift per interaction". A number is only meaningful once the metric,
the input regime, and the normalisation are pinned down, so this notebook fixes all three *before*
looking at any model:

- **Drift per interaction** — under a stream of near-identical inputs (same skill, same correctness,
  small behavioural jitter), a *stable* representation should converge to an approximately fixed point,
  not wander. Drift is the mean step-to-step displacement of the state **once past a warm-up prefix**,
  measured in a **scale-free** way so that a model with a larger-radius state is not penalised for using
  more of its space. Concretely, per step $t$ in the repeated regime,
  $\mathrm{drift} = \operatorname{median}_t
  \dfrac{\lVert h_t - h_{t-1}\rVert_2}{\tfrac12(\lVert h_t\rVert_2 + \lVert h_{t-1}\rVert_2) + \varepsilon}$.
  Ratio-based, so it is comparable across a GRU hidden state, an attention read, and the PCDT state even
  though their raw norms differ by construction.

- **Recovery time** — inject a single atypical interaction into an otherwise-repeated stream, then count
  how many subsequent steps the state takes to return within a tolerance band of the pre-perturbation
  fixed point (in the same scale-free distance). If it never returns within the window, recovery is
  reported as censored ($\ge$ window length), never as a fabricated finite number.

The stability **target** is pre-registered (`drift_target`), so "meets stability target?" is a
falsifiable column and not a verdict written after seeing the numbers.

### 0.2 The three models do not have comparable state, and pretending they do is the trap

Table 6 compares "hidden states" across a recurrent KT, an attentive KT, and the PCDT. But a GRU hidden
state, the PCDT's GRU-cell state, and an attention model's *per-step context read* are different objects:
the first two are genuine recurrent states that carry across steps; an attention model has **no single
recurrent state**, only a re-derived context vector at each position. Measuring "drift" of an attention
read and calling it the same quantity as GRU-state drift would be comparing unlike things.

This notebook therefore (i) measures each model's **most state-like internal signal**, named explicitly
per model; (ii) reports the attentive model's *context-read* drift with a footnote that it is not a
recurrent state; and (iii) adds a **prediction-level drift** row computed identically for every model —
the step-to-step change in the model's *output probability* under the repeated regime — because that one
IS comparable across all three and is arguably what actually matters for a stable substrate. Read the
prediction-level rows as the comparable ones; read the state-level rows within each model.

### 0.3 Table 7's axes name hyperparameters the audited model does not have

Manuscript Table 7 asks for drift as a function of *learning rate*, *prior strength*, and *consolidation
rate*. The Experiment P2-02 model has a learning rate and a persistence time-scale $\tau_0$ (with a
learned per-dimension decay $\lambda$), but it has **no "prior strength" and no "consolidation rate"** —
those terms come from the manuscript's pre-audit draft, the same way three rows of Table 3 named
components the audit had removed. Inventing a prior-strength knob to fill a pre-drafted cell would invert
the relationship between the model and the table.

The notebook sweeps the hyperparameters that **actually govern this model's state dynamics** —
`learning rate`, `tau0` (persistence time-scale), and `dropout` (the nearest real analogue to an update
regulariser) — and it prints, in the table's caption, that "prior strength" and "consolidation rate" are
not parameters of the audited model and are reported as `n/a`. If you adopt a belief-augmented PCDT
later (the $\Sigma_t$ module from the P2-02 v2 notebook), `belief_beta` becomes the genuine "prior
strength" axis and can be swept then — but that is a different model and would require regenerating the
other tables.

### 0.4 This measures the model as trained, not a re-initialised one

Drift is a property of the *learned* dynamics, so the sweep in Table 7 retrains a small PCDT at each
hyperparameter setting on a modest budget and measures the resulting state's drift. Table 6 uses the
capacity-matched configuration from Experiment P2-02. Where a full trained checkpoint is unavailable in
the session, the notebook trains a compact model under the identical harness rather than measuring an
untrained network, whose drift would be meaningless.

## 2. Configuration and Pre-Registered Decisions

The split and capacity parameters match Experiment P2-02 so that the Table 6 models are the audited
configuration. The stability-specific decisions — warm-up length, repeated-regime length, tolerance
band, and the drift target itself — are fixed here, before any state is inspected.


```python
@dataclass(frozen=True)
class Config:
    # ---------- data / split: match Experiment P2-02 ----------
    max_learners: int = 20_000
    min_seq_len: int = 10
    max_seq_len: int = 200
    train_frac: float = 0.70
    val_frac: float = 0.10

    # ---------- capacity / model (match Experiment P2-02) ----------
    param_budget: int = 350_000
    param_tolerance: float = 0.05
    pcdt_dimensions: Sequence[str] = ("C_t", "B_t")
    use_persistence: bool = True
    tau0_seconds: float = 3600.0
    dropout: float = 0.2
    batch_size: int = 64
    weight_decay: float = 1e-5
    grad_clip: float = 5.0

    # ---------- training budget for stability models ----------
    # Table 6 uses a full-ish budget; Table 7's sweep uses a reduced one (many cells).
    table6_epochs: int = 15
    table6_patience: int = 3
    table6_lr: float = 1e-3
    sweep_epochs: int = 6
    sweep_patience: int = 2
    sweep_max_learners: int = 6_000       # subsample for the sweep only

    # ---------- stability probe (pre-registered) ----------
    warmup_steps: int = 20                # prefix before the repeated regime begins
    repeat_steps: int = 60               # length of the near-identical-input regime
    jitter_sd: float = 0.02              # behavioural-feature jitter in the "near-identical" stream
    probe_batch: int = 256               # independent probe sequences (different warm-up histories)
    recovery_window: int = 40            # max steps to wait for recovery before censoring
    recovery_tol_mult: float = 1.5       # recovered when displacement <= mult * steady-state drift
    eps: float = 1e-8

    # ---------- pre-registered stability target ----------
    drift_target: float = 0.02           # scale-free drift per interaction judged "stable"

    # ---------- Table 7 sweep grids (only real parameters of THIS model) ----------
    sweep_lrs: Sequence[float] = (5e-4, 1e-3, 2e-3)
    sweep_tau0: Sequence[float] = (600.0, 3600.0, 86400.0)     # 10 min, 1 h, 1 day
    sweep_dropout: Sequence[float] = (0.0, 0.2, 0.4)
    # manuscript-named axes that DO NOT EXIST in the audited model -> reported n/a
    absent_axes: Sequence[str] = ("prior strength", "consolidation rate")

    # ---------- caching ----------
    cache_dir: Optional[str] = "PCDT_P2_cache"

CFG = Config()
if CFG.cache_dir:
    try: os.makedirs(CFG.cache_dir, exist_ok=True)
    except OSError as e: print(f"cache dir unavailable ({e}); caching disabled")

print("Pre-registered before any state is inspected:")
print(f"  warm-up / repeat / recovery window : {CFG.warmup_steps} / {CFG.repeat_steps} / {CFG.recovery_window} steps")
print(f"  drift target (scale-free)          : {CFG.drift_target}")
print(f"  Table 7 real axes                  : lr {list(CFG.sweep_lrs)}, "
      f"tau0 {list(CFG.sweep_tau0)}, dropout {list(CFG.sweep_dropout)}")
print(f"  Table 7 manuscript axes not in model: {list(CFG.absent_axes)} -> reported n/a")
```

## 3. Data

Stability is measured with **synthetic probe sequences** (Section 4), so real interaction data is needed
only to *train* the models whose stability is then probed. The primary loader is the Experiment
P2-02 ASSISTments 2020–2021 pipeline; if its files are absent, a fixture stands in for pipeline
verification and every output is flagged as non-manuscript. The probe itself is model-agnostic and does
not depend on which dataset trained the model.


```python
def _parse_skills(v):
    if isinstance(v, list): return v
    if isinstance(v, str) and v.strip().startswith("["):
        try: return ast.literal_eval(v)
        except (ValueError, SyntaxError): return []
    return [] if pd.isna(v) else [v]

DATA_DIR   = "dataset"
PLOGS_PATH = f"{DATA_DIR}/feb_to_apr/plogs.csv"; PDETS_PATH = f"{DATA_DIR}/feb_to_apr/pdets.csv"
BEHAV_COLS = ["response_time", "hint_frac", "attempt_count"]

def make_fixture(n_learners=1500, rng=None):
    rng = rng or np.random.default_rng(0)
    n_problems, skills = 400, [f"CCSS.{i}" for i in range(30)]
    prob_skill = {p: skills[int(rng.integers(0, len(skills)))] for p in range(n_problems)}
    skill_diff = {s: rng.normal(0, .8) for s in skills}
    rows, t0 = [], pd.Timestamp("2021-02-01", tz="UTC")
    for lid in range(n_learners):
        ability = rng.normal(0, 1); mastery = {s: rng.normal(0, .3) for s in skills}
        t = t0 + pd.Timedelta(days=int(rng.integers(0, 20)))
        for _ in range(int(rng.integers(3, 12))):
            t += pd.Timedelta(days=int(rng.integers(1, 14)))
            for _ in range(int(rng.integers(5, 30))):
                pid = int(rng.integers(0, n_problems)); sk = prob_skill[pid]
                p = 1/(1+np.exp(-(ability + mastery[sk] - skill_diff[sk])))
                corr = int(rng.random() < np.clip(p, .02, .98)); mastery[sk] += .12 if corr else .04
                rows.append(dict(learner_id=lid, item_id=pid, skill_id=sk, start_time=t,
                                 correct=float(corr),
                                 response_time=float(np.clip(rng.lognormal(3, .4), 2, 900)),
                                 hint_frac=float(np.clip(rng.beta(2,5), 0, 1)),
                                 attempt_count=float(1+rng.poisson(0.5))))
                t += pd.Timedelta(seconds=int(rng.integers(20, 240)))
    return pd.DataFrame(rows)

def load_data():
    if not os.path.exists(PLOGS_PATH):
        print("!"*76); print("! ASSISTments not found -> FIXTURE (pipeline verification only).")
        print("!"*76)
        return make_fixture(rng=RNG), True
    plogs = pd.read_csv(PLOGS_PATH, usecols=["student_id","problem_id","start_time","time_on_task",
                                             "attempt_count","fraction_of_hints_used","correct"])
    pdets = pd.read_csv(PDETS_PATH, usecols=["problem_id","skills"]); pdets["skills"]=pdets["skills"].map(_parse_skills)
    df = plogs
    df["correct"] = pd.to_numeric(df["correct"].map({True:1.0,False:0.0,"True":1.0,"False":0.0}).fillna(df["correct"]), errors="coerce")
    for c in ("time_on_task","attempt_count","fraction_of_hints_used"): df[c]=pd.to_numeric(df[c], errors="coerce")
    df = df.dropna(subset=["correct"]); df["start_time"]=pd.to_datetime(df["start_time"], format="ISO8601", utc=True)
    prim = pdets.copy(); prim["skill_id"]=prim["skills"].map(lambda s: s[0] if isinstance(s,list) and len(s) else np.nan)
    prim = prim.dropna(subset=["skill_id"])[["problem_id","skill_id"]]
    df = df.merge(prim, on="problem_id", how="inner").rename(columns={
        "student_id":"learner_id","problem_id":"item_id","time_on_task":"response_time",
        "fraction_of_hints_used":"hint_frac"})
    return df[["learner_id","item_id","skill_id","start_time","correct",*BEHAV_COLS[1:], "response_time"]].copy(), False

def build_packs(df, cfg, rng, max_learners=None):
    max_learners = max_learners or cfg.max_learners
    df = df.sort_values(["learner_id","start_time"]).copy()
    counts = df.groupby("learner_id").size(); keep = counts[counts>=cfg.min_seq_len].index
    if len(keep) > max_learners: keep = rng.choice(np.asarray(keep), max_learners, replace=False)
    df = df[df.learner_id.isin(set(keep))].copy()
    learners = np.sort(df.learner_id.unique()); rng.shuffle(learners)
    n=len(learners); n_tr,n_va=int(cfg.train_frac*n),int(cfg.val_frac*n)
    so={}
    for l in learners[:n_tr]: so[l]="train"
    for l in learners[n_tr:n_tr+n_va]: so[l]="val"
    for l in learners[n_tr+n_va:]: so[l]="test"
    df["split"]=df.learner_id.map(so)
    tsk=df.loc[df.split=="train","skill_id"].unique(); s2i={s:i+1 for i,s in enumerate(sorted(tsk,key=str))}
    df["skill_idx"]=df.skill_id.map(s2i).fillna(0).astype(int); n_skills=len(s2i)+1
    med=df[df.split=="train"][BEHAV_COLS].median(); df[BEHAV_COLS]=df[BEHAV_COLS].fillna(med)
    df["response_time"]=np.log1p(df["response_time"].clip(lower=0))
    tr=df[df.split=="train"]; mu,sd=tr[BEHAV_COLS].mean(),tr[BEHAV_COLS].std().replace(0,1.0)
    df[BEHAV_COLS]=(df[BEHAV_COLS]-mu)/sd
    df["dt_sec"]=df.groupby("learner_id")["start_time"].diff().dt.total_seconds().fillna(0).clip(lower=0)
    packs={}
    for sp in ("train","val","test"):
        sub=df[df.split==sp]; gids=sub.learner_id.unique(); L=cfg.max_seq_len
        S=np.zeros((len(gids),L),np.int64);A=np.zeros((len(gids),L),np.float32)
        Fb=np.zeros((len(gids),L,len(BEHAV_COLS)),np.float32);DT=np.zeros((len(gids),L),np.float32);M=np.zeros((len(gids),L),np.float32)
        for i,l in enumerate(gids):
            g=sub[sub.learner_id==l].tail(L);k=len(g)
            S[i,:k]=g.skill_idx.values;A[i,:k]=g.correct.values;Fb[i,:k]=g[BEHAV_COLS].values;DT[i,:k]=g.dt_sec.values;M[i,:k]=1.0
        packs[sp]=dict(skill=S,correct=A,behav=Fb,dt=DT,mask=M,learner_id=gids)
    return packs,n_skills,s2i

DF, USING_FIXTURE = load_data()
PACKS, N_SKILLS, SKILL2IDX = build_packs(DF, CFG, RNG)
for sp in ("train","val","test"):
    print(f"  {sp:5s}: {len(PACKS[sp]['learner_id']):,} learners | {int(PACKS[sp]['mask'].sum()):,} interactions")
print(f"  skills: {N_SKILLS:,}")
```

## 4. Models with State-Trajectory Access

The Experiment P2-02 models return logits only. Stability needs the *internal trajectory*, so each model
here exposes a `state_trace(batch)` method returning, per step, both its most state-like internal signal
and its output probability. The names are explicit so the table can attribute each drift number to a
concrete object:

- **Recurrent KT (DKT)** — `state` is the LSTM hidden state $h_t$ (a genuine recurrent state).
- **Attentive KT (AKT)** — `state` is the per-step **context read** $c_t$. This is **not** a recurrent
  state; it is re-derived at each position. Its drift is reported with that caveat.
- **PCDT** — `state` is the GRU-cell state $h_t$ over the audited dimensions, after $\Psi$.

Every model also returns `prob_t`, and the prediction-level drift computed from it is the quantity that
is genuinely comparable across all three.


```python
class DKT(nn.Module):
    def __init__(self, n_skills, hidden, dropout=0.2):
        super().__init__()
        self.inter_emb = nn.Embedding(2*n_skills+1, hidden, padding_idx=0)
        self.lstm = nn.LSTM(hidden, hidden, batch_first=True)
        self.drop = nn.Dropout(dropout); self.out = nn.Linear(hidden, n_skills); self.n_skills=n_skills
    def forward(self, b):
        s,a=b["skill"],b["correct"]
        inter=(s+a.long()*self.n_skills).clamp(max=self.inter_emb.num_embeddings-1)
        h,_=self.lstm(self.inter_emb(inter[:,:-1])); logits=self.out(self.drop(h))
        return logits.gather(-1, s[:,1:].unsqueeze(-1)).squeeze(-1)
    @torch.no_grad()
    def state_trace(self, b):
        s,a=b["skill"],b["correct"]
        inter=(s+a.long()*self.n_skills).clamp(max=self.inter_emb.num_embeddings-1)
        h,_=self.lstm(self.inter_emb(inter))                        # [B,L,H] hidden at every step
        logits=self.out(h); p=torch.sigmoid(logits.gather(-1, s.unsqueeze(-1)).squeeze(-1))
        return h, p                                                 # state, prob


class AKT(nn.Module):
    def __init__(self, n_skills, hidden, n_heads=4, dropout=0.2, max_len=200):
        super().__init__()
        self.c_emb=nn.Embedding(n_skills,hidden,padding_idx=0); self.d_emb=nn.Embedding(n_skills,hidden,padding_idx=0)
        self.diff=nn.Embedding(n_skills,1,padding_idx=0); self.resp_emb=nn.Embedding(3,hidden,padding_idx=2)
        self.q_proj,self.k_proj,self.v_proj=(nn.Linear(hidden,hidden) for _ in range(3))
        self.theta=nn.Parameter(torch.tensor(0.5)); self.ln=nn.LayerNorm(hidden)
        self.ffn=nn.Sequential(nn.Linear(2*hidden,hidden),nn.ReLU(),nn.Dropout(dropout),nn.Linear(hidden,1))
        self.drop=nn.Dropout(dropout); self.scale=hidden**0.5
    def _rasch(self,s): return self.c_emb(s)+self.diff(s)*self.d_emb(s)
    def _ctx(self, s, a):
        L=s.shape[1]-1
        x_past=self._rasch(s[:,:-1])+self.resp_emb(a[:,:-1].long())
        q=self.q_proj(self._rasch(s[:,1:])); k,v=self.k_proj(x_past),self.v_proj(x_past)
        scores=(q@k.transpose(-2,-1))/self.scale
        pos=torch.arange(L,device=s.device); dist=(pos.view(-1,1)-pos.view(1,-1)).float().clamp(min=0)
        scores=scores-F.softplus(self.theta)*dist
        causal=torch.triu(torch.ones(L,L,device=s.device,dtype=torch.bool),diagonal=1)
        scores=scores.masked_fill(causal,float("-inf"))
        ctx=self.drop(torch.softmax(scores,dim=-1))@v
        return ctx, q
    def forward(self, b):
        s,a=b["skill"],b["correct"]; ctx,q=self._ctx(s,a)
        return self.ffn(torch.cat([self.ln(ctx),q],dim=-1)).squeeze(-1)
    @torch.no_grad()
    def state_trace(self, b):
        s,a=b["skill"],b["correct"]; ctx,q=self._ctx(s,a)          # ctx: [B,L-1,H] context read
        p=torch.sigmoid(self.ffn(torch.cat([self.ln(ctx),q],dim=-1)).squeeze(-1))
        # pad to L for alignment; the context read is defined for predictions 1..L-1
        B,Lm1,H=ctx.shape
        state=torch.cat([torch.zeros(B,1,H,device=ctx.device), ctx],dim=1)
        prob=torch.cat([torch.zeros(B,1,device=p.device), p],dim=1)
        return state, prob


class PCDT(nn.Module):
    def __init__(self, n_skills, hidden, dims=("C_t","B_t"), use_persistence=True,
                 n_behav=3, tau0=3600.0, dropout=0.2):
        super().__init__()
        self.dims,self.use_persistence,self.tau0=list(dims),use_persistence,tau0
        self.n_dims=len(self.dims); self.dim_size=max(8,hidden//self.n_dims); self.state_size=self.dim_size*self.n_dims
        self.inter_emb=nn.Embedding(2*n_skills+1,hidden,padding_idx=0); self.ex_emb=nn.Embedding(n_skills,hidden,padding_idx=0)
        self.behav_proj=nn.Linear(n_behav,hidden) if "B_t" in self.dims else None
        in_dim=hidden+(hidden if self.behav_proj is not None else 0)
        self.cell=nn.GRUCell(in_dim,self.state_size); self.log_lambda=nn.Parameter(torch.full((self.n_dims,),-1.0))
        self.drop=nn.Dropout(dropout)
        self.head=nn.Sequential(nn.Linear(self.state_size+hidden,hidden),nn.ReLU(),nn.Dropout(dropout),nn.Linear(hidden,1))
        self.n_skills=n_skills
    def psi(self,h,dt):
        if not self.use_persistence: return h
        lam=F.softplus(self.log_lambda); dts=torch.log1p(dt.clamp(min=0)/self.tau0).unsqueeze(-1)
        return h*torch.exp(-lam.unsqueeze(0)*dts).repeat_interleave(self.dim_size,dim=-1)
    def _run(self, b, keep_state):
        s,a,f,dt=b["skill"],b["correct"],b["behav"],b["dt"]; B,L=s.shape
        inter=(s+a.long()*self.n_skills).clamp(max=self.inter_emb.num_embeddings-1)
        h=torch.zeros(B,self.state_size,device=s.device); outs=[]; states=[]
        rng_end = L if keep_state else L-1
        for t in range(rng_end):
            x=self.inter_emb(inter[:,t])
            if self.behav_proj is not None: x=torch.cat([x,self.behav_proj(f[:,t])],dim=-1)
            h=self.psi(h,dt[:,t]); h=self.cell(x,h)
            if keep_state: states.append(h.unsqueeze(1))
            nxt = s[:, t+1] if t < L-1 else s[:, t]
            outs.append(self.head(torch.cat([self.drop(h),self.ex_emb(nxt)],dim=-1)))
        logits=torch.cat(outs,dim=1)
        if keep_state: return torch.cat(states,dim=1), torch.sigmoid(logits)
        return logits
    def forward(self,b): return self._run(b, keep_state=False)
    @torch.no_grad()
    def state_trace(self,b): return self._run(b, keep_state=True)


def count_params(m): return sum(p.numel() for p in m.parameters() if p.requires_grad)
def match_capacity(builder, budget, tol, lo=8, hi=512):
    best,bg=None,float("inf")
    while lo<=hi:
        mid=(lo+hi)//2
        try: n=count_params(builder(mid))
        except Exception: hi=mid-1; continue
        g=abs(n-budget)
        if g<bg: best,bg=(mid,n),g
        if n<budget: lo=mid+1
        else: hi=mid-1
    h,n=best; return h,n,abs(n-budget)/budget<=tol

BUILDERS = {
    "Recurrent KT (DKT)": lambda h: DKT(N_SKILLS, h, dropout=CFG.dropout),
    "Attentive KT (AKT)": lambda h: AKT(N_SKILLS, h, dropout=CFG.dropout, max_len=CFG.max_seq_len),
    "PCDT":               lambda h: PCDT(N_SKILLS, h, dims=CFG.pcdt_dimensions,
                                         use_persistence=CFG.use_persistence, n_behav=len(BEHAV_COLS),
                                         tau0=CFG.tau0_seconds, dropout=CFG.dropout),
}
CAPACITY={}
print(f"Capacity matching to {CFG.param_budget:,} (+/- {CFG.param_tolerance:.0%})\n")
for name,b in BUILDERS.items():
    h,n,ok=match_capacity(b,CFG.param_budget,CFG.param_tolerance)
    CAPACITY[name]=dict(hidden=h,n_params=n,within_tol=ok)
    print(f"  {name:22s} hidden={h:4d}  params={n:9,}  {'OK' if ok else 'OUTSIDE TOL'}")
```

```python
class _DS(Dataset):
    def __init__(self,p): self.p=p
    def __len__(self): return self.p["skill"].shape[0]
    def __getitem__(self,i):
        return dict(skill=torch.from_numpy(self.p["skill"][i]),correct=torch.from_numpy(self.p["correct"][i]),
                    behav=torch.from_numpy(self.p["behav"][i]),dt=torch.from_numpy(self.p["dt"][i]),
                    mask=torch.from_numpy(self.p["mask"][i]),idx=i)

def masked_bce(logits,targets,mask):
    loss=F.binary_cross_entropy_with_logits(logits,targets,reduction="none")
    return (loss*mask).sum()/mask.sum().clamp(min=1.0)

def train(builder, hidden, packs, epochs, patience, lr, device):
    seed_everything(SEED); model=builder(hidden).to(device)
    opt=torch.optim.AdamW(model.parameters(),lr=lr,weight_decay=CFG.weight_decay)
    tr=DataLoader(_DS(packs["train"]),batch_size=CFG.batch_size,shuffle=True)
    va=DataLoader(_DS(packs["val"]),batch_size=CFG.batch_size,shuffle=False)
    best,bstate,bad=-np.inf,None,0
    for ep in range(epochs):
        model.train()
        for b in tr:
            b={k:(v.to(device) if torch.is_tensor(v) else v) for k,v in b.items()}
            loss=masked_bce(model(b),b["correct"][:,1:],b["mask"][:,1:])
            opt.zero_grad(); loss.backward(); nn.utils.clip_grad_norm_(model.parameters(),CFG.grad_clip); opt.step()
        model.eval(); P=[]; T=[]
        with torch.no_grad():
            for b in va:
                b={k:(v.to(device) if torch.is_tensor(v) else v) for k,v in b.items()}
                p=torch.sigmoid(model(b)); m=b["mask"][:,1:]>0
                P.append(p[m].cpu().numpy()); T.append(b["correct"][:,1:][m].cpu().numpy())
        auc=roc_auc_score(np.concatenate(T),np.concatenate(P))
        if auc>best+1e-5: best,bad,bstate=auc,0,{k:v.detach().cpu().clone() for k,v in model.state_dict().items()}
        else:
            bad+=1
            if bad>=patience: break
    if bstate is not None: model.load_state_dict(bstate)
    return model, best

print("Harness + models ready.")
```

## 5. The Stability Probe

Each probe sequence has three phases: a **warm-up prefix** drawn from a real learner history (so the
state is in a realistic region rather than at its zero initialisation), a **repeated regime** of
near-identical inputs (a single fixed skill, fixed correctness, small behavioural jitter of SD
`jitter_sd`, and near-zero inter-step gaps), and — for the recovery probe only — a **single injected
perturbation** (a different skill with flipped correctness) partway through the repeated regime.

`probe_batch` independent sequences are built with different warm-up histories, so drift is averaged
over many starting states rather than read off one trajectory. The same probe tensor is fed to every
model, so the three models are compared on identical inputs.


```python
def build_probe(packs, cfg, rng, with_perturbation=False):
    """Warm-up prefix (real history) + repeated near-identical regime [+ one perturbation]."""
    test = packs["test"]; n_avail = test["skill"].shape[0]
    idx = rng.choice(n_avail, size=min(cfg.probe_batch, n_avail), replace=False)
    B = len(idx); Wm = cfg.warmup_steps; Rp = cfg.repeat_steps; L = Wm + Rp
    S = np.zeros((B, L), np.int64); A = np.zeros((B, L), np.float32)
    Fb = np.zeros((B, L, len(BEHAV_COLS)), np.float32); DT = np.zeros((B, L), np.float32)
    M = np.ones((B, L), np.float32)

    for bi, si in enumerate(idx):
        valid = np.where(test["mask"][si] > 0)[0]
        w = valid[:Wm] if len(valid) >= Wm else valid
        k = len(w)
        S[bi, :k]  = test["skill"][si][w]; A[bi, :k] = test["correct"][si][w]
        Fb[bi, :k] = test["behav"][si][w]; DT[bi, :k] = test["dt"][si][w]
        if k < Wm:                              # pad short warm-ups by repeating the last real step
            S[bi, k:Wm] = S[bi, k-1] if k else 1
            Fb[bi, k:Wm] = Fb[bi, k-1] if k else 0.0

    # repeated regime: one fixed skill per sequence, fixed correctness, small jitter, ~0 gap
    rep_skill = rng.integers(1, max(2, N_SKILLS), size=B)
    rep_corr  = rng.integers(0, 2, size=B).astype(np.float32)
    base_feat = rng.normal(0, 1, size=(B, len(BEHAV_COLS))).astype(np.float32)
    for bi in range(B):
        for t in range(Wm, L):
            S[bi, t] = rep_skill[bi]; A[bi, t] = rep_corr[bi]
            Fb[bi, t] = base_feat[bi] + rng.normal(0, cfg.jitter_sd, size=len(BEHAV_COLS)).astype(np.float32)
            DT[bi, t] = rng.uniform(30, 90)     # seconds; within-session
    perturb_at = None
    if with_perturbation:
        perturb_at = Wm + Rp // 2
        alt_skill = ((rep_skill % max(1, N_SKILLS-1)) + 1).astype(np.int64)
        for bi in range(B):
            S[bi, perturb_at] = alt_skill[bi]
            A[bi, perturb_at] = 1.0 - rep_corr[bi]                 # flipped correctness
            Fb[bi, perturb_at] = base_feat[bi] + rng.normal(0, 5*cfg.jitter_sd, size=len(BEHAV_COLS)).astype(np.float32)
    return dict(skill=S, correct=A, behav=Fb, dt=DT, mask=M), Wm, perturb_at

print("Probe builder ready.")
```

## 6. Drift and Recovery Metrics

Both metrics use the scale-free step distance of Section 0.1,
$\delta_t = \lVert x_t - x_{t-1}\rVert_2 / (\tfrac12(\lVert x_t\rVert + \lVert x_{t-1}\rVert) +
\varepsilon)$, applied to whatever trajectory $x$ is passed — a model's state trace, or its probability
trace (a scalar, where the norm is $|\cdot|$).

- **Drift per interaction** is the median of $\delta_t$ over the repeated regime (excluding the warm-up),
  averaged across probe sequences. Median, not mean, so a single jitter spike does not dominate.
- **Recovery time** perturbs at a known step and reports the number of steps until $\delta_t$ falls back
  within `recovery_tol_mult` $\times$ the pre-perturbation steady-state drift and stays there; censored at
  `recovery_window` if it never does.


```python
def _scalefree_steps(x):
    """x: [B, T, D] or [B, T]. Returns per-step scale-free distance [B, T-1]."""
    if x.ndim == 2: x = x[:, :, None]
    diff = np.linalg.norm(x[:, 1:] - x[:, :-1], axis=-1)
    scale = 0.5 * (np.linalg.norm(x[:, 1:], axis=-1) + np.linalg.norm(x[:, :-1], axis=-1)) + CFG.eps
    return diff / scale

def drift_per_interaction(trace, warmup):
    """Median scale-free step in the repeated regime, averaged over sequences."""
    steps = _scalefree_steps(trace)                 # [B, T-1]; index t is step from t to t+1
    reg = steps[:, warmup:]                         # repeated-regime steps only
    per_seq = np.median(reg, axis=1)
    return float(np.mean(per_seq)), float(np.std(per_seq))

def recovery_time(trace, warmup, perturb_at, tol_mult, window):
    """Steps to return within tol_mult * steady-state drift after the perturbation. Censored at window."""
    steps = _scalefree_steps(trace)                 # step index t == displacement from t -> t+1
    pre = steps[:, warmup:perturb_at-1]
    steady = np.median(pre, axis=1) if pre.shape[1] > 0 else np.median(steps[:, warmup:], axis=1)
    tol = tol_mult * np.maximum(steady, CFG.eps)
    B = trace.shape[0]; rec = np.full(B, window, np.float32); censored = np.zeros(B, bool)
    for bi in range(B):
        got = window; ok = False
        for s in range(perturb_at, min(perturb_at + window, steps.shape[1])):
            if steps[bi, s] <= tol[bi]:
                got = s - perturb_at + 1; ok = True; break
        rec[bi] = got; censored[bi] = not ok
    return float(np.median(rec)), float(np.mean(censored))

@torch.no_grad()
def trace_of(model, probe, device):
    ds = _DS({**probe, "mask": probe["mask"]})
    loader = DataLoader(ds, batch_size=CFG.batch_size, shuffle=False)
    states, probs = [], []
    model.eval()
    for b in loader:
        b = {k:(v.to(device) if torch.is_tensor(v) else v) for k,v in b.items()}
        st, pr = model.state_trace(b)
        states.append(st.cpu().numpy()); probs.append(pr.cpu().numpy())
    return np.concatenate(states, 0), np.concatenate(probs, 0)

print("Metric functions ready.")
```

```python
# --- train (or reuse) each model at the capacity-matched config, then probe ---
MODELS6 = ["Recurrent KT (DKT)", "Attentive KT (AKT)", "PCDT"]
TRAINED = {}
print("Training models for Table 6 (capacity-matched config)\n")
for name in MODELS6:
    m, auc = train(BUILDERS[name], CAPACITY[name]["hidden"], PACKS,
                   CFG.table6_epochs, CFG.table6_patience, CFG.table6_lr, DEVICE)
    TRAINED[name] = m
    print(f"  {name:22s} val AUC={auc:.4f}")

seed_everything(SEED)
PROBE_DRIFT, _wm, _ = build_probe(PACKS, CFG, RNG, with_perturbation=False)
PROBE_REC,  _wm2, _pat = build_probe(PACKS, CFG, RNG, with_perturbation=True)

ATTN_MODELS = {"Attentive KT (AKT)"}     # state is a context read, not a recurrent state
rows6, ext6 = [], []
for name in MODELS6:
    st_d, pr_d = trace_of(TRAINED[name], PROBE_DRIFT, DEVICE)
    st_r, pr_r = trace_of(TRAINED[name], PROBE_REC, DEVICE)
    sd_state, _ = drift_per_interaction(st_d, _wm)
    sd_prob,  _ = drift_per_interaction(pr_d, _wm)
    rec_state, cens_s = recovery_time(st_r, _wm2, _pat, CFG.recovery_tol_mult, CFG.recovery_window)
    rec_prob,  cens_p = recovery_time(pr_r, _wm2, _pat, CFG.recovery_tol_mult, CFG.recovery_window)
    meets = "yes" if sd_state <= CFG.drift_target else "no"
    note = " (context read, not a recurrent state)" if name in ATTN_MODELS else ""
    rec_disp = f">= {int(rec_state)} (censored)" if cens_s > 0.5 else f"{rec_state:.0f}"
    rows6.append({"Model": name + note,
                  "Drift per interaction": f"{sd_state:.4f}",
                  "Recovery time (steps)": rec_disp,
                  "Meets stability target?": meets})
    ext6.append({"Model": name, "State drift": f"{sd_state:.4f}", "Prob drift": f"{sd_prob:.4f}",
                 "State recovery": rec_disp, "Prob recovery": f"{rec_prob:.0f}",
                 "% censored (state)": f"{100*cens_s:.0f}%", "target": f"{CFG.drift_target}",
                 "meets": meets})

table6 = pd.DataFrame(rows6); table6_ext = pd.DataFrame(ext6)
print("\nTable 6. State drift under repeated near-identical inputs, and comparison to baseline "
      "hidden-state trajectories.\n")
print(table6.to_markdown(index=False))
print("\nExtended (state-level vs prediction-level; prediction-level is the comparable one):\n")
print(table6_ext.to_markdown(index=False))
table6.to_csv("table6.csv", index=False); table6_ext.to_csv("table6_extended.csv", index=False)
with open("table6.md","w") as fh: fh.write(table6.to_markdown(index=False))

t6_notes = [
 f"Drift = median over the {CFG.repeat_steps}-step repeated regime of the scale-free step distance "
 f"||x_t - x_(t-1)|| / (0.5(||x_t||+||x_(t-1)||)+eps), averaged over {PROBE_DRIFT['skill'].shape[0]} probe "
 f"sequences with distinct real warm-up histories. Scale-free so states of different radius are comparable.",
 "The three 'states' are different objects: DKT and PCDT states are recurrent (carried across steps); "
 "the AKT 'state' is a per-step context read that is re-derived each position, NOT a recurrent state. "
 "Its row is marked accordingly and its drift is not strictly comparable to the recurrent states.",
 "The extended table adds prediction-level drift (step-to-step change in output probability), computed "
 "identically for every model. THIS is the quantity comparable across all three; the manuscript should "
 "report it alongside the state-level row or in place of it.",
 f"Recovery time: steps after a single injected atypical interaction (different skill, flipped "
 f"correctness at step {_pat}) until the scale-free step returns within {CFG.recovery_tol_mult}x the "
 f"pre-perturbation steady-state drift; censored at {CFG.recovery_window} steps and reported as "
 f"'>= N (censored)' rather than a fabricated finite value.",
 f"Stability target {CFG.drift_target} is pre-registered; 'meets target?' compares state drift to it. "
 "A design-property target (Section 4.4), instrumentally valuable for downstream inference, not a "
 "predictive-performance claim.",
]
print("\nCaption notes:")
for i,n in enumerate(t6_notes,1): print(f"  ({i}) {n}")
with open("table6_caption_notes.txt","w") as fh: fh.write("\n".join(f"({i}) {n}" for i,n in enumerate(t6_notes,1)))
```

## 7. Table 7 — Hyperparameter Sensitivity of Drift

Table 7 asks how drift depends on the update mechanism's hyperparameters. As Section 0.3 sets out, the
manuscript's named axes include two — *prior strength* and *consolidation rate* — that the audited model
does not have. This notebook sweeps the parameters that genuinely govern the PCDT's state dynamics
(**learning rate**, **persistence time-scale $\tau_0$**, **dropout**), reports drift for each, and lists
the absent axes as `n/a` in the caption rather than inventing knobs to fill them.

A small PCDT is retrained at each cell on the reduced sweep budget, then probed with the same stability
probe. The result is a long-format table (one row per configuration) plus a per-axis marginal summary;
render it as a heat map over any two axes if the manuscript wants the grid form.


```python
SWEEP_PACKS, SWEEP_NS, _ = build_packs(DF, CFG, np.random.default_rng(SEED),
                                       max_learners=CFG.sweep_max_learners)
# keep the probe consistent with the sweep's skill vocabulary
def probe_for(ns):
    global N_SKILLS
    _old = N_SKILLS; N_SKILLS = ns
    seed_everything(SEED)
    pr, wm, _ = build_probe(SWEEP_PACKS, CFG, np.random.default_rng(SEED), with_perturbation=False)
    N_SKILLS = _old
    return pr, wm

grid = list(itertools.product(CFG.sweep_lrs, CFG.sweep_tau0, CFG.sweep_dropout))
print(f"Table 7 sweep: {len(grid)} configurations x (train {CFG.sweep_epochs} epochs + probe)\n")

def build_sweep_pcdt(h, tau0, dropout):
    return PCDT(SWEEP_NS, h, dims=CFG.pcdt_dimensions, use_persistence=CFG.use_persistence,
                n_behav=len(BEHAV_COLS), tau0=tau0, dropout=dropout)

sweep_rows = []
pr_probe, pr_wm = probe_for(SWEEP_NS)
h_sweep, _, _ = match_capacity(lambda h: build_sweep_pcdt(h, CFG.tau0_seconds, CFG.dropout),
                               CFG.param_budget, CFG.param_tolerance)
t0 = time.time()
for i, (lr, tau0, dropout) in enumerate(grid):
    builder = lambda h, tt=tau0, dd=dropout: build_sweep_pcdt(h, tt, dd)
    model, auc = train(builder, h_sweep, SWEEP_PACKS, CFG.sweep_epochs, CFG.sweep_patience, lr, DEVICE)
    st, pr = trace_of(model, pr_probe, DEVICE)
    d_state, _ = drift_per_interaction(st, pr_wm)
    d_prob, _  = drift_per_interaction(pr, pr_wm)
    sweep_rows.append(dict(lr=lr, tau0=tau0, dropout=dropout, val_auc=auc,
                           state_drift=d_state, prob_drift=d_prob,
                           meets=("yes" if d_state <= CFG.drift_target else "no")))
    print(f"  [{i+1:2d}/{len(grid)}] lr={lr:<6g} tau0={tau0:<7g} drop={dropout:<3g} "
          f"-> drift={d_state:.4f} (AUC={auc:.3f})")
print(f"\nsweep runtime: {(time.time()-t0)/60:.1f} min")

sweep = pd.DataFrame(sweep_rows)
sweep.to_csv("table7_full.csv", index=False)

print("\nTable 7 (full grid; drift = scale-free state drift per interaction):\n")
show = sweep.copy()
show["state_drift"] = show["state_drift"].map(lambda v: f"{v:.4f}")
show["prob_drift"]  = show["prob_drift"].map(lambda v: f"{v:.4f}")
show["val_auc"]     = show["val_auc"].map(lambda v: f"{v:.3f}")
print(show.to_markdown(index=False))

print("\nPer-axis marginal (mean state drift, holding one axis):\n")
for ax in ["lr", "tau0", "dropout"]:
    marg = sweep.groupby(ax)["state_drift"].agg(["mean", "std"]).reset_index()
    marg["mean"] = marg["mean"].map(lambda v: f"{v:.4f}"); marg["std"] = marg["std"].map(lambda v: f"{v:.4f}")
    print(f"  by {ax}:")
    print("    " + marg.to_string(index=False).replace("\n", "\n    "))
    print()

t7_notes = [
 "Axes swept are the parameters that actually govern this model's state dynamics: learning rate, "
 "persistence time-scale tau0, and dropout (the nearest real update-regularisation analogue).",
 f"The manuscript-named axes {list(CFG.absent_axes)} are NOT parameters of the audited {{C_t,B_t}}+Psi "
 "model and are reported as n/a. 'Prior strength' would become meaningful only under the belief-augmented "
 "PCDT (the Sigma_t module), which is a different model and would require regenerating the other tables.",
 "Each cell retrains a compact PCDT on a reduced budget (sweep_epochs) and a subsample; absolute drift "
 "may differ slightly from Table 6's fully-trained model, but the SENSITIVITY pattern across axes is the "
 "reportable object, not the absolute level.",
 "Render as a heat map over any two axes (e.g. lr x tau0) for the manuscript's grid form; the CSV "
 "table7_full.csv has every cell.",
]
print("Caption notes:")
for i,n in enumerate(t7_notes,1): print(f"  ({i}) {n}")
with open("table7_caption_notes.txt","w") as fh: fh.write("\n".join(f"({i}) {n}" for i,n in enumerate(t7_notes,1)))

# optional heat map (lr x tau0, averaged over dropout)
try:
    import matplotlib.pyplot as plt
    piv = sweep.groupby(["lr","tau0"])["state_drift"].mean().unstack()
    fig, ax = plt.subplots(figsize=(6,4))
    im = ax.imshow(piv.values, aspect="auto", cmap="viridis")
    ax.set_xticks(range(len(piv.columns))); ax.set_xticklabels([f"{c:g}" for c in piv.columns])
    ax.set_yticks(range(len(piv.index)));   ax.set_yticklabels([f"{r:g}" for r in piv.index])
    ax.set_xlabel("tau0 (s)"); ax.set_ylabel("learning rate"); ax.set_title("State drift (lower = more stable)")
    for (yy,xx),v in np.ndenumerate(piv.values): ax.text(xx,yy,f"{v:.3f}",ha="center",va="center",color="w",fontsize=8)
    fig.colorbar(im, label="scale-free drift"); plt.tight_layout(); plt.savefig("table7_heatmap.png", dpi=120); plt.show()
    print("\nsaved: table7_heatmap.png")
except Exception as e:
    print(f"\n(heat map skipped: {e})")
```

## 8. Interpretation

> Written before the run. Read the tables for the numbers.

**Table 6.** If the PCDT state drifts *less* than the recurrent baseline under repeated inputs, the claim
is a design-property claim (Section 4.4): the representation is a more stable substrate for the
generative inference of Paper 3a. It is instrumentally, not intrinsically, valuable, and the manuscript
should say so rather than presenting low drift as if it were a predictive win. If drift is comparable to
the baselines, that is still worth reporting — the KT literature has not measured this property at all,
so a null is a contribution to knowledge about these models, not a failure.

**Read the prediction-level rows as the comparable ones.** The state-level comparison across a GRU state,
a context read, and the PCDT state mixes unlike objects; the step-to-step change in output probability is
computed identically for all three and is what a downstream consumer of the representation actually sees.

**Table 7.** The reportable object is the *sensitivity pattern*, not the absolute drift level at any one
cell. If drift is flat across `tau0` and `lr`, the mechanism is robust and that robustness is the finding.
If drift explodes at high learning rate or short `tau0`, that bounds the usable hyperparameter region and
should be stated as a constraint on the model rather than hidden.

**Do not let Table 7 quietly acquire the manuscript's phantom axes.** "Prior strength" and "consolidation
rate" are not in this model. Reporting them as `n/a` with a one-line reason is honest; fabricating a knob
to produce a number for them is the same error as filling Table 3's rejected-component rows.
