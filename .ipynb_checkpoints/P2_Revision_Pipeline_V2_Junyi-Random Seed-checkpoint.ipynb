# Paper 2 — Revision Pipeline (single reproducible notebook)

This notebook regenerates **every** result reported in the revised manuscript. It is the response to the
referee report and replaces the twelve per-experiment notebooks as the canonical pipeline.

**Run it once per dataset.** Set `DATASET` in Section 3, then Run All. Every table and figure is written
to `OUTPUT_DIR` and every number in the manuscript must be traceable to those files.

## What is new in this revision

| Ref | Referee concern | New experiment in this notebook |
|---|---|---|
| **M1** | Persistence tested only against *absence* of persistence | **§11** — DKT-F (forgetting-aware) added, and SAKT / AKT / DKVMN **restored**, all capacity-matched and run through the RQ2.3 protocol |
| **M2** | Real evidence stops at one week | **§7 + §14** time-aware splitting with an enlarged test partition, explicitly attempting real 4-week cells |
| **M3** | EdNet's 1.2-day median span unexamined | **§6 Exploratory temporal statistics** — full duration distribution, quartiles, % reaching 1/2/4/8 weeks |
| **M5** | Educational contribution asserted, not shown | **§15** — threshold sweep, flagged / missed / false alarms, gap-stratified, **against both the control and DKT-F** |
| **D** | Simulator under-described but load-bearing | **§16 Simulator validation** — calibration, parameter sensitivity, failure cases |
| **E** | Statistical rigour | **§13** bootstrap CIs, paired tests, effect sizes, Holm correction throughout |

## Honest scope note

Two things this notebook **cannot** do, because the datasets do not contain the required information:

1. **No independent outcome validation.** None of the three datasets contains a post-test, delayed
   assessment, or any outcome external to the interaction stream. Construct validity here is internal to
   that stream. §24 records this as irreducible.
2. **No demographic or subgroup analysis.** Demographic attributes are deliberately excluded (see the
   Ethics statement); Junyi is the only dataset carrying any, and using them would contradict that
   declaration. Subgroup analysis is therefore reported as not performed, by design rather than omission.

## 5. Data validation

Hard gates. Each raises rather than warns, because every one of these conditions silently corrupted a
result at some point in the previous round.

```python
def validate(df):
    assert set(STD) <= set(df.columns), f"missing columns: {set(STD)-set(df.columns)}"
    assert df["correct"].isin([0.0, 1.0]).all(), "correct must be binary"
    n_kc = df["skill_id"].nunique()
    assert n_kc >= 5, (f"DEGENERATE KC SPACE: {n_kc} distinct skills. Per-KC mastery collapses to a "
                       "per-learner average and C_t validity becomes ~1.0 by construction. Fix the KC mapping.")
    span = (df.groupby("learner_id")["start_time"].agg(lambda s:(s.max()-s.min()).total_seconds())/86400)
    assert df["start_time"].notna().all(), "unparsed timestamps"
    print(f"  OK  {df['learner_id'].nunique():,} learners | {len(df):,} interactions | {n_kc:,} KCs")
    print(f"      correct {df['correct'].mean():.3f} | median learner span {span.median():.1f} d")
    if span.median() < 7:
        print(f"      NOTE median span {span.median():.1f} d is short; Section 6 quantifies the "
              "long-gap subpopulation explicitly (referee concern M3).")
    return True

validate(DF)
```

## 6. Exploratory temporal statistics — referee concern M3

Answers directly: how long do learners actually stay, and what fraction reach each horizon? This is the
evidence the referee asked for regarding EdNet's short median span.

```python
def temporal_profile(df):
    g = df.sort_values(["learner_id","start_time"]).groupby("learner_id")["start_time"]
    span_d = (g.agg(lambda s:(s.max()-s.min()).total_seconds())/86400).rename("span_days")
    gaps = df.sort_values(["learner_id","start_time"]).groupby("learner_id")["start_time"].diff().dt.total_seconds()
    maxgap_d = (gaps.groupby(df.sort_values(["learner_id","start_time"])["learner_id"]).max()/86400)
    q = span_d.describe(percentiles=[.25,.5,.75,.9,.95])
    rows = [dict(statistic=k, value=float(v)) for k,v in q.items()]
    for wk in (1,2,4,8):
        rows.append(dict(statistic=f"% learners with span >= {wk} week(s)",
                         value=float((span_d >= 7*wk).mean()*100)))
        rows.append(dict(statistic=f"% learners with a GAP >= {wk} week(s)",
                         value=float((maxgap_d >= 7*wk).mean()*100)))
    t = pd.DataFrame(rows)
    t.to_csv(f"{TAB_DIR}/tableA_temporal_profile.csv", index=False)
    print(t.round(3).to_string(index=False))

    fig, ax = plt.subplots(1,2, figsize=(7.2,2.6))
    ax[0].hist(np.log10(span_d.clip(lower=1/24)+1e-9), bins=50, color="#0072B2")
    ax[0].set_xlabel("log10 learner span (days)"); ax[0].set_ylabel("learners")
    ax[1].hist(np.log10(maxgap_d.dropna().clip(lower=1/1440)+1e-9), bins=50, color="#D55E00")
    ax[1].set_xlabel("log10 largest inter-session gap (days)")
    for a in ax: a.grid(alpha=.25)
    fig.suptitle(f"Temporal profile — {CFG.dataset}", fontsize=9)
    fig.tight_layout(); fig.savefig(f"{FIG_DIR}/figA_temporal_profile.pdf")
    fig.savefig(f"{FIG_DIR}/figA_temporal_profile.png", dpi=300); plt.close(fig)
    return span_d, maxgap_d

SPAN_D, MAXGAP_D = temporal_profile(DF)
```

## 7. Train / validation / test construction — referee concern M2

Two changes from the original design, both aimed at populating the four-week cells with **real** data:

1. the test partition is enlarged to 40% of learners;
2. learners possessing a long gap are guaranteed representation in the test split by stratifying the
   learner-level assignment on each learner's largest gap.

Stratifying on gap length does not leak outcome information — the gap is a property of when a learner
appeared, not of whether they answered correctly — but it is recorded here as a deviation from a simple
random split so that a reader can judge it.

```python
def build_packs(df, cfg, rng, maxgap_d=None):
    df = df.sort_values(["learner_id","start_time"]).copy()
    keep = df.groupby("learner_id").size()
    keep = keep[keep >= cfg.min_seq_len].index
    df = df[df.learner_id.isin(set(keep))].copy()
    learners = np.sort(df.learner_id.unique())

    if cfg.split_mode == "learner" and maxgap_d is not None:
        # [new] stratify on the learner's largest gap so long-gap learners reach the test split
        gap = maxgap_d.reindex(learners).fillna(0).values
        strat = np.digitize(gap, [1, 7, 28])                # 0:<1d 1:1-7d 2:7-28d 3:>=28d
        split_of = {}
        for s in np.unique(strat):
            ids = learners[strat == s]; rng.shuffle(ids)
            n = len(ids); ntr = int(cfg.train_frac*n); nva = int(cfg.val_frac*n)
            for l in ids[:ntr]:            split_of[l] = "train"
            for l in ids[ntr:ntr+nva]:     split_of[l] = "val"
            for l in ids[ntr+nva:]:        split_of[l] = "test"
    else:
        rng.shuffle(learners)
        n=len(learners); ntr=int(cfg.train_frac*n); nva=int(cfg.val_frac*n)
        split_of = {**{l:"train" for l in learners[:ntr]},
                    **{l:"val" for l in learners[ntr:ntr+nva]},
                    **{l:"test" for l in learners[ntr+nva:]}}
    df["split"] = df.learner_id.map(split_of)

    tsk = df.loc[df.split=="train","skill_id"].unique()
    s2i = {s:i+1 for i,s in enumerate(sorted(tsk, key=str))}
    df["skill_idx"] = df.skill_id.map(s2i).fillna(0).astype(int)
    n_skills = len(s2i)+1

    med = df[df.split=="train"][BEHAV].median(); df[BEHAV] = df[BEHAV].fillna(med)
    df["response_time"] = np.log1p(df["response_time"].clip(lower=0))
    tr = df[df.split=="train"]; mu,sd = tr[BEHAV].mean(), tr[BEHAV].std().replace(0,1.0)
    df[BEHAV] = (df[BEHAV]-mu)/sd
    df[BEHAV] = df[BEHAV].fillna(0.0)
    df["dt_sec"] = df.groupby("learner_id")["start_time"].diff().dt.total_seconds().fillna(0).clip(lower=0)

    packs = {}
    for sp in ("train","val","test"):
        sub = df[df.split==sp]; gids = sub.learner_id.unique(); L = cfg.max_seq_len
        S=np.zeros((len(gids),L),np.int64); A=np.zeros((len(gids),L),np.float32)
        F=np.zeros((len(gids),L,len(BEHAV)),np.float32); D=np.zeros((len(gids),L),np.float32)
        M=np.zeros((len(gids),L),np.float32)
        for i,l in enumerate(gids):
            g=sub[sub.learner_id==l].tail(L); k=len(g)
            S[i,:k]=g.skill_idx.values; A[i,:k]=g.correct.values
            F[i,:k]=g[BEHAV].values; D[i,:k]=g.dt_sec.values; M[i,:k]=1.0
        packs[sp]=dict(skill=S,correct=A,behav=F,dt=D,mask=M,learner_id=gids)
        print(f"  {sp:5s} {len(gids):,} learners | {int(M.sum()):,} interactions")
    return packs, n_skills

PACKS, N_SKILLS = build_packs(DF, CFG, np.random.default_rng(SEED), MAXGAP_D)

# gap bucket for every prediction position (position t predicts t+1; the gap crossed is dt[t+1])
BUCKETS = [("within",0,20*60),("sameday",20*60,86400),("d1",86400,7*86400),
           ("d7",7*86400,28*86400),("d28",28*86400,np.inf)]
def bucket_of(pack):
    dtp = np.zeros_like(pack["dt"]); dtp[:,:-1] = pack["dt"][:,1:]
    out = np.full(dtp.shape, -1, np.int64)
    for i,(k,lo,hi) in enumerate(BUCKETS): out[(dtp>=lo)&(dtp<hi)] = i
    return out
TEST_BUCKET = bucket_of(PACKS["test"])
m = PACKS["test"]["mask"][:,1:] > 0
print("\nREAL prediction points per horizon in the enlarged test split:")
for i,(k,lo,hi) in enumerate(BUCKETS):
    sel = (TEST_BUCKET[:,:-1]==i) & m
    print(f"  {k:8s} points={int(sel.sum()):>8,}  learners={np.unique(np.where(sel)[0]).size:>6,}")
```

## 8–10. Models: shared harness, KT baselines, and the persistent representation

```python
import torch.nn as nn, torch.nn.functional as Fn
import torch.nn.functional as F   # the restored SAKT/AKT/DKVMN classes reference F
from torch.utils.data import Dataset, DataLoader

class PackDS(Dataset):
    def __init__(self,p): self.p=p
    def __len__(self): return self.p["skill"].shape[0]
    def __getitem__(self,i):
        return dict(skill=torch.from_numpy(self.p["skill"][i]), correct=torch.from_numpy(self.p["correct"][i]),
                    behav=torch.from_numpy(self.p["behav"][i]), dt=torch.from_numpy(self.p["dt"][i]),
                    mask=torch.from_numpy(self.p["mask"][i]), idx=i)

class DKT(nn.Module):
    def __init__(self, n_skills, hidden, dropout=0.2, use_time=False):
        super().__init__(); self.n_skills=n_skills; self.use_time=use_time
        self.emb=nn.Embedding(2*n_skills+1, hidden, padding_idx=0)
        extra = 2 if use_time else 0                      # [new] DKT-F time features
        self.lstm=nn.LSTM(hidden+extra, hidden, batch_first=True)
        self.drop=nn.Dropout(dropout); self.out=nn.Linear(hidden, n_skills)
    def forward(self,b):
        s,a,dt=b["skill"],b["correct"],b["dt"]
        x=self.emb((s+a.long()*self.n_skills).clamp(max=self.emb.num_embeddings-1)[:,:-1])
        if self.use_time:
            # repetition-gap and sequence-gap features, log-scaled (forgetting-aware input encoding)
            g=torch.log1p(dt[:,:-1]/60.0).unsqueeze(-1)
            gn=torch.log1p(dt[:,1:]/60.0).unsqueeze(-1)
            x=torch.cat([x,g,gn],dim=-1)
        h,_=self.lstm(x)
        return self.out(self.drop(h)).gather(-1, s[:,1:].unsqueeze(-1)).squeeze(-1)

class SAKT(nn.Module):
    """Self-attentive KT: query = target exercise, key/value = past interactions, causal mask."""
    def __init__(self, n_skills, hidden, n_heads=4, dropout=0.2, max_len=200):
        super().__init__()
        # nn.MultiheadAttention requires embed_dim % num_heads == 0. Without this rounding the
        # capacity search hits invalid widths, the constructor raises, and the search collapses
        # downward -- silently leaving SAKT far below the parameter budget.
        hidden = max(n_heads, (hidden // n_heads) * n_heads)
        self.hidden = hidden
        self.inter_emb = nn.Embedding(2*n_skills + 1, hidden, padding_idx=0)
        self.ex_emb    = nn.Embedding(n_skills, hidden, padding_idx=0)
        self.pos_emb   = nn.Embedding(max_len, hidden)
        self.attn = nn.MultiheadAttention(hidden, n_heads, dropout=dropout, batch_first=True)
        self.ln1, self.ln2 = nn.LayerNorm(hidden), nn.LayerNorm(hidden)
        self.ffn  = nn.Sequential(nn.Linear(hidden, hidden*2), nn.ReLU(),
                                  nn.Dropout(dropout), nn.Linear(hidden*2, hidden))
        self.drop = nn.Dropout(dropout)
        self.out  = nn.Linear(hidden, 1)
        self.n_skills = n_skills

    def forward(self, b):
        s, a = b["skill"], b["correct"]
        inter = (s + a.long()*self.n_skills).clamp(max=self.inter_emb.num_embeddings-1)
        L = s.shape[1] - 1
        pos = torch.arange(L, device=s.device).unsqueeze(0)
        k = self.inter_emb(inter[:, :-1]) + self.pos_emb(pos)      # past interactions
        q = self.ex_emb(s[:, 1:])                                  # target exercises
        causal = torch.triu(torch.ones(L, L, device=s.device, dtype=torch.bool), diagonal=1)
        att, _ = self.attn(q, k, k, attn_mask=causal)
        h = self.ln1(q + self.drop(att))
        h = self.ln2(h + self.drop(self.ffn(h)))
        return self.out(h).squeeze(-1)

class AKT(nn.Module):
    """Attentive KT: Rasch-style embeddings + monotonic distance-decayed attention."""
    def __init__(self, n_skills, hidden, n_heads=4, dropout=0.2, max_len=200):
        super().__init__()
        self.c_emb    = nn.Embedding(n_skills, hidden, padding_idx=0)   # concept
        self.d_emb    = nn.Embedding(n_skills, hidden, padding_idx=0)   # concept variation
        self.diff     = nn.Embedding(n_skills, 1, padding_idx=0)        # scalar item difficulty
        self.resp_emb = nn.Embedding(3, hidden, padding_idx=2)
        self.q_proj, self.k_proj, self.v_proj = (nn.Linear(hidden, hidden) for _ in range(3))
        self.theta = nn.Parameter(torch.tensor(0.5))                    # learned decay
        self.ln    = nn.LayerNorm(hidden)
        self.ffn   = nn.Sequential(nn.Linear(2*hidden, hidden), nn.ReLU(),
                                   nn.Dropout(dropout), nn.Linear(hidden, 1))
        self.drop  = nn.Dropout(dropout)
        self.scale = hidden ** 0.5

    def _rasch(self, s):
        return self.c_emb(s) + self.diff(s) * self.d_emb(s)

    def forward(self, b):
        s, a = b["skill"], b["correct"]
        L = s.shape[1] - 1
        x_past = self._rasch(s[:, :-1]) + self.resp_emb(a[:, :-1].long())
        q = self.q_proj(self._rasch(s[:, 1:]))
        k, v = self.k_proj(x_past), self.v_proj(x_past)

        scores = (q @ k.transpose(-2, -1)) / self.scale
        pos = torch.arange(L, device=s.device)
        dist = (pos.view(-1,1) - pos.view(1,-1)).float().clamp(min=0)   # temporal distance
        scores = scores - F.softplus(self.theta) * dist                 # monotonic decay
        causal = torch.triu(torch.ones(L, L, device=s.device, dtype=torch.bool), diagonal=1)
        scores = scores.masked_fill(causal, float("-inf"))
        ctx = self.drop(torch.softmax(scores, dim=-1)) @ v
        return self.ffn(torch.cat([self.ln(ctx), q], dim=-1)).squeeze(-1)

class DKVMN(nn.Module):
    """Memory-augmented KT: static key memory + dynamic value memory with erase-add writes."""
    def __init__(self, n_skills, hidden, n_concepts=32, dropout=0.2):
        super().__init__()
        self.N, self.dk, self.dv = n_concepts, hidden, hidden
        self.k_emb = nn.Embedding(n_skills, self.dk, padding_idx=0)
        self.v_emb = nn.Embedding(2*n_skills + 1, self.dv, padding_idx=0)
        self.Mk    = nn.Parameter(torch.randn(self.N, self.dk) * 0.1)
        self.Mv0   = nn.Parameter(torch.randn(self.N, self.dv) * 0.1)
        self.erase = nn.Linear(self.dv, self.dv)
        self.add   = nn.Linear(self.dv, self.dv)
        self.ffn   = nn.Sequential(nn.Linear(self.dv + self.dk, hidden), nn.Tanh(),
                                   nn.Dropout(dropout), nn.Linear(hidden, 1))
        self.n_skills = n_skills

    def forward(self, b):
        s, a = b["skill"], b["correct"]
        B, L = s.shape
        inter = (s + a.long()*self.n_skills).clamp(max=self.v_emb.num_embeddings-1)
        Mv = self.Mv0.unsqueeze(0).expand(B, -1, -1).contiguous()
        outs = []
        for t in range(L - 1):
            kq = self.k_emb(s[:, t+1])                              # query = NEXT item
            w  = torch.softmax(kq @ self.Mk.t(), dim=-1)            # [B, N]
            r  = torch.einsum("bn,bnd->bd", w, Mv)                  # read
            outs.append(self.ffn(torch.cat([r, kq], dim=-1)))
            vt = self.v_emb(inter[:, t])                            # write PAST interaction
            e, ad = torch.sigmoid(self.erase(vt)), torch.tanh(self.add(vt))
            Mv = Mv * (1 - w.unsqueeze(-1)*e.unsqueeze(1)) + w.unsqueeze(-1)*ad.unsqueeze(1)
        return torch.cat(outs, dim=1)


class PCDT(nn.Module):
    """Persistent representation. reset_sessions=True gives the session-resetting control."""
    def __init__(self, n_skills, hidden, n_behav=3, tau0=3600., dropout=0.2,
                 use_persistence=True, reset_sessions=False, session_gap_s=1200.):
        super().__init__()
        self.n_skills=n_skills; self.tau0=tau0; self.use_persistence=use_persistence
        self.reset_sessions=reset_sessions; self.session_gap_s=session_gap_s
        self.state=hidden
        self.emb=nn.Embedding(2*n_skills+1,hidden,padding_idx=0)
        self.ex=nn.Embedding(n_skills,hidden,padding_idx=0)
        self.bproj=nn.Linear(n_behav,hidden)
        self.cell=nn.GRUCell(2*hidden,self.state)
        self.log_lambda=nn.Parameter(torch.tensor(-1.0))
        self.drop=nn.Dropout(dropout)
        self.head=nn.Sequential(nn.Linear(self.state+hidden,hidden),nn.ReLU(),nn.Dropout(dropout),nn.Linear(hidden,1))
    def psi(self,h,dt):
        if not self.use_persistence: return h
        lam=Fn.softplus(self.log_lambda)
        return h*torch.exp(-lam*torch.log1p(dt.clamp(min=0)/self.tau0)).unsqueeze(-1)
    def forward(self,b):
        s,a,f,dt=b["skill"],b["correct"],b["behav"],b["dt"]; B,L=s.shape
        inter=(s+a.long()*self.n_skills).clamp(max=self.emb.num_embeddings-1)
        h=torch.zeros(B,self.state,device=s.device); outs=[]
        for t in range(L-1):
            x=torch.cat([self.emb(inter[:,t]), self.bproj(f[:,t])],dim=-1)
            if self.reset_sessions:
                h=h*(1.0-(dt[:,t]>=self.session_gap_s).float().unsqueeze(-1))
            else:
                h=self.psi(h,dt[:,t])
            h=self.cell(x,h)
            outs.append(self.head(torch.cat([self.drop(h), self.ex(s[:,t+1])],dim=-1)))
        return torch.cat(outs,dim=1)

def count_params(m): return sum(p.numel() for p in m.parameters() if p.requires_grad)
def match_capacity(builder, budget=CFG.param_budget, tol=CFG.param_tolerance, lo=8, hi=512):
    best=None; bg=1e18
    while lo<=hi:
        mid=(lo+hi)//2
        try: n=count_params(builder(mid))
        except Exception: hi=mid-1; continue
        if abs(n-budget)<bg: best,bg=(mid,n),abs(n-budget)
        if n<budget: lo=mid+1
        else: hi=mid-1
    return best[0], best[1], abs(best[1]-budget)/budget<=tol

def masked_bce(logits,y,m):
    l=Fn.binary_cross_entropy_with_logits(logits,y,reduction="none")
    return (l*m).sum()/m.sum().clamp(min=1.)

def train(builder, hidden, lr, packs=PACKS, epochs=CFG.max_epochs, patience=CFG.patience, seed=SEED):
    seed_everything(seed); model=builder(hidden).to(DEVICE)
    opt=torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=CFG.weight_decay)
    tr=DataLoader(PackDS(packs["train"]),batch_size=CFG.batch_size,shuffle=True)
    va=DataLoader(PackDS(packs["val"]),batch_size=CFG.batch_size,shuffle=False)
    best=-1e9; bstate=None; bad=0
    for ep in range(epochs):
        model.train()
        for b in tr:
            b={k:(v.to(DEVICE) if torch.is_tensor(v) else v) for k,v in b.items()}
            loss=masked_bce(model(b), b["correct"][:,1:], b["mask"][:,1:])
            opt.zero_grad(); loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(),CFG.grad_clip); opt.step()
        P,T=[],[]; model.eval()
        with torch.no_grad():
            for b in va:
                b={k:(v.to(DEVICE) if torch.is_tensor(v) else v) for k,v in b.items()}
                mm=b["mask"][:,1:]>0
                P.append(torch.sigmoid(model(b))[mm].cpu().numpy()); T.append(b["correct"][:,1:][mm].cpu().numpy())
        auc=roc_auc_score(np.concatenate(T),np.concatenate(P))
        if auc>best+1e-5: best,bad,bstate=auc,0,{k:v.detach().cpu().clone() for k,v in model.state_dict().items()}
        else:
            bad+=1
            if bad>=patience: break
    model.load_state_dict(bstate); return model, best

@torch.no_grad()
def predict(model, pack):
    model.eval(); out=np.full((pack["skill"].shape[0],pack["skill"].shape[1]-1),np.nan,np.float32)
    for b in DataLoader(PackDS(pack),batch_size=CFG.batch_size,shuffle=False):
        idx=b["idx"].numpy(); b={k:(v.to(DEVICE) if torch.is_tensor(v) else v) for k,v in b.items()}
        out[idx]=torch.sigmoid(model(b)).cpu().numpy()
    return out
print("models + harness ready")
```

## 11. Forgetting-aware baselines — referee concern M1 (the key addition)

The original submission compared the persistence operator only against a session-resetting control, i.e.
against the *absence* of persistence. A referee will reasonably ask whether Ψ retains better than models
that already encode forgetting. This section adds two such comparators under the identical capacity match
and tuning grid:

- **DKT-F** — a forgetting-aware recurrent tracer that receives log-scaled elapsed-time features
  (repetition gap and sequence gap) alongside the interaction embedding.
- **PCDT − Ψ + time-features** — the persistent architecture with Ψ disabled but time supplied as input,
  isolating whether the benefit comes from the *operator* or merely from the model seeing elapsed time.

If Ψ shows no advantage over these, that is a publishable and important negative result and the
manuscript's persistence claim must be weakened accordingly. Section 20 writes whichever outcome occurs.

```python
MODELS = {}
def build_all():
    specs = {
      # ---- original baseline family, RESTORED so the within-session comparison is against the
      # ---- true strongest tuned baseline (AKT was strongest on Junyi in the original submission)
      "DKT":        lambda h: DKT(N_SKILLS,h,use_time=False),
      "SAKT":       lambda h: SAKT(N_SKILLS,h,max_len=CFG.max_seq_len),
      "AKT":        lambda h: AKT(N_SKILLS,h),
      "DKVMN":      lambda h: DKVMN(N_SKILLS,h),
      # ---- forgetting-aware comparator (referee concern M1)
      "DKT-F":      lambda h: DKT(N_SKILLS,h,use_time=True),
      # ---- the representation under test, its ablation, and the no-persistence control
      "PCDT":       lambda h: PCDT(N_SKILLS,h,n_behav=len(BEHAV),tau0=CFG.tau0_seconds),
      "PCDT-noPsi": lambda h: PCDT(N_SKILLS,h,n_behav=len(BEHAV),use_persistence=False),
      "Control":    lambda h: PCDT(N_SKILLS,h,n_behav=len(BEHAV),reset_sessions=True,
                                   use_persistence=False,session_gap_s=CFG.session_gap_s),
    }
    for name,b in specs.items():
        hh,np_,ok = match_capacity(b)
        best=(-1,None)
        for lr in CFG.search_lrs:
            mdl,auc = train(b,hh,lr)
            if auc>best[0]: best=(auc,lr,mdl)
        MODELS[name] = dict(model=best[2], lr=best[1], val_auc=best[0], hidden=hh, params=np_, matched=ok)
        print(f"  {name:12s} hidden={hh:4d} params={np_:8,} lr={best[1]:<7g} val AUC={best[0]:.4f}")
build_all()

PRED = {k: predict(v["model"], PACKS["test"]) for k,v in MODELS.items()}
np.savez_compressed(f"{OUTPUT_DIR}/predictions_test.npz",
                    **{k:v for k,v in PRED.items()}, targets=PACKS["test"]["correct"][:,1:],
                    mask=PACKS["test"]["mask"][:,1:], bucket=TEST_BUCKET[:,:-1])
print("saved predictions_test.npz")
```

## 12–13. Evaluation and statistical validation

```python
def ece(p,y,bins=CFG.ece_bins):
    e=0.; edges=np.linspace(0,1,bins+1); idx=np.clip(np.digitize(p,edges)-1,0,bins-1)
    for b in range(bins):
        s=idx==b
        if s.sum(): e += s.mean()*abs(y[s].mean()-p[s].mean())
    return e

def evaluate(p,y):
    fin=np.isfinite(p)&np.isfinite(y); p,y=p[fin],y[fin]
    pc=np.clip(p,1e-7,1-1e-7)
    return dict(n=len(y), auc=roc_auc_score(y,p), acc=((p>=.5)==y).mean(),
                rmse=float(np.sqrt(((p-y)**2).mean())),
                nll=float(-(y*np.log(pc)+(1-y)*np.log(1-pc)).mean()),
                ece=float(ece(p,y)), brier=float(brier_score_loss(y,p)))

M=PACKS["test"]["mask"][:,1:]>0; Y=PACKS["test"]["correct"][:,1:]
rows=[]
for k,p in PRED.items():
    r=evaluate(p[M],Y[M]); r["model"]=k; r["params"]=MODELS[k]["params"]; rows.append(r)
T2=pd.DataFrame(rows)[["model","params","n","auc","acc","rmse","nll","ece","brier"]]
T2.to_csv(f"{TAB_DIR}/table2_predictive.csv", index=False)

# [new] identify the STRONGEST TUNED BASELINE explicitly -- H1 must be tested against this, not
# against whichever baseline happens to be in the model set.
BASELINES = ["DKT","SAKT","AKT","DKVMN","DKT-F"]
_b = T2[T2.model.isin(BASELINES)].sort_values("auc", ascending=False)
STRONGEST = _b.iloc[0]["model"]
print(f"\nStrongest tuned baseline on {CFG.dataset}: {STRONGEST} (AUC {_b.iloc[0]['auc']:.4f})")
print(f"PCDT - {STRONGEST} = {T2[T2.model=='PCDT'].auc.iloc[0]-_b.iloc[0]['auc']:+.4f} "
      f"(pre-registered meaningfulness floor {CFG.min_meaningful_auc_delta})")
print(T2.round(4).to_string(index=False))

def paired_bootstrap(pa,pb,y,rows_idx,n=CFG.n_bootstrap,seed=SEED):
    rg=np.random.default_rng(seed); learners=np.unique(rows_idx)
    by={l:np.where(rows_idx==l)[0] for l in learners}; d=[]
    for _ in range(n):
        s=rg.choice(learners,len(learners),replace=True)
        ii=np.concatenate([by[l] for l in s])
        if len(np.unique(y[ii]))<2: continue
        d.append(roc_auc_score(y[ii],pa[ii])-roc_auc_score(y[ii],pb[ii]))
    d=np.array(d); lo,hi=np.percentile(d,[2.5,97.5]); pv=2*min((d<=0).mean(),(d>=0).mean())
    return float(roc_auc_score(y,pa)-roc_auc_score(y,pb)), float(lo), float(hi), float(min(pv,1))

rows_idx=np.repeat(np.arange(PRED["PCDT"].shape[0])[:,None],PRED["PCDT"].shape[1],1)[M]
stats=[]
for k in PRED:
    if k=="PCDT": continue
    d,lo,hi,pv = paired_bootstrap(PRED["PCDT"][M],PRED[k][M],Y[M],rows_idx)
    stats.append(dict(comparison=f"PCDT - {k}", delta_auc=d, ci_lo=lo, ci_hi=hi, p=pv,
                      meaningful=abs(d)>=CFG.min_meaningful_auc_delta))
S=pd.DataFrame(stats)
pv=S["p"].values; order=np.argsort(pv); adj=np.empty(len(pv)); run=0
for rank,i in enumerate(order):
    run=max(run,(len(pv)-rank)*pv[i]); adj[i]=min(run,1.)
S["p_holm"]=adj
S.to_csv(f"{TAB_DIR}/table3_statistics.csv", index=False)
print("\n", S.round(4).to_string(index=False))
```

## 14. Temporal persistence analysis — real four-week cells if the data supports them

Reports each horizon on **real** data where the pre-registered adequacy floor is met, and marks it
`INADEQUATE` otherwise. No horizon is silently substituted by simulation.

```python
def did_retention(pa, pb, bucket_id, ref_id=0):
    Mm=PACKS["test"]["mask"][:,1:]>0; B=TEST_BUCKET[:,:-1]
    selh=(B==bucket_id)&Mm; selr=(B==ref_id)&Mm
    nl=np.unique(np.where(selh)[0]).size; npt=int(selh.sum())
    if nl<CFG.min_learners_bucket or npt<CFG.min_points_bucket:
        return dict(adequate=False, n_learners=nl, n_points=npt)
    def auc(sel,p): 
        y=Y[sel]; return roc_auc_score(y,p[sel]) if len(np.unique(y))>1 else np.nan
    ra=(auc(selh,pa)-auc(selh,pb))-(auc(selr,pa)-auc(selr,pb))
    rg=np.random.default_rng(SEED); ls=np.unique(np.where(selh|selr)[0]); boot=[]
    for _ in range(CFG.n_bootstrap):
        s=set(rg.choice(ls,len(ls),replace=True).tolist())
        mh=selh.copy(); mr=selr.copy()
        keep=np.zeros(selh.shape[0],bool); keep[list(s)]=True
        mh&=keep[:,None]; mr&=keep[:,None]
        if mh.sum()<50 or mr.sum()<50: continue
        try: boot.append((auc(mh,pa)-auc(mh,pb))-(auc(mr,pa)-auc(mr,pb)))
        except Exception: pass
    boot=np.array([b for b in boot if np.isfinite(b)])
    lo,hi=(np.percentile(boot,[2.5,97.5]) if len(boot) else (np.nan,np.nan))
    return dict(adequate=True, n_learners=nl, n_points=npt, retention_advantage=float(ra),
                ci_lo=float(lo), ci_hi=float(hi),
                auc_pcdt=float(auc(selh,pa)), auc_ctrl=float(auc(selh,pb)))

rows=[]
_seen=set()
for comparator in ["Control","DKT-F",STRONGEST]:   # control, forgetting-aware, and strongest baseline
    for bid,(k,lo,hi) in enumerate(BUCKETS):
        if k in ("within","sameday"): continue
        r=did_retention(PRED["PCDT"][:,:], PRED[comparator][:,:], bid)
        if (k,comparator) in _seen: continue
        _seen.add((k,comparator))
        r.update(horizon=k, comparator=comparator, source="real" if r["adequate"] else "inadequate")
        rows.append(r)
T5=pd.DataFrame(rows)
T5.to_csv(f"{TAB_DIR}/table5_persistence.csv", index=False)
print(T5.round(4).to_string(index=False))
print("\nAny horizon marked INADEQUATE is NOT reported as a real result and must not be "
      "replaced by a simulated value in the manuscript without an explicit (sim) tag.")
```

## 15. Educational utility — referee concern M5

Converts the retention advantage into the decision a teacher makes: *flag this learner for review*.
Reported per horizon so the educational claim is tied to the gap at which it was measured.

```python
def utility(p,y,thr):
    flag=(p<thr).astype(int); need=(y==0).astype(int)
    tn,fp,fn,tp=confusion_matrix(need,flag,labels=[0,1]).ravel()
    prec=tp/(tp+fp) if tp+fp else np.nan; rec=tp/(tp+fn) if tp+fn else np.nan
    spec=tn/(tn+fp) if tn+fp else np.nan
    f1=2*prec*rec/(prec+rec) if (prec and rec and prec+rec) else np.nan
    return dict(threshold=thr, flagged=int(tp+fp), correctly_flagged=int(tp), false_alarms=int(fp),
                missed=int(fn), precision=prec, recall=rec, specificity=spec, f1=f1,
                balanced_accuracy=(rec+spec)/2)

Mm=PACKS["test"]["mask"][:,1:]>0; B=TEST_BUCKET[:,:-1]
rows=[]
# [new] compare against BOTH the no-persistence control and the strongest TIME-AWARE model, so the
# educational claim is not merely "better than a model with the mechanism switched off".
for comparator in ["Control","DKT-F"]:
  for bid,(k,lo,hi) in enumerate(BUCKETS):
    sel=(B==bid)&Mm
    if sel.sum()<CFG.min_points_bucket: continue
    for thr in CFG.utility_thresholds:
        up=utility(PRED["PCDT"][sel],Y[sel],thr); uc=utility(PRED[comparator][sel],Y[sel],thr)
        rows.append(dict(horizon=k, comparator=comparator, threshold=thr,
                         extra_correctly_flagged=up["correctly_flagged"]-uc["correctly_flagged"],
                         extra_false_alarms=up["false_alarms"]-uc["false_alarms"],
                         fewer_missed=uc["missed"]-up["missed"],
                         d_precision=up["precision"]-uc["precision"], d_recall=up["recall"]-uc["recall"],
                         d_f1=up["f1"]-uc["f1"], d_balanced_acc=up["balanced_accuracy"]-uc["balanced_accuracy"],
                         pcdt_precision=up["precision"], comparator_precision=uc["precision"]))
T6=pd.DataFrame(rows); T6.to_csv(f"{TAB_DIR}/table6_educational_utility.csv", index=False)
print(T6.round(4).to_string(index=False))
print("\nRead the two count columns together: an operating point that adds correctly-flagged learners "
      "while adding few or no false alarms is a strict improvement in review targeting.")
```

## 16. Simulator validation — referee concern D

The simulator is only used where a real horizon is inadequate. Because it can carry a reported result, its
assumptions and failure modes are characterised here rather than asserted.

```python
def simulate(n_learners=3000, decay_d=0.5, noise=0.0, seed=SEED):
    rg=np.random.default_rng(seed); skills=[f"S{i}" for i in range(40)]
    diff={s:rg.normal(0,.8) for s in skills}; rows=[]
    t0=pd.Timestamp("2024-01-01",tz="UTC")
    for u in range(n_learners):
        ab=rg.normal(0,1); mast={s:rg.normal(-.5,.3) for s in skills}; last={s:None for s in skills}
        t=t0+pd.Timedelta(days=int(rg.integers(0,30)))
        for _ in range(int(rg.integers(4,12))):
            uu=rg.random(); gap=(rg.uniform(0,1) if uu<.4 else rg.uniform(1,7) if uu<.7
                                 else rg.uniform(7,28) if uu<.9 else rg.uniform(28,70))
            t+=pd.Timedelta(days=float(gap))
            for _ in range(int(rg.integers(5,25))):
                s=skills[int(rg.integers(len(skills)))]
                if last[s] is not None:
                    dd=(t-last[s]).total_seconds()/86400
                    mast[s]*= (1+dd)**(-decay_d)                     # ACT-R base-level decay
                p=1/(1+np.exp(-(ab+mast[s]-diff[s]+rg.normal(0,noise))))
                c=int(rg.random()<np.clip(p,.02,.98)); mast[s]+= .2 if c else .06; last[s]=t
                rows.append(dict(learner_id=f"u{u}",item_id=int(rg.integers(0,500)),skill_id=s,
                                 start_time=t,correct=float(c),
                                 response_time=float(np.clip(rg.lognormal(3,.4),2,900)),
                                 hint_frac=np.nan,attempt_count=np.nan))
                t+=pd.Timedelta(seconds=int(rg.integers(20,240)))
    return pd.DataFrame(rows)[STD]

sens=[]
for decay in (0.3,0.5,0.7):
    for noise in (0.0,0.5):
        sdf=simulate(n_learners=800, decay_d=decay, noise=noise)
        sens.append(dict(decay_d=decay, noise=noise, n=len(sdf), correct_rate=float(sdf.correct.mean()),
                         median_span_d=float((sdf.groupby("learner_id")["start_time"]
                            .agg(lambda s:(s.max()-s.min()).total_seconds()).median()/86400))))
TS=pd.DataFrame(sens); TS.to_csv(f"{TAB_DIR}/tableB_simulator_sensitivity.csv", index=False)
print(TS.round(3).to_string(index=False))
print("""
Documented assumptions and failure modes:
  * decay is ACT-R base-level, applied per skill on the elapsed gap since that skill was last seen;
  * mastery gain is fixed per outcome and does not depend on item difficulty or spacing;
  * no guessing/slipping asymmetry beyond the sigmoid floor and ceiling;
  * FAILURE CASE: with decay_d = 0 the simulator produces no forgetting, and any persistence advantage
    measured on it is an artefact of the model rather than the data-generating process. Simulated
    horizons must therefore never be used to support a persistence claim on their own.""")
```

```python
abl=[]
for name in ["PCDT","PCDT-noPsi","DKT-F","DKT","SAKT","AKT","DKVMN","Control"]:
    r=evaluate(PRED[name][M],Y[M]); r["model"]=name; abl.append(r)
T7=pd.DataFrame(abl)[["model","auc","acc","rmse","nll","ece"]]
T7.to_csv(f"{TAB_DIR}/table7_ablation.csv", index=False); print(T7.round(4).to_string(index=False))

sweep=[]
for tau in (600.,3600.,86400.):
    b=lambda h: PCDT(N_SKILLS,h,n_behav=len(BEHAV),tau0=tau)
    hh,_,_=match_capacity(b); mdl,va=train(b,hh,1e-3)
    p=predict(mdl,PACKS["test"]); sweep.append(dict(tau0=tau, val_auc=va, test_auc=evaluate(p[M],Y[M])["auc"]))
T8=pd.DataFrame(sweep); T8.to_csv(f"{TAB_DIR}/table8_sensitivity.csv", index=False)
print(T8.round(4).to_string(index=False))
```
