# Experiment P2-01 — Construct-Validity Audit of Candidate PCDT State Dimensions

**Paper 2:** *A Persistent Multidimensional State Representation for Cognitive Digital Twins*
**Research question addressed:** RQ2.1
**Dataset:** ASSISTments 2020–2021 School Year (Feb–Apr release)
**Notebook version:** 2.0 (regenerated, corrected, and documented)

---

## Experiment Overview

This experiment audits which of seven *candidate* dimensions of a Persistent Cognitive Digital Twin (PCDT)
state can be **operationalised with defensible construct validity** from ASSISTments interaction logs.

The audit is deliberately built to be able to **fail**. It is not a demonstration that the seven-dimensional
representation works; it is a test of whether each dimension survives three independent measurement criteria:

| Criterion | Question it answers | Applies to |
|---|---|---|
| **Reliability** | Is the proxy internally consistent across independent slices of a learner's history? | all dimensions |
| **Convergent validity** | Does the proxy relate to an independent behavioural outcome? | dimensions with a real outcome target |
| **Simulator recovery** | Can the proxy recover a *known* latent value from synthetic data? | dimensions the simulator instruments |

A dimension is retained only if it passes reliability **and** the validity criterion appropriate to it.

> **Epistemic commitment maintained throughout.** No observable variable in this dataset is treated as
> cognitive ground truth. Correctness, response time, hint usage, and attempt counts are *observations*
> that carry evidence about latent state; they are not the latent state.


## 1. Objectives and Research Question

### RQ2.1
> Which subset of the seven proposed PCDT dimensions $\{I_t, C_t, B_t, M_t, G_t, U_t, H_t\}$ can be
> operationalised with defensible construct validity using observations available in the target
> educational-interaction datasets, and which measurement protocol, proxy variable, or
> synthetic-simulator validation supports each?

### Concrete objectives

1. Construct an explicit, documented proxy for each candidate dimension from ASSISTments observables.
2. Estimate the **reliability** of each proxy using a split-half design with a defensible precision floor.
3. Estimate **convergent validity** against a temporally held-out behavioural outcome.
4. Estimate **simulator recovery** against known latent values from an ACT-R-style generative simulator.
5. Apply a **pre-registered** decision rule and report a verdict per dimension — including rejections.

### What this experiment does *not* do

- It does not claim to measure cognition directly.
- It does not validate the *full* seven-dimensional structure; it tests each dimension separately.
- It does not establish predictive superiority (that is RQ2.2) or persistence (RQ2.3) or stability (RQ2.4).


## 2. Theoretical Framing and Mathematical Formulation

### 2.1 Latent state, observations, estimates, belief

We model a learner over discrete interaction steps $t = 1, 2, \dots, T$.

| Symbol | Meaning | Status in this experiment |
|---|---|---|
| $S_t$ | true latent learner state | **unobservable** — never accessed on real data |
| $O_t$ | observable evidence at step $t$ | correctness, response time, hint fraction, attempts, skill tag, timestamp |
| $\hat{S}_t$ | model estimate of latent state | what each proxy produces |
| $b_t$ | belief (distribution) over latent state | *not* estimated here — see note below |

The belief state is defined as

$$b_t \;=\; P\!\left(S_t \mid O_{1:t},\, A_{1:t-1}\right)$$

where $A_{1:t-1}$ is the sequence of instructional actions preceding step $t$.

> **This experiment computes point estimates $\hat{S}_t$, not beliefs $b_t$.** Constructing $b_t$ requires the
> generative inference model of Paper 3a. Consequently the **uncertainty dimension $U_t$ is excluded from
> this audit**: under the corrected formal framework, $U_t$ is a property of the belief distribution
> (its dispersion, e.g. $\Sigma_t$), not a coordinate of the latent state. Auditing it as though it were a
> state dimension would be a category error.

### 2.2 The candidate PCDT state

$$\mathrm{CDT}_t \;=\; \bigl(I_t,\; C_t,\; B_t,\; M_t,\; G_t,\; U_t,\; H_t\bigr)$$

| Symbol | Construct | Audited here? |
|---|---|---|
| $I_t$ | identity / longitudinal continuity | ✔ |
| $C_t$ | cognition / mastery | ✔ |
| $B_t$ | behaviour | ✔ |
| $M_t$ | metacognition | ✔ |
| $G_t$ | goals | ✔ |
| $U_t$ | uncertainty | ✘ — belief-level quantity, see §2.1 |
| $H_t$ | historical memory | ✔ |

Six dimensions are audited. The relationship between the observable stream and each proxy is

$$O_{1:T} \;\xrightarrow{\;\text{proxy operator } \pi_d\;}\; \hat{S}^{(d)} \qquad d \in \{I, C, B, M, G, H\}$$

where $\pi_d$ is the deterministic, hand-specified aggregation defined in §7. Each $\hat{S}^{(d)}$ is a
scalar per learner, **not** a claim about $S_t$.

### 2.3 Proxy operators

Let $\mathcal{L}$ be the set of learners and $R_\ell$ the interaction rows of learner $\ell$.
Let $x_i$ denote observable $x$ on row $i$.

**Cognition / mastery** — mean first-attempt correctness:
$$\pi_C(\ell) \;=\; \frac{1}{|R_\ell|}\sum_{i \in R_\ell} c_i, \qquad c_i \in \{0,1\}$$

**Behaviour** — standardised effort composite:
$$\pi_B(\ell) \;=\; \tfrac{1}{2}\!\left(\frac{\bar{r}_\ell - \mu_r}{\sigma_r} + \frac{\bar{a}_\ell - \mu_a}{\sigma_a}\right)$$
with $\bar{r}_\ell,\bar{a}_\ell$ the learner's mean response time and mean attempt count, and
$(\mu_r,\sigma_r),(\mu_a,\sigma_a)$ **reference-population** statistics (§7.1).

**Metacognition** — inverted standardised help-seeking:
$$\pi_M(\ell) \;=\; -\,\frac{\bar{h}_\ell - \mu_h}{\sigma_h}$$
where $\bar{h}_\ell$ is mean `fraction_of_hints_used`. Lower help-seeking is treated as a *proxy indicator*
of stronger self-regulation. This is a contested assumption and is revisited in Limitations.

**Identity** — accuracy combined with breadth of engagement:
$$\pi_I(\ell) \;=\; \tfrac{1}{2}\pi_C(\ell) \;+\; \tfrac{1}{2}\cdot\tfrac{1}{2}\!\left(\frac{n^{\text{sess}}_\ell - \mu_s}{\sigma_s} + \frac{n^{\text{skill}}_\ell - \mu_k}{\sigma_k}\right)$$

**Goals** — Shannon entropy of the skill-attempt distribution:
$$\pi_G(\ell) \;=\; -\sum_{k \in \mathcal{K}_\ell} p_{\ell k}\log p_{\ell k}, \qquad
p_{\ell k} = \frac{n_{\ell k}}{\sum_{k'} n_{\ell k'}}$$

**Historical memory** — recency-weighted interaction mass:
$$\pi_H(\ell) \;=\; \sum_{i \in R_\ell} \exp\!\bigl(-\lambda\,(s^{\max}_\ell - s_i)\bigr), \qquad \lambda = 0.3$$
where $s_i$ is the session index of row $i$ and $s^{\max}_\ell$ the learner's last session.

### 2.4 Measurement model and the precision floor

Treat an observed proxy value as a true score plus measurement error:
$$\hat{S}^{(d)}_{\ell} \;=\; S^{(d)}_{\ell} \;+\; \varepsilon_{\ell}, \qquad
\operatorname{Var}(\varepsilon_\ell) \;\propto\; \frac{1}{n_\ell}$$

For a proportion-type proxy such as $\pi_C$ estimated from $n$ Bernoulli trials with success probability $p$,
$$\operatorname{Var}(\hat{p}) = \frac{p(1-p)}{n}$$

so the reliability of a split-half comparison is **mechanically bounded above** by the number of
observations per half. Classical test theory gives the attenuation relation
$$r_{\text{obs}} \;=\; r_{\text{true}}\sqrt{\rho_1 \rho_2}$$
where $\rho_j$ is the reliability of half $j$. **This is the single most important methodological fact in
this notebook**: including learners with 1–3 observations per half drives estimated reliability toward zero
regardless of whether the underlying construct is stable. §10 quantifies this and derives the precision
floor used throughout.

### 2.5 Reliability estimator

Sessions are partitioned into odd/even halves. With $r_{hh}$ the Pearson correlation between
half-scores across learners, the Spearman–Brown prophecy formula gives the full-length reliability:

$$\rho_{SB} \;=\; \frac{2\,r_{hh}}{1 + r_{hh}}$$

### 2.6 Convergent validity

$$r_d \;=\; \operatorname{corr}\!\left(\hat{S}^{(d)}_{\ell},\; y_\ell\right)$$

where $y_\ell$ is the temporally held-out outcome defined in §11.

### 2.7 Simulator recovery

For a simulator that instruments dimension $d$ with a known latent value $\theta_\ell$:

$$r^{\text{sim}}_d \;=\; \operatorname{corr}\!\left(\theta_\ell,\; \pi_d(\text{sim}(\theta_\ell))\right)$$

Recovery is **defined only if the simulator's generative process actually depends on $\theta$ for that
dimension**. If it does not, the correlation is noise and must be reported as undefined, not as evidence.

### 2.8 ACT-R base-level activation

The simulator's memory mechanism uses the ACT-R base-level learning equation:

$$B_i(t) \;=\; \ln\!\left(\sum_{j\,:\,t_j < t} (t - t_j)^{-\delta}\right)$$

with $t_j$ the times of prior retrievals of item $i$ and decay $\delta = 0.5$. Retrieval success is

$$P(\text{correct}) \;=\; \sigma\!\bigl(B_i(t) + \theta\bigr), \qquad \sigma(z) = \frac{1}{1+e^{-z}}$$


## 3. Corrections Log — changes from notebook v1.0

Every substantive change is listed. Corrections marked **[CRITICAL]** altered a reported result.

### [CRITICAL] C-1 — Reliability was attenuated by learners with almost no data
**Symptom.** $C_t$ (cognition/mastery) returned $\rho_{SB} = 0.576$ and a verdict of **REJECT** — a result
that contradicts the entire knowledge-tracing literature and would not survive peer review.
**Diagnosis.** v1.0 computed split-half reliability over *all* learners, including those contributing a
single observation per half. Simulation with a **perfectly stable** true ability shows $\rho_{SB}$ of
$0.29$ at 1 item/half, $0.59$ at 3, $0.68$ at 5, $0.81$ at 10 (reproduced in §10). The observed $0.576$
corresponds to roughly 3 items/half — i.e. the estimate reflected *measurement precision*, not construct
instability.
**Correction.** A pre-registered precision floor `MIN_ITEMS_PER_HALF` is applied, plus a sensitivity
analysis across floor values (§13) so the result is not an artifact of one arbitrary choice.

### [CRITICAL] C-2 — `NaN` propagation silently voided two dimensions
**Symptom.** $B_t$ and $M_t$ reliability returned `nan`.
**Diagnosis.** `hint_count` and `response_time` contain missing values; a learner with all-missing values in
one half yields a `NaN` half-score, and `scipy.stats.pearsonr` propagates a single `NaN` to the whole
statistic. The convergent-validity routine masked `NaN`s; the reliability routine did not.
**Correction.** Pairwise-complete masking applied consistently in all three metric routines.

### [CRITICAL] C-3 — Simulator recovery reported for an uninstrumented dimension
**Symptom.** $I_t$ reported simulator recovery $r = -0.180$.
**Diagnosis.** The simulator branches on `dimension` for `C_t, B_t, M_t, G_t, H_t` only. For `I_t` the
generative process **does not depend on the true value at all**, so the synthetic data carries no signal
about $\theta$. The reported $-0.180$ is sampling noise presented as evidence.
**Correction.** An explicit `SIMULATOR_INSTRUMENTS` map is declared; uninstrumented dimensions report
`None` ("not testable"), and the verdict rule returns **INCONCLUSIVE** rather than silently rejecting.

### C-4 — Normalisation constants computed on a pre-cleaning row set
`GS` was computed before 1,725,178 rows with unusable `correct` were dropped, so proxies were standardised
against a different population than they were evaluated on. Cleaning now precedes all statistics.

### C-5 — Simulator used real-data normalisation constants
Synthetic learners (6 sessions, 8 skills) were z-scored against real-data means (7.5 sessions, 425 skills),
making the breadth term meaningless on synthetic data. Normalisation is now computed **within the reference
population being analysed**, passed explicitly as an argument.

### C-6 — Degenerate convergent-validity target
The held-out outcome was the mean of a learner's final 30% of interactions, which for many learners is 1–3
items, yielding values of exactly 0.0 or 1.0. A pre-registered `MIN_HOLDOUT_ITEMS` floor is applied and the
retained-learner count reported.

### C-7 — `abs()` in the verdict rule masked sign disagreement
$G_t$ was RETAINed on $|{-0.945}|$. The negative sign is a *parameterisation artifact* (higher simulated
goal-focus produces lower entropy). Expected signs are now declared per dimension and checked explicitly.

### C-8 — Thresholds declared after results were visible
Decision thresholds appeared in the final cell. They are now declared in §5, before any statistic is
computed, and the notebook states that they must not be edited after inspecting results.

### C-9 — Deduplicated, vectorised, and cleaned
Duplicate convergent-validity cells removed; `pd.to_datetime` called once instead of twice on 8.9M rows;
`groupby.apply` replaced with vectorised `transform` (removes the pandas `DeprecationWarning` and is
substantially faster); per-group entropy replaced with a vectorised computation; `SettingWithCopyWarning`
eliminated via explicit `.copy()`; unused `sdets` load and unused imports removed; global RNG seeded.

## 5. Configuration and Pre-Registered Decision Thresholds

> **Pre-registration discipline.** Everything in this cell is a decision that must be fixed *before* any
> statistic is inspected. Editing a threshold after seeing a result converts this audit into a
> confirmation exercise. If a threshold must change, re-run the audit on a fresh data partition and
> document the change.

**Justification of each threshold**

| Parameter | Value | Basis |
|---|---|---|
| `reliability_min` | 0.70 | Conventional floor for group-level research instruments (Nunnally-style convention). |
| `convergent_r_min` | 0.30 | Cohen's "medium" effect; below this a proxy explains <10% of outcome variance. |
| `simulator_r_min` | 0.50 | Recovery from *synthetic* data with known ground truth should be substantially stronger than a real-world convergent correlation; a weak recovery under ideal conditions indicates the operator does not track the construct. |
| `MIN_ITEMS_PER_HALF` | 5 | Derived in §10 from the measurement-precision analysis: below ~5 items/half, reliability is bounded by sampling noise rather than construct stability. |
| `MIN_HOLDOUT_ITEMS` | 5 | Same precision argument applied to the outcome variable. |
| `SESSION_GAP_HOURS` | 4 | ASSISTments defines no session; a 4-hour inactivity gap is the operational boundary. Sensitivity to this choice is a documented limitation. |
| `HOLDOUT_FRACTION` | 0.30 | Final 30% of each learner's timeline reserved as the outcome window. |


```python
@dataclass(frozen=True)
class Config:
    # --- pre-registered decision thresholds ---
    reliability_min: float = 0.70
    convergent_r_min: float = 0.30
    simulator_r_min: float = 0.50

    # --- measurement-precision floors (justified in section 10) ---
    min_items_per_half: int = 5
    min_holdout_items: int = 5

    # --- data construction ---
    session_gap_hours: float = 4.0
    holdout_fraction: float = 0.30

    # --- proxy hyper-parameters ---
    recency_lambda: float = 0.30      # H_t exponential recency decay
    actr_decay: float = 0.50          # ACT-R base-level decay delta

    # --- simulator ---
    n_sim_learners: int = 300
    sim_sessions: int = 6
    sim_items_per_session: int = 15
    sim_session_gap_days: float = 7.0

    # --- sensitivity analysis grid ---
    sensitivity_floors: Sequence[int] = (1, 2, 3, 5, 10, 20)

CFG = Config()

DATA_DIR = "/home/hamed/Downloads/Research/dataset/feb_to_apr"
PLOGS_PATH = f"{DATA_DIR}/plogs.csv"
PDETS_PATH = f"{DATA_DIR}/pdets.csv"

DIMENSIONS = ["I_t", "C_t", "B_t", "M_t", "G_t", "H_t"]   # U_t excluded: belief-level (see section 2.1)

# Which dimensions have a defensible REAL convergent-validity target vs. must rely on the simulator.
HAS_REAL_TARGET: Dict[str, bool] = {
    "I_t": False,   # no external identity measure exists in ASSISTments
    "C_t": True,    # future correctness is a defensible mastery-related outcome
    "B_t": True,    # behavioural effort plausibly relates to future performance
    "M_t": False,   # no external metacognition measure exists
    "G_t": False,   # no external goal measure exists
    "H_t": True,    # engagement mass plausibly relates to future performance
}

# Does the generative simulator's data-generating process actually depend on the dimension's
# true value?  If not, "recovery" is undefined -- see correction C-3.
SIMULATOR_INSTRUMENTS: Dict[str, bool] = {
    "I_t": False,   # NOT instrumented: simulator has no stable-trait mechanism
    "C_t": True,
    "B_t": True,
    "M_t": True,
    "G_t": True,
    "H_t": True,
}

# Expected sign of simulator recovery, given the simulator's parameterisation (see correction C-7).
EXPECTED_SIM_SIGN: Dict[str, Optional[int]] = {
    "I_t": None,    # undefined -- not instrumented
    "C_t": +1,      # higher ability -> higher correctness
    "B_t": +1,      # higher effort parameter -> higher RT/attempts
    "M_t": +1,      # higher metacognition -> fewer hints -> proxy is negated -> positive
    "G_t": -1,      # higher goal-focus -> fewer distinct skills -> LOWER entropy
    "H_t": +1,      # higher value -> more sessions -> greater recency-weighted mass
}

print("Thresholds fixed BEFORE any statistic is computed:")
print(f"  reliability >= {CFG.reliability_min}")
print(f"  |convergent r| >= {CFG.convergent_r_min}")
print(f"  |simulator r| >= {CFG.simulator_r_min}  (sign must match expectation)")
print(f"  precision floors: {CFG.min_items_per_half} items/half, {CFG.min_holdout_items} holdout items")
```

## 6. Dataset Description

**ASSISTments 2020–2021 School Year dataset**, February–April release. The dataset is *relational*: ten
tables joined on shared identifiers. This experiment uses two.

### `plogs` — problem logs (one row per student × problem attempt)

| Column | Type | Role here |
|---|---|---|
| `student_id` | int | learner identifier → `learner_id` |
| `problem_id` | int | item identifier → `item_id` |
| `start_time` | timestamp (tz-aware) | session derivation + temporal split |
| `time_on_task` | float (s) | → `response_time`, input to $\pi_B$ |
| `attempt_count` | int | input to $\pi_B$ |
| `fraction_of_hints_used` | float | → `hint_count`, input to $\pi_M$ |
| `correct` | bool / missing | first-attempt correctness, input to $\pi_C$, $\pi_I$ |

### `pdets` — problem details

| Column | Type | Role here |
|---|---|---|
| `problem_id` | int | join key |
| `skills` | list of Common Core codes | exploded to `skill_id`, input to $\pi_G$, $\pi_I$ |

### Structural properties that shape the analysis

1. **No session concept.** Sessions must be *derived* from inter-arrival times (§9). This is a modelling
   decision, not a property of the data.
2. **Multi-label skills.** One problem may map to several standards; the explode-join therefore
   **duplicates interaction rows**, one per (problem, skill) pair. This is correct for skill-indexed
   quantities but means row counts are not interaction counts.
3. **Missing `correct`.** Ungraded, incomplete, and teacher-graded-open-response items have null
   correctness. In the v1.0 run this removed 1,725,178 rows.
4. **`fraction_of_hints_used` is a proportion, not a count** — and per the official documentation it can
   exceed 1.0 when hints were later deleted from the platform. It must never be reported as a hint count.
5. **Teacher-assigned content.** Problem sets and skill builders are assigned by teachers, so any
   "goal" signal derived from content selection substantially reflects *curriculum*, not learner intent.
   This is decisive for interpreting $G_t$ (see Interpretation).


## 7. Data Loading

Only the required columns are read, which materially reduces memory on an ~8.9M-row table.
The loader falls back to a synthetic fixture in the same schema when the Drive path is unavailable, so the
notebook's logic is executable and testable outside Colab. The fixture is **clearly flagged** and its
results must never be reported as findings.


```python
import os, ast

PLOGS_COLS = ["student_id", "problem_id", "start_time", "time_on_task",
              "attempt_count", "fraction_of_hints_used", "correct"]

def _parse_skills(value):
    """`skills` arrives as a stringified list, e.g. "['4.OA.A.1','4.OA.A.2']"."""
    if isinstance(value, list):
        return value
    if isinstance(value, str) and value.strip().startswith("["):
        try:
            return ast.literal_eval(value)
        except (ValueError, SyntaxError):
            return []
    return [] if pd.isna(value) else [value]

def make_synthetic_fixture(n_learners=800, rng=None):
    """Synthetic data in the REAL plogs/pdets schema, for pipeline verification only."""
    rng = rng or np.random.default_rng(0)
    n_problems, all_skills = 300, [f"CCSS.{i}" for i in range(40)]
    pdets = pd.DataFrame({
        "problem_id": range(n_problems),
        "skills": [list(rng.choice(all_skills, rng.integers(1, 3), replace=False))
                   for _ in range(n_problems)],
    })
    rows, t0_base = [], pd.Timestamp("2021-02-01", tz="UTC")
    for lid in range(n_learners):
        ability, metacog = rng.normal(0, 1), rng.beta(2, 2)
        t = t0_base + pd.Timedelta(days=int(rng.integers(0, 20)))
        for _ in range(int(rng.integers(3, 12))):                 # sessions
            t += pd.Timedelta(days=int(rng.integers(1, 10)), hours=int(rng.integers(0, 20)))
            for _ in range(int(rng.integers(4, 25))):             # items in session
                p = 1 / (1 + np.exp(-(ability + rng.normal(0, .5))))
                rows.append(dict(
                    student_id=lid, problem_id=int(rng.integers(0, n_problems)), start_time=t,
                    time_on_task=float(np.clip(rng.lognormal(3 - metacog * .3, .4), 2, 900)),
                    attempt_count=int(1 + rng.poisson(max(0, (1 - metacog) * 1.5))),
                    fraction_of_hints_used=(np.nan if rng.random() < .15
                                            else float(np.clip(rng.beta(2, 5) * (1.3 - metacog), 0, 1.3))),
                    correct=(np.nan if rng.random() < .18 else bool(rng.binomial(1, np.clip(p, .01, .99)))),
                ))
                t += pd.Timedelta(seconds=int(rng.integers(20, 200)))
    return pd.DataFrame(rows), pdets

USING_SYNTHETIC = not os.path.exists(PLOGS_PATH)

if USING_SYNTHETIC:
    print("!" * 78)
    print("! Drive path not found -> using SYNTHETIC FIXTURE.")
    print("! Pipeline verification only. These numbers are NOT construct-validity results.")
    print("!" * 78)
    plogs_raw, pdets_raw = make_synthetic_fixture(rng=RNG)
else:
    plogs_raw = pd.read_csv(PLOGS_PATH, usecols=PLOGS_COLS)
    pdets_raw = pd.read_csv(PDETS_PATH, usecols=["problem_id", "skills"])
    pdets_raw["skills"] = pdets_raw["skills"].map(_parse_skills)

print(f"\nplogs: {len(plogs_raw):,} rows | {plogs_raw['student_id'].nunique():,} learners")
print(f"pdets: {len(pdets_raw):,} problems")
plogs_raw.head()
```

## 8. Data Cleaning and Preprocessing

**Order matters and was wrong in v1.0** (correction C-4): all type coercion and row filtering happens
*before* any statistic — including normalisation constants — is computed.

Steps, in order:

1. **Type coercion.** `correct` arrives as an object column mixing `True`/`False`/`NaN`; pandas
   aggregations on object dtype raise obscure errors. All analysis columns are coerced to numeric with
   `errors="coerce"`.
2. **Drop unusable correctness.** Rows with null `correct` carry no mastery evidence. The count dropped is
   reported explicitly — this is a substantial and non-random deletion, revisited in Limitations.
3. **Timestamp parsing.** ASSISTments timestamps are timezone-aware with *inconsistent* sub-second
   precision, so `format="ISO8601"` is required; `utc=True` normalises offsets and prevents
   tz-aware/naive arithmetic errors downstream. Parsed **once** (v1.0 parsed twice).
4. **Skill explode-join.** Multi-label skills are exploded and inner-joined; problems with no tagged skill
   are dropped, and the loss is reported rather than imputed.


```python
def clean_and_join(plogs: pd.DataFrame, pdets: pd.DataFrame) -> pd.DataFrame:
    df = plogs.copy()

    # 1. coerce numerics (correct is object dtype: True/False/NaN)
    df["correct"] = pd.to_numeric(
        df["correct"].map({True: 1.0, False: 0.0, "True": 1.0, "False": 0.0}).fillna(df["correct"]),
        errors="coerce")
    for col in ("time_on_task", "attempt_count", "fraction_of_hints_used"):
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # 2. drop rows with no usable correctness signal
    n0 = len(df)
    df = df.dropna(subset=["correct"])
    print(f"dropped {n0 - len(df):,} rows ({(n0-len(df))/n0:.1%}) with missing/unusable `correct`")

    # 3. parse timestamps ONCE (ISO8601 handles inconsistent sub-second precision)
    df["start_time"] = pd.to_datetime(df["start_time"], format="ISO8601", utc=True)

    # 4. explode multi-label skills, inner join
    sk = pdets[["problem_id", "skills"]].explode("skills").rename(columns={"skills": "skill_id"})
    sk = sk.dropna(subset=["skill_id"])
    n_prob_before = df["problem_id"].nunique()
    df = df.merge(sk, on="problem_id", how="inner")
    n_prob_after = df["problem_id"].nunique()
    print(f"skill join: {n_prob_before:,} -> {n_prob_after:,} problems retained "
          f"({n_prob_before - n_prob_after:,} lacked skill tags)")

    df = df.rename(columns={"student_id": "learner_id", "problem_id": "item_id",
                            "time_on_task": "response_time",
                            "fraction_of_hints_used": "hint_count"})
    return df[["learner_id", "item_id", "skill_id", "start_time",
               "correct", "response_time", "hint_count", "attempt_count"]]

interactions = clean_and_join(plogs_raw, pdets_raw)
print(f"\nanalysis table: {len(interactions):,} rows | "
      f"{interactions['learner_id'].nunique():,} learners | "
      f"{interactions['skill_id'].nunique():,} skills")
interactions.dtypes
```

## 9. Session Derivation

ASSISTments has no session identifier, so sessions are derived: a new session begins whenever the gap since
a learner's previous interaction exceeds `SESSION_GAP_HOURS`. Formally, for learner $\ell$ with ordered
timestamps $t_1 < t_2 < \dots$,

$$s_i \;=\; \sum_{j\le i} \mathbb{1}\!\left[(t_j - t_{j-1}) > \tau\right], \qquad \tau = 4\ \text{hours}$$

The implementation is vectorised via `groupby(...).transform(...)`, replacing v1.0's
`groupby(...).apply(...)` — which is both slower on ~136k groups and emits a `DeprecationWarning` because it
operated on the grouping column (correction C-9).

> **Assumption.** Sessions are a *derived construct*. The 4-hour threshold is defensible but arbitrary;
> reliability is estimated by splitting on these sessions, so the threshold propagates into every
> reliability figure. Sensitivity to $\tau$ is listed in Limitations.


```python
def derive_sessions(df: pd.DataFrame, gap_hours: float) -> pd.DataFrame:
    df = df.sort_values(["learner_id", "start_time"]).copy()
    gap = df.groupby("learner_id")["start_time"].diff().dt.total_seconds() / 3600.0
    df["session_id"] = (gap.fillna(0) > gap_hours).groupby(df["learner_id"]).cumsum().astype(int)
    return df

interactions = derive_sessions(interactions, CFG.session_gap_hours)

_sess = interactions.groupby("learner_id")["session_id"].nunique()
print(f"sessions/learner: mean={_sess.mean():.1f}  median={_sess.median():.0f}  max={_sess.max()}")
print(f"learners with only 1 session: {(_sess == 1).sum():,} ({(_sess == 1).mean():.1%})")
```

## 10. Exploratory Analysis and the Measurement-Precision Floor

This section exists because it **explains the single largest error in v1.0**.

### 10.1 The problem

Split-half reliability compares two independent estimates of a learner's proxy score. If each estimate is
based on very few observations, the estimates disagree due to *sampling noise alone*, and the reliability
coefficient collapses — even when the underlying trait is perfectly stable.

### 10.2 Quantifying the floor

We simulate learners whose true ability is **fixed by construction** (so true reliability is 1.0 by
definition), draw two independent halves of $n$ Bernoulli responses each, and compute $\rho_{SB}$. Any
shortfall below 1.0 is pure measurement noise.

This is not a side analysis — it determines whether a REJECT verdict means "this construct is unstable" or
"we did not measure it precisely enough to tell."


```python
def reliability_ceiling_by_n(n_items_grid, n_learners=4000, rng=None):
    """Max attainable split-half reliability when the true trait is perfectly stable."""
    rng = rng or np.random.default_rng(SEED)
    out = []
    for n in n_items_grid:
        ability = rng.normal(0, 1, n_learners)
        p = 1 / (1 + np.exp(-ability))
        h1 = rng.binomial(n, p) / n
        h2 = rng.binomial(n, p) / n
        r = stats.pearsonr(h1, h2)[0]
        out.append({"items_per_half": n, "raw_r": r, "spearman_brown": 2 * r / (1 + r)})
    return pd.DataFrame(out)

ceiling = reliability_ceiling_by_n([1, 2, 3, 5, 10, 20, 50])
print("Attainable reliability ceiling (true trait is PERFECTLY stable by construction):\n")
print(ceiling.round(3).to_string(index=False))
print(f"\n-> v1.0 reported C_t reliability = 0.576, which corresponds to ~3 items/half.")
print("   That is a measurement-precision artifact, not evidence about the construct.")
```

```python
import matplotlib.pyplot as plt

items_per_learner = interactions.groupby("learner_id").size()

fig, axes = plt.subplots(1, 3, figsize=(16, 4.2))

axes[0].hist(np.log10(items_per_learner.clip(lower=1)), bins=50,
             color="#2c3e50", edgecolor="white")
axes[0].set_xlabel("log$_{10}$(interactions per learner)")
axes[0].set_ylabel("learners")
axes[0].set_title("Interaction volume is highly skewed")

axes[1].plot(ceiling["items_per_half"], ceiling["spearman_brown"],
             "o-", color="#c0392b", lw=2)
axes[1].axhline(CFG.reliability_min, ls="--", c="grey",
                label=f"threshold = {CFG.reliability_min}")
axes[1].axvline(CFG.min_items_per_half, ls=":", c="#2980b9",
                label=f"floor = {CFG.min_items_per_half}")
axes[1].set_xscale("log")
axes[1].set_xlabel("items per half")
axes[1].set_ylabel(r"attainable $\rho_{SB}$")
axes[1].set_title("Reliability ceiling vs. measurement precision")
axes[1].legend(fontsize=8)

frac_retained = [(items_per_learner >= 2 * f).mean() for f in CFG.sensitivity_floors]
axes[2].plot(CFG.sensitivity_floors, frac_retained, "s-", color="#27ae60", lw=2)
axes[2].axvline(CFG.min_items_per_half, ls=":", c="#2980b9")
axes[2].set_xlabel("min items per half")
axes[2].set_ylabel("fraction of learners retained")
axes[2].set_title("Precision/coverage trade-off")

plt.tight_layout(); plt.show()

print(f"median interactions/learner: {items_per_learner.median():.0f}")
print(f"learners with >= {2*CFG.min_items_per_half} interactions: "
      f"{(items_per_learner >= 2*CFG.min_items_per_half).mean():.1%}")
```

## 11. Convergent-Validity Target Construction

For each learner the timeline is split at the `HOLDOUT_FRACTION` quantile of `start_time`. Proxies are
computed on the **earlier** window; the outcome is computed on the **later** window:

$$y_\ell \;=\; \frac{1}{|H_\ell|}\sum_{i \in H_\ell} c_i, \qquad
H_\ell = \{\,i \in R_\ell : t_i > q_{0.7}(t_{R_\ell})\,\}$$

**Two properties must be stated plainly.**

1. **This is not an independent post-test.** It is the same platform, same task family, a later slice of the
   same activity stream. It is a *temporally held-out behavioural outcome*, and must be described that way
   in any manuscript. It provides no external construct validation.

2. **Leakage is prevented by construction:** the outcome window contributes no rows to proxy computation.
   Correction C-6 adds `MIN_HOLDOUT_ITEMS` because a mean over 1–3 items is nearly degenerate — in the v1.0
   run many learners had $y_\ell \in \{0.0, 1.0\}$ exactly.


```python
def build_holdout_target(df, holdout_fraction, min_holdout_items):
    """Split each learner's timeline temporally; return (analysis_window, outcome)."""
    df = df.sort_values(["learner_id", "start_time"]).copy()

    cutoff = df.groupby("learner_id")["start_time"].transform(
        lambda s: s.quantile(1.0 - holdout_fraction))
    is_holdout = df["start_time"] > cutoff

    hold = df.loc[is_holdout]
    agg = hold.groupby("learner_id")["correct"].agg(["mean", "size"])
    eligible = agg[agg["size"] >= min_holdout_items]
    target = eligible["mean"].rename("outcome")

    analysis = df.loc[~is_holdout].copy()

    print(f"holdout rows: {is_holdout.sum():,} ({is_holdout.mean():.1%})")
    print(f"learners with >= {min_holdout_items} holdout items: {len(target):,} "
          f"of {agg.shape[0]:,} ({len(target)/max(agg.shape[0],1):.1%})")
    print(f"outcome distribution: mean={target.mean():.3f} sd={target.std():.3f}")
    print(f"  exactly 0.0 or 1.0: {((target == 0) | (target == 1)).mean():.1%} "
          f"(v1.0 had no floor, so this fraction was much higher)")
    return analysis, target

analysis_df, outcome = build_holdout_target(
    interactions, CFG.holdout_fraction, CFG.min_holdout_items)
```

## 12. State Construction — Proxy Operators

All six proxies are computed in a **single vectorised pass** returning a learner × dimension matrix.
This replaces v1.0's six separate `groupby` traversals (and its `.apply`-based entropy, which was the
slowest operation in the notebook).

### Normalisation reference (correction C-5)

Standardised proxies require a reference population $(\mu, \sigma)$. v1.0 z-scored *synthetic* learners
against *real-data* constants, which is a scale mismatch that renders the breadth term meaningless. The
reference is now an explicit argument, so each population is standardised against itself:

$$z_x(\ell) = \frac{\bar{x}_\ell - \mu_x^{\text{ref}}}{\sigma_x^{\text{ref}}}$$

### Output

A `DataFrame` indexed by `learner_id` with columns `I_t, C_t, B_t, M_t, G_t, H_t` plus `n_items`
(retained for the precision filter). Every value is an estimate $\hat{S}^{(d)}_\ell$ — never $S_t$.


```python
def normalisation_reference(df: pd.DataFrame) -> Dict[str, float]:
    """Reference-population statistics for z-scored proxies."""
    per_learner = df.groupby("learner_id").agg(
        rt=("response_time", "mean"), att=("attempt_count", "mean"),
        hint=("hint_count", "mean"),
        n_sessions=("session_id", "nunique"), n_skills=("skill_id", "nunique"))
    ref = {}
    for col in ("rt", "att", "hint", "n_sessions", "n_skills"):
        ref[f"{col}_mu"] = per_learner[col].mean()
        sd = per_learner[col].std()
        ref[f"{col}_sd"] = sd if (pd.notna(sd) and sd > 0) else 1.0   # guard degenerate populations
    return ref


def compute_proxies(df: pd.DataFrame, ref: Dict[str, float],
                    recency_lambda: float) -> pd.DataFrame:
    """Vectorised computation of all six proxy operators. Index = learner_id."""
    g = df.groupby("learner_id")
    base = g.agg(n_items=("correct", "size"), acc=("correct", "mean"),
                 rt=("response_time", "mean"), att=("attempt_count", "mean"),
                 hint=("hint_count", "mean"),
                 n_sessions=("session_id", "nunique"), n_skills=("skill_id", "nunique"))

    z = lambda col, key: (base[col] - ref[f"{key}_mu"]) / ref[f"{key}_sd"]

    # --- G_t: Shannon entropy of skill distribution (vectorised; no per-group apply) ---
    counts = df.groupby(["learner_id", "skill_id"]).size().rename("n")
    totals = counts.groupby(level="learner_id").transform("sum")
    p = counts / totals
    entropy = (-p * np.log(p)).groupby(level="learner_id").sum()

    # --- H_t: recency-weighted interaction mass ---
    max_sess = df.groupby("learner_id")["session_id"].transform("max")
    w = np.exp(-recency_lambda * (max_sess - df["session_id"]))
    hist_mass = w.groupby(df["learner_id"]).sum()

    out = pd.DataFrame(index=base.index)
    out["n_items"] = base["n_items"]
    out["C_t"] = base["acc"]
    out["B_t"] = 0.5 * (z("rt", "rt") + z("att", "att"))
    out["M_t"] = -z("hint", "hint")
    out["I_t"] = 0.5 * base["acc"] + 0.5 * (0.5 * (z("n_sessions", "n_sessions")
                                                   + z("n_skills", "n_skills")))
    out["G_t"] = entropy.reindex(out.index)
    out["H_t"] = hist_mass.reindex(out.index)
    return out[["n_items"] + DIMENSIONS]

REF = normalisation_reference(analysis_df)
proxies = compute_proxies(analysis_df, REF, CFG.recency_lambda)

print(f"proxy matrix: {proxies.shape[0]:,} learners x {len(DIMENSIONS)} dimensions")
print("\nmissing values per dimension:")
print(proxies[DIMENSIONS].isna().sum().to_string())
proxies[DIMENSIONS].describe().T.round(3)
```

## 13. Metric 1 — Reliability

### Design

Sessions are partitioned into odd and even index sets; each learner's proxy is computed independently on
each half; the correlation across learners is Spearman–Brown corrected:

$$\rho_{SB} = \frac{2 r_{hh}}{1 + r_{hh}}$$

### Two corrections applied here

- **C-1 (precision floor).** Only learners with $\ge$ `MIN_ITEMS_PER_HALF` observations *in both halves*
  enter the estimate. §10 shows that without this, the coefficient measures sampling noise.
- **C-2 (NaN masking).** Pairwise-complete masking before correlation. A single `NaN` makes
  `pearsonr` return `NaN` for the entire statistic — this silently voided $B_t$ and $M_t$ in v1.0.

A **95% bootstrap confidence interval** is added so the estimate's precision is visible rather than implied.


```python
def _clean_pair(a: pd.Series, b: pd.Series):
    """Align two series and drop non-finite pairs (correction C-2)."""
    idx = a.index.intersection(b.index)
    x, y = a.reindex(idx), b.reindex(idx)
    ok = np.isfinite(x) & np.isfinite(y)
    return x[ok], y[ok]


def bootstrap_ci(x, y, stat_fn, n_boot=1000, alpha=0.05, rng=None):
    rng = rng or np.random.default_rng(SEED)
    if len(x) < 10:
        return (np.nan, np.nan)
    idx = rng.integers(0, len(x), size=(n_boot, len(x)))
    vals = [stat_fn(np.asarray(x)[i], np.asarray(y)[i]) for i in idx]
    vals = np.asarray([v for v in vals if np.isfinite(v)])
    if vals.size == 0:
        return (np.nan, np.nan)
    return tuple(np.quantile(vals, [alpha / 2, 1 - alpha / 2]))


def split_half_reliability(df, ref_builder, min_items, recency_lambda, n_boot=500):
    """Odd/even session split-half reliability with a precision floor. Returns dict per dimension."""
    sessions = np.sort(df["session_id"].unique())
    odd, even = set(sessions[0::2]), set(sessions[1::2])

    d1 = df[df["session_id"].isin(odd)]
    d2 = df[df["session_id"].isin(even)]

    # normalise each half against the FULL analysis population (a stable common reference)
    ref = ref_builder
    p1 = compute_proxies(d1, ref, recency_lambda)
    p2 = compute_proxies(d2, ref, recency_lambda)

    eligible = p1.index[(p1["n_items"] >= min_items)].intersection(
               p2.index[(p2["n_items"] >= min_items)])

    results = {}
    for dim in DIMENSIONS:
        x, y = _clean_pair(p1.loc[eligible, dim], p2.loc[eligible, dim])
        if len(x) < 10 or x.std() == 0 or y.std() == 0:
            results[dim] = dict(rho=None, n=len(x), ci=(np.nan, np.nan))
            continue
        r = stats.pearsonr(x, y)[0]
        rho = 2 * r / (1 + r) if (1 + r) != 0 else np.nan
        lo, hi = bootstrap_ci(x, y,
                              lambda a, b: (lambda rr: 2 * rr / (1 + rr))(np.corrcoef(a, b)[0, 1]),
                              n_boot=n_boot, rng=RNG)
        results[dim] = dict(rho=rho, n=len(x), ci=(lo, hi))
    return results

reliability = split_half_reliability(
    analysis_df, REF, CFG.min_items_per_half, CFG.recency_lambda)

print(f"Split-half reliability (Spearman-Brown), floor = {CFG.min_items_per_half} items/half\n")
for dim in DIMENSIONS:
    r = reliability[dim]
    if r["rho"] is None:
        print(f"  {dim}: insufficient data (n={r['n']})")
    else:
        print(f"  {dim}: {r['rho']:.3f}  [95% CI {r['ci'][0]:.3f}, {r['ci'][1]:.3f}]  n={r['n']:,}")
```

### 13.1 Sensitivity to the precision floor

A single threshold choice should never carry a REJECT verdict on its own. Here reliability is recomputed
across a grid of floors, showing (a) how each dimension's estimate responds to measurement precision and
(b) how many learners each floor retains.

Read this as: *if a dimension's reliability rises steeply with the floor, its v1.0 estimate was
precision-limited; if it stays flat and low, the construct itself is unstable.* That distinction is the
difference between a defensible rejection and an artifact.


```python
sens_rows = []
for floor in CFG.sensitivity_floors:
    res = split_half_reliability(analysis_df, REF, floor, CFG.recency_lambda, n_boot=0)
    row = {"min_items_per_half": floor}
    for dim in DIMENSIONS:
        row[dim] = res[dim]["rho"]
    row["n_learners"] = max(res[d]["n"] for d in DIMENSIONS)
    sens_rows.append(row)

sensitivity = pd.DataFrame(sens_rows).set_index("min_items_per_half")
print("Reliability as a function of the precision floor:\n")
print(sensitivity.round(3).to_string())

fig, ax = plt.subplots(figsize=(9, 5))
for dim in DIMENSIONS:
    ax.plot(sensitivity.index, sensitivity[dim], "o-", lw=2, label=dim)
ax.axhline(CFG.reliability_min, ls="--", c="grey", label=f"threshold {CFG.reliability_min}")
ax.axvline(CFG.min_items_per_half, ls=":", c="#2980b9", label=f"pre-registered floor")
ax.set_xscale("log"); ax.set_xlabel("minimum items per half")
ax.set_ylabel(r"$\rho_{SB}$"); ax.set_title("Reliability vs. measurement precision")
ax.legend(ncol=2, fontsize=9); plt.tight_layout(); plt.show()
```

## 14. Metric 2 — Convergent Validity

$$r_d = \operatorname{corr}\!\left(\hat{S}^{(d)}_\ell,\; y_\ell\right)$$

computed over learners who (i) meet the precision floor in the analysis window and (ii) have a valid
held-out outcome. Both Pearson and Spearman are reported: Pearson assumes linearity, while several proxies
(entropy, recency mass) are strongly skewed, so a large Pearson/Spearman divergence is itself diagnostic.

**Interpretive caution.** $y_\ell$ is later-window correctness. It is therefore *conceptually closest to*
$C_t$, and a high $r_{C}$ is close to tautological (past correctness predicts future correctness). The
convergent-validity criterion is consequently more informative as a **falsifier** for the non-cognitive
dimensions than as confirmation for $C_t$.


```python
def convergent_validity(proxy_df, target, min_items, n_boot=500):
    eligible = proxy_df.index[proxy_df["n_items"] >= min_items]
    results = {}
    for dim in DIMENSIONS:
        x, y = _clean_pair(proxy_df.loc[eligible, dim], target)
        if len(x) < 10 or x.std() == 0 or y.std() == 0:
            results[dim] = dict(r=None, rho_s=None, n=len(x), ci=(np.nan, np.nan))
            continue
        r = stats.pearsonr(x, y)[0]
        rs = stats.spearmanr(x, y)[0]
        lo, hi = bootstrap_ci(x, y, lambda a, b: np.corrcoef(a, b)[0, 1],
                              n_boot=n_boot, rng=RNG)
        results[dim] = dict(r=r, rho_s=rs, n=len(x), ci=(lo, hi))
    return results

convergent = convergent_validity(proxies, outcome, CFG.min_items_per_half)

print("Convergent validity vs. temporally held-out correctness\n")
for dim in DIMENSIONS:
    c = convergent[dim]
    if c["r"] is None:
        print(f"  {dim}: insufficient data")
    else:
        flag = "" if abs(c["r"] - c["rho_s"]) < 0.10 else "   <- Pearson/Spearman diverge (non-linear)"
        print(f"  {dim}: r={c['r']:+.3f} [{c['ci'][0]:+.3f},{c['ci'][1]:+.3f}]  "
              f"rho_s={c['rho_s']:+.3f}  n={c['n']:,}{flag}")
```

## 15. ACT-R Generative Simulator

### Purpose

Real data has no cognitive ground truth. The simulator supplies learners whose latent parameter $\theta$ is
**known by construction**, so a proxy can be tested for whether it recovers what it claims to measure.

### Memory mechanism

Base-level activation with power-law decay:

$$B_i(t) = \ln\!\left(\sum_{j: t_j<t}(t-t_j)^{-\delta}\right), \qquad
P(\text{correct}) = \sigma\!\left(B_i(t) + \theta\right)$$

### Instrumentation map — the decisive methodological point (correction C-3)

Recovery is meaningful **only if the data-generating process depends on $\theta$** for that dimension:

| Dimension | How $\theta$ enters the generative process | Instrumented |
|---|---|---|
| $C_t$ | added to activation → shifts $P(\text{correct})$ | ✔ |
| $B_t$ | scales response-time and attempt-count distributions | ✔ |
| $M_t$ | scales hint-usage distribution | ✔ |
| $G_t$ | narrows the range of skills sampled | ✔ |
| $H_t$ | sets the number of sessions generated | ✔ |
| $I_t$ | **nothing — $\theta$ never enters the process** | ✘ |

Because $I_t$ is not instrumented, its simulated data is statistically independent of $\theta$ and any
correlation is sampling noise. v1.0 reported $r=-0.180$ for it as if it were evidence. Here it is reported
as **not testable**. Instrumenting $I_t$ requires a two-timescale simulator (a stable trait plus
session-varying state), which is deferred to future work rather than silently fabricated.

### Population-level normalisation

All simulated learners are generated into **one pooled frame** and standardised against *that* population
(correction C-5), rather than against real-data constants. This is both correct and considerably faster
than v1.0's per-learner Python loop.


```python
def simulate_population(dimension: str, n_learners: int, cfg: Config, rng) -> tuple:
    """Generate a pooled population of simulated learners for one dimension.

    Returns (long_frame, true_theta_series). Only dimensions in SIMULATOR_INSTRUMENTS
    with value True have data that actually depends on theta.
    """
    thetas, rows = {}, []
    for lid in range(n_learners):
        theta = rng.uniform(0, 1) if dimension in ("M_t", "G_t") else rng.normal(0, 1)
        thetas[lid] = theta

        n_sessions = cfg.sim_sessions
        if dimension == "H_t":                      # theta -> history depth
            n_sessions = int(np.clip(round(4 + theta * 2), 2, 10))

        retrieval, clock = {k: [] for k in range(8)}, 0.0
        for sid in range(n_sessions):
            for i in range(cfg.sim_items_per_session):
                skill = int(rng.integers(0, 8))
                if dimension == "G_t":              # theta -> narrower skill repertoire
                    skill = int(rng.integers(0, max(1, int(8 * (1 - theta)) + 1)))

                past = np.asarray(retrieval[skill], dtype=float)
                past = past[past < clock]
                activation = (np.log(np.sum((clock - past) ** (-cfg.actr_decay)))
                              if past.size else -np.inf)

                shift = theta if dimension == "C_t" else rng.normal(0, 0.3)
                p_correct = 1 / (1 + np.exp(-(activation + shift)))
                correct = rng.binomial(1, float(np.clip(p_correct, 0.01, 0.99)))
                retrieval[skill].append(clock)

                if dimension == "M_t":              # theta -> less help-seeking
                    hint = float(np.clip(rng.beta(2, 5) * (1.3 - theta), 0, 1.3))
                else:
                    hint = float(rng.beta(2, 5))

                if dimension == "B_t":              # theta -> more effortful behaviour
                    rt = float(rng.lognormal(2.5 + theta * 0.4, 0.4))
                    attempts = int(1 + rng.poisson(max(0.0, theta + 1)))
                else:
                    rt = float(rng.lognormal(2.5, 0.4))
                    attempts = int(1 + rng.poisson(1))

                rows.append(dict(learner_id=lid, session_id=sid, skill_id=f"S{skill}",
                                 correct=float(correct), response_time=rt,
                                 hint_count=hint, attempt_count=attempts))
                clock += float(rng.uniform(20, 120))
            clock += cfg.sim_session_gap_days * 86400.0

    return pd.DataFrame(rows), pd.Series(thetas, name="theta")

print("Simulator defined.")
print("Instrumented dimensions:", [d for d, v in SIMULATOR_INSTRUMENTS.items() if v])
print("NOT instrumented      :", [d for d, v in SIMULATOR_INSTRUMENTS.items() if not v])
```

## 16. Metric 3 — Simulator Recovery

For each instrumented dimension we generate a population, compute the proxy on the pooled synthetic data,
and correlate the recovered value against the known $\theta$. The **sign** is checked against
`EXPECTED_SIM_SIGN` (correction C-7): a strong correlation in the *wrong* direction indicates the operator
does not track the construct as theorised, and must not be counted as a pass simply because $|r|$ is large.


```python
def simulator_recovery(cfg: Config, rng, n_boot=500):
    results = {}
    for dim in DIMENSIONS:
        if not SIMULATOR_INSTRUMENTS[dim]:
            results[dim] = dict(r=None, n=0, ci=(np.nan, np.nan),
                                note="not instrumented -- recovery undefined")
            continue

        sim_df, theta = simulate_population(dim, cfg.n_sim_learners, cfg, rng)
        sim_ref = normalisation_reference(sim_df)          # standardise within the SIM population
        sim_proxies = compute_proxies(sim_df, sim_ref, cfg.recency_lambda)

        x, y = _clean_pair(sim_proxies[dim], theta)
        if len(x) < 10 or x.std() == 0:
            results[dim] = dict(r=None, n=len(x), ci=(np.nan, np.nan), note="degenerate")
            continue
        r = stats.pearsonr(x, y)[0]
        lo, hi = bootstrap_ci(x, y, lambda a, b: np.corrcoef(a, b)[0, 1], n_boot=n_boot, rng=rng)

        expected = EXPECTED_SIM_SIGN[dim]
        sign_ok = (expected is None) or (np.sign(r) == np.sign(expected))
        results[dim] = dict(r=r, n=len(x), ci=(lo, hi), sign_ok=sign_ok,
                            note="" if sign_ok else f"SIGN MISMATCH (expected {expected:+d})")
    return results

recovery = simulator_recovery(CFG, RNG)

print("Simulator recovery (correlation between known theta and recovered proxy)\n")
for dim in DIMENSIONS:
    s = recovery[dim]
    if s["r"] is None:
        print(f"  {dim}: n/a  -- {s['note']}")
    else:
        print(f"  {dim}: r={s['r']:+.3f} [{s['ci'][0]:+.3f},{s['ci'][1]:+.3f}]  "
              f"n={s['n']}  {s['note']}")
```

## 17. Decision Rule and the Audit Table

### Rule

Let $\rho$ be reliability, $r_c$ convergent validity, $r_s$ simulator recovery.

$$
\text{verdict}(d)=
\begin{cases}
\textbf{REJECT} & \rho < \rho_{\min} \ \text{ or } \ \rho \text{ undefined}\\[4pt]
\textbf{INCONCLUSIVE} & \rho \ge \rho_{\min},\ \text{validity criterion unavailable}\\[4pt]
\textbf{RETAIN} & \rho \ge \rho_{\min}\ \text{ and }\ |r| \ge r_{\min}\ \text{ and sign as expected}\\[4pt]
\textbf{PROXY-ONLY} & \rho \ge \rho_{\min}\ \text{ but validity criterion not met}
\end{cases}
$$

**INCONCLUSIVE is a new category** (correction C-3). A dimension that is reliably measured but for which no
validity test could be run is *not* the same as one that failed a test, and collapsing the two — as v1.0
did for $I_t$ — misrepresents the evidence.

Which criterion applies is set by `HAS_REAL_TARGET`: dimensions with a defensible real-world outcome are
judged on convergent validity; the rest on simulator recovery.


```python
def decide(dim, rel, conv, rec, cfg) -> tuple:
    rho = rel["rho"]
    if rho is None or not np.isfinite(rho):
        return "REJECT", "reliability could not be estimated"
    if rho < cfg.reliability_min:
        return "REJECT", f"reliability {rho:.3f} < {cfg.reliability_min}"

    if HAS_REAL_TARGET[dim]:
        r, kind, thr = conv["r"], "convergent", cfg.convergent_r_min
    else:
        r, kind, thr = rec["r"], "simulator", cfg.simulator_r_min

    if r is None or not np.isfinite(r):
        return "INCONCLUSIVE", f"reliable, but {kind} validity not testable"

    if kind == "simulator":
        expected = EXPECTED_SIM_SIGN[dim]
        if expected is not None and np.sign(r) != np.sign(expected):
            return "PROXY-ONLY", f"recovery |r|={abs(r):.3f} but sign contradicts theory"

    if abs(r) >= thr:
        return "RETAIN", f"reliable and {kind} |r|={abs(r):.3f} >= {thr}"
    return "PROXY-ONLY", f"reliable but {kind} |r|={abs(r):.3f} < {thr}"


def fmt(v, spec="+.3f"):
    return "n/a" if v is None or (isinstance(v, float) and not np.isfinite(v)) else format(v, spec)

rows = []
for dim in DIMENSIONS:
    rel, conv, rec = reliability[dim], convergent[dim], recovery[dim]
    v, why = decide(dim, rel, conv, rec, CFG)
    rows.append({
        "Component": dim,
        "Reliability (SB)": fmt(rel["rho"], ".3f"),
        "Convergent r": fmt(conv["r"]),
        "Simulator r": fmt(rec["r"]),
        "Criterion": "convergent" if HAS_REAL_TARGET[dim] else "simulator",
        "Verdict": v,
        "Basis": why,
    })

audit_table = pd.DataFrame(rows)
print(audit_table.to_markdown(index=False))
```

```python
# Manuscript-ready Table 1 (drop the diagnostic columns)
table1 = audit_table[["Component", "Reliability (SB)", "Convergent r", "Simulator r", "Verdict"]]
print("\n--- Manuscript Table 1 ---\n")
print(table1.to_markdown(index=False))

audit_table.to_csv("rq2_1_audit_full.csv", index=False)
table1.to_csv("rq2_1_table1.csv", index=False)
sensitivity.to_csv("rq2_1_reliability_sensitivity.csv")
print("\nsaved: rq2_1_audit_full.csv, rq2_1_table1.csv, rq2_1_reliability_sensitivity.csv")
```

## 18. Results Visualisation

Each dimension is shown against both decision thresholds with bootstrap confidence intervals, so a
borderline result is visibly borderline rather than being hidden behind a point estimate.


```python
fig, axes = plt.subplots(1, 3, figsize=(17, 4.8))
xs = np.arange(len(DIMENSIONS))
palette = {"RETAIN": "#27ae60", "PROXY-ONLY": "#f39c12",
           "INCONCLUSIVE": "#7f8c8d", "REJECT": "#c0392b"}
verdicts = dict(zip(audit_table["Component"], audit_table["Verdict"]))
colors = [palette[verdicts[d]] for d in DIMENSIONS]

def panel(ax, values, cis, title, threshold, ylabel):
    vals = [v if (v is not None and np.isfinite(v)) else np.nan for v in values]
    ax.bar(xs, vals, color=colors, edgecolor="black", lw=.6)
    for i, (v, ci) in enumerate(zip(vals, cis)):
        if np.isfinite(v) and np.isfinite(ci[0]):
            ax.plot([i, i], ci, color="black", lw=1.4)
    for i, v in enumerate(vals):
        if not np.isfinite(v):
            ax.text(i, 0.02, "n/a", ha="center", fontsize=9, style="italic", color="#555")
    ax.axhline(threshold, ls="--", c="grey", lw=1.2)
    ax.axhline(0, c="black", lw=.8)
    ax.set_xticks(xs); ax.set_xticklabels(DIMENSIONS)
    ax.set_title(title); ax.set_ylabel(ylabel)

panel(axes[0], [reliability[d]["rho"] for d in DIMENSIONS],
      [reliability[d]["ci"] for d in DIMENSIONS],
      "Reliability (Spearman-Brown)", CFG.reliability_min, r"$\rho_{SB}$")
panel(axes[1], [convergent[d]["r"] for d in DIMENSIONS],
      [convergent[d]["ci"] for d in DIMENSIONS],
      "Convergent validity", CFG.convergent_r_min, "r")
panel(axes[2], [recovery[d]["r"] for d in DIMENSIONS],
      [recovery[d]["ci"] for d in DIMENSIONS],
      "Simulator recovery", CFG.simulator_r_min, "r")

handles = [plt.Rectangle((0, 0), 1, 1, color=c, ec="black", lw=.6) for c in palette.values()]
fig.legend(handles, palette.keys(), loc="lower center", ncol=4,
           bbox_to_anchor=(0.5, -0.10), frameon=False)
plt.suptitle("RQ2.1 construct-validity audit of candidate PCDT dimensions", y=1.04, fontsize=13)
plt.tight_layout(); plt.show()
```

## 19. Interpretation

> Read the table produced above, not this section, for the numbers. The guidance below states how each
> possible outcome should be interpreted — written before the run so interpretation is not fitted to
> whatever appeared.

### $C_t$ — cognition / mastery
The only dimension with three decades of converging evidence behind it. If it now passes at the corrected
precision floor while failing at floor = 1 (visible in §13.1), that is direct confirmation that the v1.0
REJECT was a measurement artifact. Note the near-tautology caveat from §14: a strong $r_C$ against later
correctness is partly definitional and should not be reported as strong construct validation.

### $B_t$ — behaviour
Directly observed rather than inferred, so high reliability is expected and is *not* evidence about
cognition. A negative convergent correlation is theoretically coherent — greater time-on-task and more
attempts typically indicate difficulty, not proficiency. Interpret sign, not magnitude alone.

### $M_t$ — metacognition
The weakest theoretical link in the audit. `fraction_of_hints_used` measures *help-seeking*, and inverting
it presumes that seeking less help indicates better self-regulation. The self-regulated-learning literature
supports the opposite reading at least as strongly: *adaptive* help-seeking is a hallmark of skilled
metacognition. A pass here should be reported as "a reliable help-seeking indicator", never as
"metacognition measured".

### $G_t$ — goals
The decisive consideration is structural, not statistical: **ASSISTments content is teacher-assigned**.
Skill-selection entropy therefore largely reflects the teacher's assignment breadth. Even a statistically
clean pass does not license a *learner-goal* interpretation on this dataset, and the sign check (§16) guards
against the v1.0 outcome in which $G_t$ was retained on a large negative correlation that was purely a
simulator parameterisation artifact.

### $H_t$ — historical memory
Effectively a recency-weighted engagement-volume measure, so the honest reading is "engagement mass", not
memory. Convergent validity is expected to be weak because volume is not attainment.

> **Design interaction discovered during verification — read before interpreting any $H_t$ reliability
> figure.** $\pi_H$ weights each row by its recency *relative to the learner's last session within the
> slice being scored*, i.e. $s^{\max}_\ell$ is recomputed inside each half. Splitting sessions odd/even
> therefore changes the weighting origin between halves, so the two half-scores are not measuring on a
> common scale. A low split-half $\rho_{SB}$ for $H_t$ is consequently **partly an artifact of applying a
> session-partitioned reliability design to a session-relative statistic**, and the sensitivity curve for
> $H_t$ may *decline* as the floor rises (more sessions ⇒ more divergent weighting origins). Before
> rejecting $H_t$ on reliability grounds, re-estimate it either with a global time origin
> ($s^{\max}$ fixed across the full history) or with a test–retest design over contiguous
> rather than interleaved blocks. This is a limitation of the estimator, not evidence about the construct.

### $I_t$ — identity
Reliable-but-not-testable is the most likely outcome, and **INCONCLUSIVE is the correct verdict** for it.
The construct requires separating a stable trait from session-varying state, which needs either a
two-timescale simulator or longitudinal data spanning far more than one term.

### On $U_t$
Excluded by design, not by failure. Under the corrected framework it is the dispersion of $b_t$, so it
becomes measurable only once Paper 3a's generative inference model exists.


## 20. Limitations and Threats to Validity

### Construct validity
1. **No cognitive ground truth exists in the data.** Every real-data conclusion concerns proxy behaviour,
   never latent cognition.
2. **The outcome is not an independent post-test** — it is later-window correctness on the same platform.
   Describing it otherwise in a manuscript would be a material misrepresentation.
3. **Proxy specifications are hand-designed**, and plausible alternatives exist for every dimension
   (e.g. BKT-derived mastery for $C_t$, adaptive-help-seeking for $M_t$). Verdicts are conditional on these
   specific operationalisations.
4. **Log-only data.** Studies reporting stronger affective and metacognitive measurement use multimodal
   signals (gaze, facial, physiological) absent here. Conclusions do not transfer to multimodal deployments.

### Internal validity
5. **Sessions are derived, not observed.** The 4-hour threshold propagates into every reliability estimate;
   robustness to $\tau$ is not tested here.
6. **Missing-correctness deletion is non-random** — ungraded and open-response items are removed
   preferentially, plausibly shifting the sample toward auto-gradable content.
7. **The skill explode-join duplicates rows** for multi-skill problems, up-weighting them in
   learner-level means.
8. **The precision floor trades coverage for precision**, retaining a more engaged sub-population; §10's
   third panel quantifies this trade-off.
9. **The split-half design is not scale-invariant for $H_t$.** Because $\pi_H$ is defined relative to the
   last session *within the scored slice*, odd/even partitioning shifts its weighting origin between halves.
   $H_t$ reliability is therefore estimator-dependent and should be re-derived with a fixed time origin or a
   contiguous-block test–retest design before any verdict on $H_t$ is reported (see §19).

### Simulator validity
10. **The simulator embodies assumptions rather than testing them.** Recovery shows a proxy can invert
   *this* generative process; if the process is wrong, recovery is uninformative about real learners.
11. **Single-exponential/power-law decay is contested** by recent retention research suggesting phased
    forgetting, so $H_t$-related recovery is assumption-dependent.
12. **$I_t$ is not instrumented at all** — reported as not testable rather than assigned a noise value.

### Statistical
13. **Thresholds are conventional, not derived** from a decision-theoretic loss.
14. **No multiplicity correction** across six dimensions × three metrics; CIs are descriptive.
15. **Very large $n$ on real data** makes trivially small correlations significant — hence effect sizes and
    CIs are reported throughout rather than $p$-values.


## 21. Reproducibility Notes

  v1.0 mixed a seeded generator with unseeded global `np.random` calls, so its simulator "true values" were

  reliability split partitions sessions disjointly.

## 22. Conclusion

This notebook implements the RQ2.1 construct-validity audit as a genuinely falsifiable test of the candidate
PCDT dimensions, and corrects three errors that materially changed the v1.0 results:

1. Reliability was **precision-limited, not construct-limited** — $C_t$'s REJECT verdict was an artifact of
   including learners with ~3 observations per half (§10, §13.1).
2. `NaN` propagation **silently voided** $B_t$ and $M_t$ rather than reporting them.
3. Simulator recovery was reported for $I_t$, a dimension the simulator **never instruments**, turning
   sampling noise into apparent evidence.

The corrected pipeline additionally normalises each population against itself, enforces a pre-registered
precision floor with an accompanying sensitivity analysis, checks recovery signs against theory, and
introduces **INCONCLUSIVE** so that "untested" is never reported as "failed".

### Output feeding Paper 3a

The retained dimension set defines the state space over which Paper 3a constructs generative belief-state
inference $b_t = P(S_t \mid O_{1:t}, A_{1:t-1})$. This ordering is not merely convenient: inference over an
unaudited representation could not distinguish *"the model cannot infer this dimension"* from *"this
dimension was never defensibly measurable."*
