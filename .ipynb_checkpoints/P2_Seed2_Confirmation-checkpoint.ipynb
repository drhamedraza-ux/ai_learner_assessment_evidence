# Paper 2 — Junyi seed-2 confirmation (final experimental run)

Everything else in the revision is settled. This notebook resolves the last open question.

## The question

Three of the paper's conclusions rest on one comparison — PCDT versus DKVMN, on Junyi, at seed 20260202:

1. the persistence operator Ψ confers **no retention advantage** over an established memory-augmented tracer;
2. the within-session and cross-session model rankings **invert** — the paper's most novel claim;
3. the educational advantage **does not survive** against the strongest baseline.

One model, one dataset, one seed. If that run is anomalous, all three change. This notebook re-runs Junyi
at seed 20260203 and tests whether each conclusion holds.

## Decision rules, fixed before the numbers are seen

| Finding | Confirmed only if |
|---|---|
| **Retention** | every adequate PCDT-vs-DKVMN cell keeps its **sign** in both seeds **and** both 95% CIs exclude zero |
| **Ranking inversion** | in **both** seeds, PCDT ranks first within session **and** DKVMN ranks first at ≥1 real gap horizon |
| **Educational utility** | the count of strict wins vs DKVMN (more correct flags **and** no extra false alarms) stays at the same qualitative level in both seeds |

Anything that fails is reported as **NOT CONFIRMED**, and the manuscript must then treat the finding as
unsettled rather than weakened. No result is edited, smoothed, or reinterpreted to fit.

## Cost and recomputation

Step 2 is the only expensive part (~160 min). Everything else reads saved artefacts. If
`REV_junyi_seed2` already exists the training is **skipped** unless `FORCE_RERUN = True`, so the notebook
is safe to re-execute.

## 1. Configuration

```python
import os, sys, json, time, glob, warnings, re
import numpy as np, pandas as pd
warnings.filterwarnings("ignore", category=FutureWarning)

CACHE       = "/home/hamed/Downloads/Research/P02/PCDT_P2_cache_rev_V3"
PIPELINE_NB = "/home/hamed/Downloads/Research/P02/P2_Revision_Pipeline_v3_junyi.ipynb"
DATASET     = "junyi"

RUN_STEP2   = True      # the ~160 min seed-2 training run
RUN_STEP3   = True      # cross-seed comparison: retention, ranking inversion, utility
RUN_STEP4   = True      # consolidated final numbers
FORCE_RERUN = False     # True = retrain even if REV_junyi_seed2 already exists

# analysis constants -- identical to the primary run, do not change
BUCKETS    = [("within",0,20*60),("sameday",20*60,86400),("d1",86400,7*86400),
              ("d7",7*86400,28*86400),("d28",28*86400,np.inf)]
REAL_GAPS  = ["d1","d7","d28"]
THRESHOLDS = (0.3,0.4,0.5,0.6,0.7)
MIN_POINTS = 2_000
MMDE       = 0.010
BASELINES  = ["DKT","SAKT","AKT","DKVMN","DKT-F"]

RES  = os.path.join(CACHE,"results")
S1   = os.path.join(RES,f"REV_{DATASET}")            # primary seed
S2   = os.path.join(RES,f"REV_{DATASET}_seed2")      # confirmation seed
OUT  = os.path.join(RES,"SEED2_CONFIRMATION"); os.makedirs(OUT,exist_ok=True)
pd.set_option("display.width",220); pd.set_option("display.max_columns",40)

ERRORS = []          # anything that prevents a conclusion is appended here and reported at the end
def flag(msg):
    ERRORS.append(msg); print("  !! " + msg)

print("primary  :", S1)
print("seed-2   :", S2, "(exists)" if os.path.isdir(S2) else "(to be created)")
print("outputs  :", OUT)
```

## 2. Pre-flight

```python
ok = True
for p,label in [(S1,"primary Junyi run"),
                (os.path.join(S1,"tables","table5_persistence.csv"),"primary table5"),
                (os.path.join(S1,"predictions_test.npz"),"primary predictions")]:
    if not os.path.exists(p):
        flag(f"missing {label}: {p}"); ok = False

if RUN_STEP2 and not (os.path.isdir(S2) and not FORCE_RERUN):
    if not os.path.exists(PIPELINE_NB):
        flag(f"PIPELINE_NB not found: {PIPELINE_NB}"); ok = False
    found_loader = any(os.path.exists(os.path.join(d,"pcdt_datasets.py"))
                       for d in [os.path.dirname(CACHE), CACHE, os.path.dirname(PIPELINE_NB)])
    if not found_loader:
        flag("pcdt_datasets.py not found near the cache or pipeline -- the pipeline import will fail")
        ok = False

print("PRE-FLIGHT:", "OK" if ok else "FAILED -- fix the items above before continuing")
if not ok: raise SystemExit("pre-flight failed")
```

## 3. Step 2 — Junyi at seed 20260203

Runs the v3 pipeline with **only** `dataset` and `seed_index` patched. The patch is asserted, so a silent
failure to apply it cannot happen. Output goes to `REV_junyi_seed2`; the primary run is never touched.

```python
SEED2_RUNTIME = None
if RUN_STEP2:
    if os.path.isdir(S2) and not FORCE_RERUN:
        print(f"REV_{DATASET}_seed2 already exists -- skipping the training run.")
        print("Set FORCE_RERUN = True to retrain from scratch.")
    else:
        t0 = time.time()
        nbj = json.load(open(PIPELINE_NB))
        src = "\n".join("".join(c["source"]) for c in nbj["cells"] if c["cell_type"]=="code")
        before = src
        src = re.sub(r'dataset:\s*str\s*=\s*"[^"]+"', f'dataset: str = "{DATASET}"', src, count=1)
        src = re.sub(r'seed_index:\s*int\s*=\s*\d+',   'seed_index: int = 1',        src, count=1)
        assert src != before, "config patch did not apply -- check the Config field names in the v3 notebook"
        assert 'seed_index: int = 1' in src, "seed_index patch missing"
        src = "\n".join(l for l in src.split("\n") if not l.strip().startswith("!"))
        print(f"running the v3 pipeline: dataset={DATASET}, seed_index=1 ...")
        ns = {"__name__":"__main__"}
        exec(compile(src,"<v3-seed2>","exec"), ns)
        SEED2_RUNTIME = (time.time()-t0)/60
        print(f"\nseed-2 run complete in {SEED2_RUNTIME:.1f} min")
else:
    print("RUN_STEP2 = False -- skipping")

if not os.path.isdir(S2):
    flag("REV_junyi_seed2 was not produced -- Steps 3 and 4 cannot draw conclusions")
```

## 4. Shared helpers

Post-gap AUC is computed per model per bucket directly from the saved predictions, so the ranking test
covers all eight models rather than only the pairs stored in `table5`.

```python
from sklearn.metrics import roc_auc_score, confusion_matrix

def load_preds(run_dir):
    p = os.path.join(run_dir,"predictions_test.npz")
    if not os.path.exists(p): return None
    z = np.load(p)
    return dict(z=z, models=[k for k in z.files if k not in ("targets","mask","bucket")],
                Y=z["targets"], M=z["mask"]>0, B=z["bucket"])

def auc_by_bucket(pk):
    """AUC per model per bucket. Returns tidy DataFrame."""
    rows=[]
    for bid,(name,lo,hi) in enumerate(BUCKETS):
        sel = (pk["B"]==bid) & pk["M"]
        n = int(sel.sum())
        if n < MIN_POINTS: continue
        y = pk["Y"][sel]
        if len(np.unique(y))<2: continue
        for m in pk["models"]:
            rows.append(dict(horizon=name, model=m, n_points=n,
                             auc=float(roc_auc_score(y, pk["z"][m][sel]))))
    return pd.DataFrame(rows)

def overall_auc(pk):
    rows=[]
    y = pk["Y"][pk["M"]]
    for m in pk["models"]:
        rows.append(dict(model=m, auc=float(roc_auc_score(y, pk["z"][m][pk["M"]]))))
    return pd.DataFrame(rows).sort_values("auc",ascending=False)

def utility(p,y,thr):
    flag_ = (p<thr).astype(int); need=(y==0).astype(int)
    tn,fp,fn,tp = confusion_matrix(need,flag_,labels=[0,1]).ravel()
    prec = tp/(tp+fp) if tp+fp else np.nan
    rec  = tp/(tp+fn) if tp+fn else np.nan
    return dict(correctly_flagged=int(tp), false_alarms=int(fp), precision=prec, recall=rec)

def utility_vs(pk, comparator):
    rows=[]
    if comparator not in pk["models"]: return pd.DataFrame()
    for bid,(name,lo,hi) in enumerate(BUCKETS):
        if name not in REAL_GAPS: continue
        sel=(pk["B"]==bid)&pk["M"]
        if sel.sum()<MIN_POINTS: continue
        for thr in THRESHOLDS:
            a=utility(pk["z"]["PCDT"][sel],pk["Y"][sel],thr)
            b=utility(pk["z"][comparator][sel],pk["Y"][sel],thr)
            rows.append(dict(horizon=name, comparator=comparator, threshold=thr,
                extra_correctly_flagged=a["correctly_flagged"]-b["correctly_flagged"],
                extra_false_alarms=a["false_alarms"]-b["false_alarms"],
                d_precision=a["precision"]-b["precision"]))
    return pd.DataFrame(rows)

P1 = load_preds(S1); P2 = load_preds(S2)
if P1 is None: flag("primary predictions missing")
if P2 is None: flag("seed-2 predictions missing")
print("primary models:", P1["models"] if P1 else "-")
print("seed-2  models:", P2["models"] if P2 else "-")
```

## 5. Finding 1 — retention (PCDT vs DKVMN)

Confirmed only if every adequate cell keeps its sign in both seeds and both CIs exclude zero.

```python
V_RET = {}
if RUN_STEP3:
    f1 = os.path.join(S1,"tables","table5_persistence.csv")
    f2 = os.path.join(S2,"tables","table5_persistence.csv")
    if not (os.path.exists(f1) and os.path.exists(f2)):
        flag("table5_persistence.csv missing for one or both seeds -- retention cannot be confirmed")
    else:
        a=pd.read_csv(f1); b=pd.read_csv(f2); k=["horizon","comparator"]
        cols=k+["retention_advantage","ci_lo","ci_hi","adequate"]
        m=a[cols].merge(b[cols],on=k,suffixes=("_s1","_s2"))
        m["sign_agrees"]=np.sign(m.retention_advantage_s1)==np.sign(m.retention_advantage_s2)
        m["s1_excl_zero"]=m.ci_lo_s1*m.ci_hi_s1>0
        m["s2_excl_zero"]=m.ci_lo_s2*m.ci_hi_s2>0
        m["abs_change"]=(m.retention_advantage_s2-m.retention_advantage_s1).abs()
        m.to_csv(f"{OUT}/retention_cross_seed.csv",index=False)
        print(m.round(4).to_string(index=False))

        dk=m[(m.comparator=="DKVMN") & m.adequate_s1.fillna(False) & m.adequate_s2.fillna(False)]
        if not len(dk):
            flag("no adequate PCDT-vs-DKVMN cells in both seeds"); V_RET=dict(confirmed=None)
        else:
            conf=bool(dk.sign_agrees.all() and dk.s1_excl_zero.all() and dk.s2_excl_zero.all())
            V_RET=dict(confirmed=conf, n_cells=int(len(dk)),
                       max_abs_change=float(dk.abs_change.max()),
                       horizons=dk.horizon.tolist(),
                       s1=dk.retention_advantage_s1.round(4).tolist(),
                       s2=dk.retention_advantage_s2.round(4).tolist())
            print("\nFINDING 1 -- retention:", "CONFIRMED" if conf else "NOT CONFIRMED")
            print(f"  cells={len(dk)} | largest change between seeds={dk.abs_change.max():.4f}")
            if not conf:
                print("  -> Do NOT report the DKVMN retention comparison as settled. Run a third seed.")
```

## 6. Finding 2 — ranking inversion

The paper's most novel claim: PCDT ranks first *within session*, DKVMN ranks first *across gaps*.
Tested independently in each seed, across all eight models.

```python
V_INV = {}
if RUN_STEP3 and P1 and P2:
    tabs={}
    for tag,pk in [("s1",P1),("s2",P2)]:
        ov=overall_auc(pk); gp=auc_by_bucket(pk)
        ov.to_csv(f"{OUT}/within_session_ranking_{tag}.csv",index=False)
        gp.to_csv(f"{OUT}/postgap_auc_by_model_{tag}.csv",index=False)
        tabs[tag]=(ov,gp)
        print(f"--- {tag}: within-session ranking ---")
        print(ov.round(4).to_string(index=False))
        print(f"--- {tag}: post-gap AUC, real horizons ---")
        g=gp[gp.horizon.isin(REAL_GAPS)]
        if len(g):
            piv=g.pivot(index="model",columns="horizon",values="auc").round(4)
            print(piv.to_string())
        print()

    res={}
    for tag,(ov,gp) in tabs.items():
        top_within = ov.iloc[0]["model"]
        gaps = gp[gp.horizon.isin(REAL_GAPS)]
        top_by_gap = {h: gaps[gaps.horizon==h].sort_values("auc",ascending=False).iloc[0]["model"]
                      for h in gaps.horizon.unique()}
        pcdt_first = top_within=="PCDT"
        dkvmn_gap  = [h for h,mm in top_by_gap.items() if mm=="DKVMN"]
        res[tag]=dict(top_within=top_within, top_by_gap=top_by_gap,
                      pcdt_first_within=bool(pcdt_first), dkvmn_first_gaps=dkvmn_gap,
                      inversion=bool(pcdt_first and len(dkvmn_gap)>0))
    conf = bool(res["s1"]["inversion"] and res["s2"]["inversion"])
    V_INV=dict(confirmed=conf, per_seed=res)
    json.dump(V_INV, open(f"{OUT}/verdict_ranking_inversion.json","w"), indent=2)
    print("FINDING 2 -- ranking inversion:", "CONFIRMED in both seeds" if conf else "NOT CONFIRMED")
    for tag in ("s1","s2"):
        r=res[tag]
        print(f"  {tag}: within-session top = {r['top_within']}; "
              f"DKVMN tops gaps {r['dkvmn_first_gaps'] or 'none'}")
    if not conf:
        print("  -> The inversion claim is not stable across seeds. Report it as dataset- and")
        print("     seed-specific, or drop it. Do not present it as a general finding.")
```

## 7. Finding 3 — educational utility vs DKVMN

Strict win = more learners correctly flagged **and** no extra false alarms. Recomputed for seed 2 and
compared with the primary run.

```python
V_UTIL = {}
if RUN_STEP3 and P1 and P2:
    frames=[]
    for tag,pk in [("s1",P1),("s2",P2)]:
        for comp in ["Control","DKT-F","DKVMN"]:
            u=utility_vs(pk,comp)
            if len(u): u["seed"]=tag; frames.append(u)
    if not frames:
        flag("no utility rows computed")
    else:
        U=pd.concat(frames,ignore_index=True)
        U.to_csv(f"{OUT}/utility_cross_seed.csv",index=False)
        summ=[]
        for tag in ("s1","s2"):
            for comp in ["Control","DKT-F","DKVMN"]:
                sub=U[(U.seed==tag)&(U.comparator==comp)]
                if not len(sub): continue
                strict=sub[(sub.extra_correctly_flagged>0)&(sub.extra_false_alarms<=0)]
                summ.append(dict(seed=tag,comparator=comp,cells=len(sub),strict_wins=len(strict)))
        S=pd.DataFrame(summ)
        print(S.to_string(index=False))
        dk=S[S.comparator=="DKVMN"]
        if len(dk)==2:
            w1,w2=int(dk[dk.seed=='s1'].strict_wins.iloc[0]), int(dk[dk.seed=='s2'].strict_wins.iloc[0])
            same_level = (w1==0)==(w2==0)
            V_UTIL=dict(confirmed=bool(same_level), strict_wins_s1=w1, strict_wins_s2=w2)
            print(f"\nFINDING 3 -- utility vs DKVMN: seed1 {w1} strict wins, seed2 {w2}")
            print("  ", "CONSISTENT across seeds" if same_level else "NOT CONSISTENT across seeds")
            if same_level and w1==0:
                print("  -> Educational claim stays scoped to the session-resetting control.")
            elif not same_level:
                print("  -> Utility conclusion is seed-dependent; report as unsettled.")
        json.dump(V_UTIL, open(f"{OUT}/verdict_utility_cross_seed.json","w"), indent=2)
```

## 8. Step 4 — consolidated verdict and final numbers

```python
if RUN_STEP4:
    VERDICT=dict(
        generated=time.strftime("%Y-%m-%d %H:%M"),
        dataset=DATASET, seed2_runtime_min=SEED2_RUNTIME,
        decision_rules=dict(
            retention="every adequate PCDT-vs-DKVMN cell keeps its sign in both seeds and both CIs exclude zero",
            ranking_inversion="PCDT first within session AND DKVMN first at >=1 real gap horizon, in both seeds",
            utility="strict-win count vs DKVMN at the same qualitative level in both seeds",
            mmde=MMDE),
        finding_1_retention=V_RET, finding_2_ranking_inversion=V_INV,
        finding_3_utility=V_UTIL, errors=ERRORS)
    for tag,d in [("s1",S1),("s2",S2)]:
        rp=os.path.join(d,"reproducibility.json")
        if os.path.exists(rp):
            r=json.load(open(rp))
            VERDICT[f"provenance_{tag}"]={k:r.get(k) for k in
                ("seed","seed_index","strongest_baseline","runtime_minutes","n_learners","n_kcs",
                 "horizons_real","horizons_inadequate")}
    json.dump(VERDICT, open(f"{OUT}/verdict_seed.json","w"), indent=2, default=str)
    print("wrote", f"{OUT}/verdict_seed.json")
```

## 9. Final summary

```python
def status(v):
    if not v or v.get("confirmed") is None: return "NOT DETERMINED"
    return "CONFIRMED" if v.get("confirmed") else "NOT CONFIRMED"

print("="*78); print("SEED-2 CONFIRMATION -- FINAL SUMMARY"); print("="*78)
print(f"dataset        : {DATASET}")
print(f"primary seed   : 20260202   ({S1})")
print(f"confirmation   : 20260203   ({S2})")
if SEED2_RUNTIME: print(f"seed-2 runtime : {SEED2_RUNTIME:.1f} min")
print()
print(f"1. Retention (PCDT vs DKVMN)  : {status(V_RET)}")
if V_RET.get("n_cells"):
    print(f"     cells={V_RET['n_cells']} horizons={V_RET.get('horizons')}")
    print(f"     seed1={V_RET.get('s1')}  seed2={V_RET.get('s2')}")
    print(f"     largest between-seed change={V_RET.get('max_abs_change'):.4f}")
print(f"2. Ranking inversion          : {status(V_INV)}")
if V_INV.get("per_seed"):
    for t in ("s1","s2"):
        r=V_INV["per_seed"][t]
        print(f"     {t}: within-top={r['top_within']}, DKVMN tops {r['dkvmn_first_gaps'] or 'no'} gap(s)")
print(f"3. Utility vs DKVMN           : {status(V_UTIL)}")
if V_UTIL: print(f"     strict wins: seed1={V_UTIL.get('strict_wins_s1')}, seed2={V_UTIL.get('strict_wins_s2')}")
print()
print("files written:")
for f in sorted(os.listdir(OUT)): print("   ", os.path.join(OUT,f))
print()
if ERRORS:
    print("ERRORS / MISSING INPUTS -- conclusions above may be incomplete:")
    for e in ERRORS: print("   -", e)
else:
    print("no errors; every conclusion above is backed by a file in the folder listed.")
print()
print("Send the SEED2_CONFIRMATION folder back. verdict_seed.json carries the three verdicts and")
print("the decision rules they were judged against.")
```
