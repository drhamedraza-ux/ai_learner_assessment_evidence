# Paper 2 — Seed-3 confirmation of the ranking-inversion finding

## Pre-registration

**This section is written before seed 20260204 has been run. Do not edit it afterwards.**

### The hypothesis, and where it came from

Seeds 20260202 and 20260203 on Junyi both showed the same pattern: the model a practitioner would select
on within-session AUC is *not* the model that performs best after a gap. Those two runs are the
**exploratory basis**. They generated the hypothesis; they cannot also confirm it, because the rule below
was specified after seeing them.

**Seed 20260204 is the confirmatory test.** It is the only out-of-sample evidence in this notebook, and
the verdict rests on it.

### Why the earlier rule failed, and what changed

The original rule required *PCDT ranks first within session*. It failed on seed 2 — not because the
phenomenon vanished, but because `PCDT` (0.7476) and `PCDT-noPsi` (0.7480) are statistically
indistinguishable within session (tied to four decimals on seed 1), so which one tops the table is
arbitrary. The rule was sensitive to a coin flip between two variants of the same model.

The revised rule is stated in terms of **DKVMN's own position**, which is what the finding is actually
about, and does not depend on internal ordering within the PCDT family.

### PRIMARY rule — rank-based, pre-registered

> The inversion is **CONFIRMED** on seed 20260204 if, in that run:
> 1. **DKVMN ranks outside the top three** of the eight models on within-session AUC; **and**
> 2. **DKVMN ranks first** on post-gap AUC at **every** real (adequately powered) gap horizon.

Rank rather than magnitude, deliberately: a practitioner selecting a model *chooses the top-ranked one*,
so rank is the decision-relevant quantity. Both legs must hold. Failure of either → **NOT CONFIRMED**.

### SECONDARY rule — magnitude, with confidence intervals

> The **sign flip** is supported if the learner-clustered bootstrap interval for
> `AUC(PCDT) − AUC(DKVMN)` is **positive within session** and **negative at ≥1 real gap horizon**.

This tests the same phenomenon as a magnitude rather than an ordering, and is reported alongside.

### An honesty note recorded in advance

The within-session lead of the PCDT family over DKVMN was **+0.0086** on seed 1 and **+0.0078** on seed 2
— both **below this paper's own 0.010 minimum meaningful difference**. The inversion must therefore be
reported as a *model-selection* consequence, not as a claim that PCDT is meaningfully better within
session. The manuscript wording must say so explicitly. This is written here so it cannot be quietly
dropped once the result is known.

### Cost

~165 minutes. Everything after the run reads saved artefacts.

## 1. Configuration

```python
import os, sys, json, time, re, warnings
import numpy as np, pandas as pd
warnings.filterwarnings("ignore", category=FutureWarning)

CACHE       = "/home/hamed/Downloads/Research/P02/PCDT_P2_cache_rev_V3"
PIPELINE_NB = "/home/hamed/Downloads/Research/P02/P2_Revision_Pipeline_v3_junyi.ipynb"
DATASET     = "junyi"

RUN_SEED3   = True       # the ~165 min run at seed 20260204
FORCE_RERUN = False      # True = retrain even if REV_junyi_seed3 exists

# pre-registered constants -- do not change
MIN_POINTS   = 2_000
REAL_GAPS    = ["d1","d7","d28"]
MMDE         = 0.010
N_BOOT       = 2_000
TOP_K_RULE   = 3         # DKVMN must rank OUTSIDE the top 3 within session
BUCKETS      = [("within",0,20*60),("sameday",20*60,86400),("d1",86400,7*86400),
                ("d7",7*86400,28*86400),("d28",28*86400,np.inf)]

RES  = os.path.join(CACHE,"results")
RUNS = {"s1": os.path.join(RES,f"REV_{DATASET}"),
        "s2": os.path.join(RES,f"REV_{DATASET}_seed2"),
        "s3": os.path.join(RES,f"REV_{DATASET}_seed3")}
OUT  = os.path.join(RES,"SEED3_INVERSION"); os.makedirs(OUT,exist_ok=True)
pd.set_option("display.width",220); pd.set_option("display.max_columns",40)

ERRORS=[]
def flag(m): ERRORS.append(m); print("  !! "+m)

for k,v in RUNS.items(): print(f"{k}: {v}  {'(exists)' if os.path.isdir(v) else '(to be created)' if k=='s3' else '(MISSING)'}")
print("out:", OUT)
```

## 2. Pre-flight

```python
ok=True
for k in ("s1","s2"):
    if not os.path.exists(os.path.join(RUNS[k],"predictions_test.npz")):
        flag(f"{k} predictions missing at {RUNS[k]} -- the exploratory basis is incomplete"); ok=False
if RUN_SEED3 and not (os.path.isdir(RUNS["s3"]) and not FORCE_RERUN):
    if not os.path.exists(PIPELINE_NB): flag(f"PIPELINE_NB missing: {PIPELINE_NB}"); ok=False
    if not any(os.path.exists(os.path.join(d,"pcdt_datasets.py"))
               for d in [os.path.dirname(CACHE), CACHE, os.path.dirname(PIPELINE_NB)]):
        flag("pcdt_datasets.py not found near the cache or pipeline"); ok=False
print("PRE-FLIGHT:", "OK" if ok else "FAILED")
if not ok: raise SystemExit("fix the above first")
```

## 3. Run seed 20260204

Patches `seed_index = 2` (the third entry of `CFG.seeds`) and the dataset, nothing else. Output goes to
`REV_junyi_seed3`; seeds 1 and 2 are untouched.

The patch is asserted — a silent failure to apply it cannot occur.

```python
SEED3_RUNTIME=None
if RUN_SEED3:
    if os.path.isdir(RUNS["s3"]) and not FORCE_RERUN:
        print("REV_junyi_seed3 exists -- skipping training (set FORCE_RERUN=True to redo)")
    else:
        t0=time.time()
        nbj=json.load(open(PIPELINE_NB))
        src="\n".join("".join(c["source"]) for c in nbj["cells"] if c["cell_type"]=="code")
        before=src
        src=re.sub(r'dataset:\s*str\s*=\s*"[^"]+"', f'dataset: str = "{DATASET}"', src, count=1)
        src=re.sub(r'seed_index:\s*int\s*=\s*\d+', 'seed_index: int = 2', src, count=1)
        assert src!=before and 'seed_index: int = 2' in src, "seed/dataset patch failed"
        # the pipeline builds SEED_TAG from seed_index; verify it will write to _seed3
        assert 'SEED_TAG' in src, "pipeline lacks SEED_TAG -- output would overwrite a previous run"
        src="\n".join(l for l in src.split("\n") if not l.strip().startswith("!"))
        print("running v3 pipeline: junyi, seed_index=2 (seed 20260204) ...")
        ns={"__name__":"__main__"}
        exec(compile(src,"<v3-seed3>","exec"), ns)
        SEED3_RUNTIME=(time.time()-t0)/60
        print(f"\ncomplete in {SEED3_RUNTIME:.1f} min")
        got=ns.get("OUTPUT_DIR","")
        if not got.endswith("_seed3"): flag(f"pipeline wrote to {got}, expected a _seed3 folder")
else:
    print("RUN_SEED3=False -- skipping")
if not os.path.isdir(RUNS["s3"]): flag("REV_junyi_seed3 absent -- no confirmatory test possible")
```

## 4. Helpers

```python
from sklearn.metrics import roc_auc_score

def load(run_dir):
    p=os.path.join(run_dir,"predictions_test.npz")
    if not os.path.exists(p): return None
    z=np.load(p)
    return dict(z=z, models=[k for k in z.files if k not in ("targets","mask","bucket")],
                Y=z["targets"], M=z["mask"]>0, B=z["bucket"])

def within_ranking(pk):
    sel=pk["M"]; y=pk["Y"][sel]
    r=[dict(model=m, auc=float(roc_auc_score(y, pk["z"][m][sel]))) for m in pk["models"]]
    return pd.DataFrame(r).sort_values("auc",ascending=False).reset_index(drop=True)

def gap_ranking(pk):
    rows=[]
    for bid,(name,lo,hi) in enumerate(BUCKETS):
        if name not in REAL_GAPS: continue
        sel=(pk["B"]==bid)&pk["M"]
        if sel.sum()<MIN_POINTS: continue
        y=pk["Y"][sel]
        if len(np.unique(y))<2: continue
        for m in pk["models"]:
            rows.append(dict(horizon=name, model=m, n_points=int(sel.sum()),
                             auc=float(roc_auc_score(y, pk["z"][m][sel]))))
    return pd.DataFrame(rows)

def boot_delta(pk, a, b, sel, n=N_BOOT, seed=0):
    """Learner-clustered bootstrap of AUC(a) - AUC(b) on the masked selection."""
    rg=np.random.default_rng(seed)
    rows_idx=np.repeat(np.arange(sel.shape[0])[:,None], sel.shape[1], 1)[sel]
    ya=pk["Y"][sel]; pa=pk["z"][a][sel]; pb=pk["z"][b][sel]
    learners=np.unique(rows_idx); by={l:np.where(rows_idx==l)[0] for l in learners}
    d=[]
    for _ in range(n):
        s=rg.choice(learners,len(learners),replace=True)
        ii=np.concatenate([by[l] for l in s])
        if len(np.unique(ya[ii]))<2: continue
        d.append(roc_auc_score(ya[ii],pa[ii])-roc_auc_score(ya[ii],pb[ii]))
    d=np.asarray(d)
    return (float(roc_auc_score(ya,pa)-roc_auc_score(ya,pb)),
            float(np.percentile(d,2.5)), float(np.percentile(d,97.5)))

PK={k:load(v) for k,v in RUNS.items()}
for k,v in PK.items():
    print(f"{k}: {'loaded ' + str(len(v['models'])) + ' models' if v else 'MISSING'}")
```

## 5. PRIMARY test — rank-based, applied to seed 3

Seeds 1 and 2 are shown for context and are explicitly labelled exploratory. **Only the seed-3 row
constitutes confirmation.**

```python
prim=[]
for k in ("s1","s2","s3"):
    pk=PK.get(k)
    if not pk: continue
    w=within_ranking(pk); g=gap_ranking(pk)
    w.to_csv(f"{OUT}/within_ranking_{k}.csv",index=False)
    g.to_csv(f"{OUT}/gap_ranking_{k}.csv",index=False)
    dk_rank=int(w.index[w.model=="DKVMN"][0])+1
    tops={h: g[g.horizon==h].sort_values("auc",ascending=False).iloc[0]["model"] for h in g.horizon.unique()}
    leg1 = dk_rank > TOP_K_RULE
    leg2 = bool(len(tops)) and all(m=="DKVMN" for m in tops.values())
    prim.append(dict(seed=k, role="exploratory" if k in ("s1","s2") else "CONFIRMATORY",
                     dkvmn_within_rank=dk_rank, n_models=len(w),
                     top_within=w.iloc[0]["model"], top_within_auc=round(float(w.iloc[0]["auc"]),4),
                     dkvmn_within_auc=round(float(w[w.model=="DKVMN"].auc.iloc[0]),4),
                     gaps_tested=list(tops.keys()), gap_leaders=tops,
                     leg1_outside_top3=bool(leg1), leg2_first_all_gaps=bool(leg2),
                     inversion=bool(leg1 and leg2)))
    print(f"--- {k} ({'exploratory' if k in ('s1','s2') else 'CONFIRMATORY'}) ---")
    print(w.round(4).to_string(index=False))
    print(f"  DKVMN within-session rank: {dk_rank}/{len(w)}   gap leaders: {tops}")
    print(f"  leg1 outside top {TOP_K_RULE}: {leg1} | leg2 first at every gap: {leg2} -> inversion={leg1 and leg2}\n")

PRIM=pd.DataFrame(prim); PRIM.to_csv(f"{OUT}/primary_rule_by_seed.csv",index=False)
```

## 6. SECONDARY test — magnitude with bootstrap intervals

`AUC(PCDT) − AUC(DKVMN)` within session and at each real gap. The sign flip is the inversion expressed
as a magnitude. The within-session leg is also compared against the 0.010 meaningfulness floor, so the
manuscript reports it honestly rather than implying a large lead.

```python
sec=[]
for k in ("s1","s2","s3"):
    pk=PK.get(k)
    if not pk or "PCDT" not in pk["models"] or "DKVMN" not in pk["models"]: continue
    d,lo,hi = boot_delta(pk,"PCDT","DKVMN", pk["M"], seed=0)
    sec.append(dict(seed=k, scope="within-session", delta=d, ci_lo=lo, ci_hi=hi,
                    excludes_zero=bool(lo*hi>0), above_mmde=bool(abs(d)>=MMDE)))
    for bid,(name,l,h) in enumerate(BUCKETS):
        if name not in REAL_GAPS: continue
        sel=(pk["B"]==bid)&pk["M"]
        if sel.sum()<MIN_POINTS: continue
        d,lo,hi = boot_delta(pk,"PCDT","DKVMN", sel, seed=0)
        sec.append(dict(seed=k, scope=name, delta=d, ci_lo=lo, ci_hi=hi,
                        excludes_zero=bool(lo*hi>0), above_mmde=bool(abs(d)>=MMDE)))
SEC=pd.DataFrame(sec)
if len(SEC):
    SEC.to_csv(f"{OUT}/secondary_magnitude_by_seed.csv",index=False)
    print(SEC.round(4).to_string(index=False))

    def signflip(k):
        s=SEC[SEC.seed==k]
        if not len(s): return None
        w=s[s.scope=="within-session"]
        g=s[s.scope.isin(REAL_GAPS)]
        pos = bool(len(w) and (w.delta.iloc[0]>0) and w.excludes_zero.iloc[0])
        neg = bool(len(g) and ((g.delta<0)&g.excludes_zero).any())
        return dict(within_positive_sig=pos, any_gap_negative_sig=neg, sign_flip=bool(pos and neg))
    SIGN={k:signflip(k) for k in ("s1","s2","s3") if signflip(k)}
    print("\nsign-flip test:", json.dumps(SIGN, indent=2))
    json.dump(SIGN, open(f"{OUT}/secondary_signflip.json","w"), indent=2)
else:
    flag("no magnitude rows computed"); SIGN={}
```

## 7. Verdict

```python
s3 = PRIM[PRIM.seed=="s3"] if len(PRIM) else pd.DataFrame()
VERDICT=dict(
    generated=time.strftime("%Y-%m-%d %H:%M"), dataset=DATASET,
    seed3_runtime_min=SEED3_RUNTIME,
    preregistered_primary_rule=(f"DKVMN ranks outside the top {TOP_K_RULE} within session AND first at "
                                "every real gap horizon, on seed 20260204"),
    preregistered_secondary_rule="AUC(PCDT)-AUC(DKVMN) bootstrap CI positive within session and negative at >=1 real gap",
    confirmatory_seed="s3 (20260204)",
    exploratory_seeds=["s1 (20260202)","s2 (20260203)"],
    primary_by_seed=PRIM.to_dict("records") if len(PRIM) else [],
    secondary_signflip=SIGN if 'SIGN' in dir() else {},
    errors=ERRORS)

if not len(s3):
    VERDICT["confirmed"]=None
    print("VERDICT: NOT DETERMINED -- seed 3 has not been run")
else:
    conf=bool(s3.iloc[0]["inversion"])
    VERDICT["confirmed"]=conf
    print("="*76)
    print("RANKING INVERSION --", "CONFIRMED on seed 20260204" if conf else "NOT CONFIRMED on seed 20260204")
    print("="*76)
    r=s3.iloc[0]
    print(f"  DKVMN within-session rank : {r['dkvmn_within_rank']}/{r['n_models']}  "
          f"(rule: outside top {TOP_K_RULE})")
    print(f"  gap leaders               : {r['gap_leaders']}")
    print(f"  leg 1 / leg 2             : {r['leg1_outside_top3']} / {r['leg2_first_all_gaps']}")
    if conf:
        print("\n  -> Report as a CONFIRMATORY finding, pre-registered and tested out of sample.")
        print("     Required wording: the within-session lead of the PCDT family over DKVMN is BELOW")
        print("     the 0.010 meaningfulness floor, so state the inversion as a MODEL-SELECTION")
        print("     consequence, not as a claim that PCDT is meaningfully better within session.")
    else:
        print("\n  -> The inversion is NOT confirmed out of sample. Report it as an exploratory")
        print("     observation from seeds 1-2 with this failure disclosed. Do not lead the paper")
        print("     with it.")
json.dump(VERDICT, open(f"{OUT}/verdict_inversion_seed3.json","w"), indent=2, default=str)
print("\nwrote", f"{OUT}/verdict_inversion_seed3.json")
```

## 8. Summary

```python
print("="*76); print("SEED-3 INVERSION CONFIRMATION -- SUMMARY"); print("="*76)
if SEED3_RUNTIME: print(f"seed-3 runtime: {SEED3_RUNTIME:.1f} min")
if len(PRIM):
    print("\nprimary rule by seed (only s3 is confirmatory):")
    print(PRIM[["seed","role","dkvmn_within_rank","leg1_outside_top3","leg2_first_all_gaps","inversion"]]
          .to_string(index=False))
if 'SEC' in dir() and len(SEC):
    print("\nmagnitude, PCDT - DKVMN:")
    print(SEC[["seed","scope","delta","ci_lo","ci_hi","excludes_zero","above_mmde"]].round(4).to_string(index=False))
print("\nfiles:")
for f in sorted(os.listdir(OUT)): print("   ", os.path.join(OUT,f))
if ERRORS:
    print("\nERRORS:"); [print("   -",e) for e in ERRORS]
else:
    print("\nno errors -- every statement above is backed by a file listed here.")
print("\nSend the SEED3_INVERSION folder back.")
```
