# Experiment P2-02 — Predictive Validity of a Persistent Multidimensional Representation

**Paper 2:** *A Persistent Multidimensional State Representation for Cognitive Digital Twins*
**Research question:** RQ2.2
**Produces:** Manuscript **Table 2** — next-interaction predictive performance under standardised,
capacity-matched evaluation
**Depends on:** Experiment P2-01 (RQ2.1 construct-validity audit)

---

## Experiment Overview

> **RQ2.2 —** Does a multidimensional persistent state representation composed of the **empirically
> justified dimensions identified in RQ2.1** yield higher next-interaction predictive accuracy than
> competitively tuned knowledge-only baselines?

This experiment is designed so that a *negative* answer is publishable. The comparison is deliberately
hostile to the proposed model:

| Safeguard | Why it matters |
|---|---|
| **Capacity matching** | Every neural model is tuned to the same parameter budget, so any difference is attributable to *representation*, not model size. |
| **Standardised pipeline** | Identical preprocessing, sequence construction, splits, optimiser, schedule, and evaluation code for all models. |
| **Tuned baselines** | Baselines receive the same hyper-parameter search budget as the proposed model. Weak-baseline comparison is the single most common cause of rejection in this literature. |
| **Audit-constrained state** | The PCDT model may only use dimensions that survived the RQ2.1 audit — it cannot quietly reintroduce rejected dimensions to win. |
| **Pre-registered effect size** | A minimum practically-meaningful improvement is declared *before* results are seen. |

### Why these safeguards are non-negotiable

Standardised benchmarking of deep knowledge tracing has shown that reported performance differences
between architectures shrink substantially once preprocessing, splitting, and tuning are held constant.
An uncontrolled comparison would therefore produce a result that is *expected* to be unreproducible.
The expected honest outcome for this experiment is a **small** effect, and the notebook is built to detect
and report a small effect accurately rather than to manufacture a large one.


## 1. Objectives, Hypotheses, and Pre-Registered Decision Rule

### Primary hypothesis

$$H_1:\quad \mathrm{AUC}_{\text{PCDT}} \;>\; \max_{b \in \mathcal{B}} \mathrm{AUC}_{b}$$

where $\mathcal{B}$ is the set of tuned knowledge-only baselines.

### Null and the honest alternative

$$H_0:\quad \mathrm{AUC}_{\text{PCDT}} \;\le\; \max_{b \in \mathcal{B}} \mathrm{AUC}_{b}$$

If $H_0$ is not rejected, the finding is reported as such. Given the RQ2.1 audit retained only a narrow
dimension subset, **failure to reject $H_0$ is a plausible and scientifically informative outcome**: it
would indicate that construct-valid multidimensionality does not translate into next-step predictive gain,
which redirects the value of the representation toward persistence (RQ2.3), stability (RQ2.4), and
downstream decision quality rather than toward benchmark accuracy.

### Pre-registered minimum meaningful effect

$$\Delta_{\min} = 0.010 \text{ AUC}$$

Declared before any model is trained. Differences below this are reported as **statistically detectable but
practically negligible** even when $p < 0.05$ — with hundreds of thousands of test interactions, trivial
differences will reach significance, so significance alone is not evidence of a contribution.

### Secondary objectives

1. Report **calibration** (ECE), not only discrimination (AUC). A representation that yields better
   calibrated probabilities has value even at AUC parity — this is the strongest fallback claim available.
2. Quantify the **contribution of each retained dimension** by ablation.
3. Report **paired statistical tests with correction and effect sizes**, not raw metric differences.


## 2. Mathematical Formulation

### 2.1 Task definition

For learner $\ell$ let the interaction sequence be

$$\mathcal{X}_\ell \;=\; \bigl\langle (q_1, a_1, \mathbf{f}_1, t_1),\; \dots,\; (q_T, a_T, \mathbf{f}_T, t_T) \bigr\rangle$$

| Symbol | Meaning |
|---|---|
| $q_t \in \{1,\dots,K\}$ | knowledge component (skill) of item $t$ |
| $a_t \in \{0,1\}$ | first-attempt correctness |
| $\mathbf{f}_t \in \mathbb{R}^{d_f}$ | auxiliary observable features (response time, attempts, hint fraction) |
| $t_t$ | wall-clock timestamp |

The prediction task is next-interaction correctness:

$$\hat{a}_t \;=\; P\bigl(a_t = 1 \mid q_t,\; \mathcal{X}_{1:t-1}\bigr)$$

**Epistemic note.** $\hat{a}_t$ is a prediction about an *observable*, not an estimate of latent cognition.
A model achieving high AUC has demonstrated predictive validity of its representation — it has **not**
demonstrated that its internal state corresponds to $S_t$.

### 2.2 Baseline models

**Bayesian Knowledge Tracing (per-skill).** A two-state HMM per skill $k$ with parameters
$\theta_k = (P(L_0), P(T), P(G), P(S))$. With $L_t = P(\text{mastered at } t)$:

$$P(a_t{=}1) \;=\; L_t(1-P(S)) + (1-L_t)P(G)$$

Posterior update after observing $a_t$:

$$L_t^{+} =
\begin{cases}
\dfrac{L_t(1-P(S))}{L_t(1-P(S)) + (1-L_t)P(G)} & a_t = 1\\[10pt]
\dfrac{L_t P(S)}{L_t P(S) + (1-L_t)(1-P(G))} & a_t = 0
\end{cases}
\qquad
L_{t+1} = L_t^{+} + (1-L_t^{+})P(T)$$

Parameters are fitted per skill by Expectation–Maximisation on the training split only.

**Deep Knowledge Tracing (recurrent).** With $\mathbf{x}_t$ a learned embedding of the interaction
$(q_t, a_t)$:

$$\mathbf{h}_t = \mathrm{LSTM}(\mathbf{x}_t, \mathbf{h}_{t-1}), \qquad
\hat{a}_{t+1} = \sigma\bigl(\mathbf{w}_{q_{t+1}}^\top \mathbf{h}_t + b\bigr)$$

**Self-Attentive KT.** Query from the target exercise, keys/values from past interactions, with a strict
causal mask $M$:

$$\mathrm{Attn}(Q,K,V) = \mathrm{softmax}\!\left(\frac{QK^\top}{\sqrt{d}} + M\right)V,
\qquad M_{ij} = -\infty \ \text{ if } j \ge i$$

**Attentive KT.** Rasch-style item embeddings plus *distance-decayed monotonic* attention:

$$\mathbf{e}_{q} = \mathbf{c}_{q} + \mu_{q}\,\mathbf{d}_{c(q)},
\qquad
\alpha_{ij} \propto \exp\!\left(\frac{q_i k_j^\top}{\sqrt{d}} - \theta\, \delta(i,j)\right)$$

where $\mu_q$ is item difficulty, $\mathbf{d}$ a concept-variation vector, and $\delta(i,j)$ a
context-aware temporal distance. Decay $\theta \ge 0$ is learned, encoding that distant interactions
matter less.

**Memory-Augmented KT.** A static key memory $\mathbf{M}^k \in \mathbb{R}^{N \times d_k}$ and a dynamic
value memory $\mathbf{M}^v_t \in \mathbb{R}^{N \times d_v}$. Attention over concepts:

$$w_i = \mathrm{softmax}\bigl(\mathbf{k}_{q_t}^\top \mathbf{M}^k_i\bigr),
\qquad \mathbf{r}_t = \sum_i w_i \mathbf{M}^v_{t,i}$$

with erase–add write:

$$\mathbf{M}^v_{t+1,i} = \mathbf{M}^v_{t,i}\bigl(1 - w_i \mathbf{e}_t\bigr) + w_i \mathbf{a}_t$$

### 2.3 The PCDT model

The proposed model differs from the baselines in **two** specified ways, and in no other way — this is what
makes the comparison interpretable.

**(i) Structured multidimensional state.** The hidden state is partitioned by audited dimension:

$$\mathbf{h}_t = \bigl[\,\mathbf{h}^{C}_t \;\|\; \mathbf{h}^{B}_t\,\bigr]$$

with $\mathbf{h}^{C}$ updated from correctness evidence and $\mathbf{h}^{B}$ from behavioural observables
$\mathbf{f}_t$. Dimensions rejected by the RQ2.1 audit are **not** included.

**(ii) Explicit persistence operator $\Psi$.** Between interactions separated by elapsed time
$\Delta t$, the state is decayed rather than carried forward unchanged:

$$\Psi(\mathbf{h}, \Delta t) \;=\; \mathbf{h} \odot \exp\bigl(-\boldsymbol{\lambda} \odot \Delta \tilde{t}\bigr),
\qquad \Delta\tilde{t} = \log\!\left(1 + \frac{\Delta t}{\tau_0}\right)$$

where $\boldsymbol{\lambda} \ge 0$ is a **learned per-dimension** decay rate and $\tau_0$ a time-scale
constant. Learning $\boldsymbol{\lambda}$ per dimension operationalises the claim that different components
of learner state persist at different rates. Baselines are indexed by interaction only and have no
$\Delta t$ pathway — this is precisely the representational difference RQ2.2 tests.

Recurrent update, then prediction:

$$\mathbf{h}_t = \mathrm{GRU}\bigl([\mathbf{x}_t \| \mathbf{f}_t],\; \Psi(\mathbf{h}_{t-1}, \Delta t_t)\bigr),
\qquad \hat{a}_{t+1} = \sigma\bigl(\mathrm{MLP}([\mathbf{h}_t \| \mathbf{e}_{q_{t+1}}])\bigr)$$

### 2.4 Loss

Masked binary cross-entropy over valid positions $\mathcal{V}$:

$$\mathcal{L} = -\frac{1}{|\mathcal{V}|}\sum_{(\ell,t)\in\mathcal{V}}
\Bigl[a_{t}\log\hat{a}_{t} + (1-a_{t})\log(1-\hat{a}_{t})\Bigr]$$

### 2.5 Evaluation metrics

**AUC-ROC** — threshold-free discrimination; the field-standard primary metric.

**Accuracy** — $\frac{1}{|\mathcal{V}|}\sum \mathbb{1}[\hat{a} \ge 0.5] = a$. Reported for comparability but
sensitive to class imbalance.

**RMSE** — $\sqrt{\frac{1}{|\mathcal{V}|}\sum (\hat{a}-a)^2}$; penalises confident errors.

**NLL** — $-\frac{1}{|\mathcal{V}|}\sum [a\log\hat{a} + (1-a)\log(1-\hat{a})]$; a strictly proper scoring
rule, sensitive to calibration in a way AUC is not.

**Expected Calibration Error** — with $M$ equal-width confidence bins $B_m$:

$$\mathrm{ECE} = \sum_{m=1}^{M}\frac{|B_m|}{n}\Bigl|\mathrm{acc}(B_m) - \mathrm{conf}(B_m)\Bigr|$$

ECE is reported for **every** model. *(Deviation from the original Table 2 template, which marked ECE as
unavailable for baselines: every model here emits a probability, so every model has a measurable
calibration error. Omitting it for baselines would have made the proposed model's calibration claim
unfalsifiable.)*

### 2.6 Statistical comparison

Paired bootstrap over **test learners** (the independent unit — interactions within a learner are not
independent):

$$\Delta^{(b)} = m\bigl(\mathcal{T}^{(b)}_{\text{PCDT}}\bigr) - m\bigl(\mathcal{T}^{(b)}_{\text{baseline}}\bigr)$$

resampling learners with replacement for $b = 1..B$. Reported: the $95\%$ percentile interval of $\Delta$,
and a Holm–Bonferroni corrected $p$-value across the family of baseline comparisons.

## 4. Configuration and Pre-Registered Decisions

Everything that constitutes an analytic choice is fixed here, **before** any model is trained.

**Capacity matching.** `param_budget` is the target parameter count for every neural model. Each model's
hidden width is searched to land within `param_tolerance` of it. Bayesian KT is excluded from matching —
it has $4K$ parameters by construction and cannot be scaled; this is stated rather than hidden.

**The audited subset.** `pcdt_dimensions` is constrained by the RQ2.1 result. On the ASSISTments
2020–2021 audit, $C_t$ was retained (reliability $0.733$, convergent $r = 0.569$) and $B_t$ carried real
but negatively-signed convergent validity ($r = -0.233$, interpretable as a *struggle* signal) while
failing the reliability floor. $M_t$, $G_t$, and $I_t$ produced no real-data validity evidence.

The default configuration therefore uses $\{C_t, B_t\}$ plus the architectural persistence operator $\Psi$.
$\Psi$ is included as a **mechanism**, not as a claim about the $H_t$ *proxy*, which failed its audit — the
distinction matters and is documented in §12.


```python
@dataclass(frozen=True)
class Config:
    # ---------- data ----------
    max_learners: int = 20_000        # subsample for tractability; raise for the final run
    min_seq_len: int = 10             # learners with fewer interactions are excluded
    max_seq_len: int = 200            # sequence window (KT-literature standard)
    train_frac: float = 0.70
    val_frac: float = 0.10            # test = remainder

    # ---------- capacity matching ----------
    param_budget: int = 350_000
    param_tolerance: float = 0.05     # +/- 5 percent

    # ---------- optimisation (identical for every neural model) ----------
    batch_size: int = 64
    lr: float = 1e-3
    weight_decay: float = 1e-5
    max_epochs: int = 30
    patience: int = 5                 # early stopping on validation AUC
    grad_clip: float = 5.0

    # ---------- hyper-parameter search (identical budget for every model) ----------
    search_lrs: Sequence[float] = (5e-4, 1e-3, 2e-3)

    # ---------- PCDT ----------
    pcdt_dimensions: Sequence[str] = ("C_t", "B_t")   # audited subset from RQ2.1
    use_persistence: bool = True                       # the Psi operator
    tau0_seconds: float = 3600.0                       # persistence time-scale

    # ---------- evaluation ----------
    ece_bins: int = 15
    n_bootstrap: int = 2000
    min_meaningful_auc_delta: float = 0.010            # pre-registered

CFG = Config()

DATA_DIR   = "dataset/feb_to_apr"
PLOGS_PATH = f"{DATA_DIR}/plogs.csv"
PDETS_PATH = f"{DATA_DIR}/pdets.csv"

MODELS = ["BKT", "DKT", "SAKT", "AKT", "DKVMN", "PCDT"]

print("Pre-registered before any training:")
print(f"  parameter budget      : {CFG.param_budget:,} (+/- {CFG.param_tolerance:.0%})")
print(f"  min meaningful dAUC   : {CFG.min_meaningful_auc_delta}")
print(f"  PCDT audited subset   : {list(CFG.pcdt_dimensions)} | persistence={CFG.use_persistence}")
print(f"  identical search grid : lr in {list(CFG.search_lrs)} for every neural model")
```

## 5. Data Loading and Preprocessing

The cleaning pipeline is **identical to Experiment P2-01** so that RQ2.1 and RQ2.2 operate on the same
population. Steps: numeric coercion → drop unusable correctness → single ISO8601 timestamp parse →
skill explode-join.

**One deliberate difference.** P2-01's explode-join duplicates a row per (problem, skill) pair, which is
correct for skill-indexed *aggregates* but wrong for *sequence modelling* — it would insert duplicate
interactions into the temporal stream and inflate apparent history. Here each interaction is kept **once**,
using its primary (first-listed) skill tag. This is documented as a divergence rather than applied silently.


```python
def _parse_skills(v):
    if isinstance(v, list): return v
    if isinstance(v, str) and v.strip().startswith("["):
        try: return ast.literal_eval(v)
        except (ValueError, SyntaxError): return []
    return [] if pd.isna(v) else [v]


def make_fixture(n_learners=1200, rng=None):
    """Synthetic data in the real plogs/pdets schema — pipeline verification only."""
    rng = rng or np.random.default_rng(0)
    n_problems, skills = 400, [f"CCSS.{i}" for i in range(30)]
    pdets = pd.DataFrame({"problem_id": range(n_problems),
                          "skills": [list(rng.choice(skills, rng.integers(1, 3), replace=False))
                                     for _ in range(n_problems)]})
    prob_skill = {p: s[0] for p, s in zip(pdets.problem_id, pdets.skills)}
    skill_diff = {s: rng.normal(0, .8) for s in skills}
    rows, t0 = [], pd.Timestamp("2021-02-01", tz="UTC")
    for lid in range(n_learners):
        ability, metacog = rng.normal(0, 1), rng.beta(2, 2)
        mastery = {s: rng.normal(0, .3) for s in skills}
        t = t0 + pd.Timedelta(days=int(rng.integers(0, 20)))
        for _ in range(int(rng.integers(2, 10))):
            t += pd.Timedelta(days=int(rng.integers(1, 14)), hours=int(rng.integers(0, 20)))
            for _ in range(int(rng.integers(5, 30))):
                pid = int(rng.integers(0, n_problems)); sk = prob_skill[pid]
                logit = ability + mastery[sk] - skill_diff[sk]
                p = 1 / (1 + np.exp(-logit))
                corr = rng.binomial(1, np.clip(p, .02, .98))
                mastery[sk] += 0.15 if corr else 0.05           # learning
                rows.append(dict(student_id=lid, problem_id=pid, start_time=t,
                                 time_on_task=float(np.clip(rng.lognormal(3 - metacog*.3, .4), 2, 900)),
                                 attempt_count=int(1 + rng.poisson(max(0, (1-metacog)*1.5))),
                                 fraction_of_hints_used=(np.nan if rng.random() < .12 else
                                                         float(np.clip(rng.beta(2,5)*(1.3-metacog), 0, 1.3))),
                                 correct=(np.nan if rng.random() < .15 else bool(corr))))
                t += pd.Timedelta(seconds=int(rng.integers(20, 240)))
    return pd.DataFrame(rows), pdets


def load_and_clean():
    using_fixture = not os.path.exists(PLOGS_PATH)
    if using_fixture:
        print("!"*76); print("! Drive path not found -> SYNTHETIC FIXTURE (pipeline verification only).")
        print("! Numbers produced are NOT results for the manuscript."); print("!"*76)
        plogs, pdets = make_fixture(rng=RNG)
    else:
        plogs = pd.read_csv(PLOGS_PATH, usecols=["student_id","problem_id","start_time","time_on_task",
                                                 "attempt_count","fraction_of_hints_used","correct"])
        pdets = pd.read_csv(PDETS_PATH, usecols=["problem_id","skills"])
        pdets["skills"] = pdets["skills"].map(_parse_skills)

    df = plogs.copy()
    df["correct"] = pd.to_numeric(
        df["correct"].map({True:1.0, False:0.0, "True":1.0, "False":0.0}).fillna(df["correct"]),
        errors="coerce")
    for c in ("time_on_task","attempt_count","fraction_of_hints_used"):
        df[c] = pd.to_numeric(df[c], errors="coerce")
    n0 = len(df); df = df.dropna(subset=["correct"])
    print(f"dropped {n0-len(df):,} rows ({(n0-len(df))/n0:.1%}) with unusable `correct`")

    df["start_time"] = pd.to_datetime(df["start_time"], format="ISO8601", utc=True)

    # PRIMARY skill only -- one row per interaction (see markdown above)
    prim = pdets.copy()
    prim["skill_id"] = prim["skills"].map(lambda s: s[0] if isinstance(s, list) and len(s) else np.nan)
    prim = prim.dropna(subset=["skill_id"])[["problem_id","skill_id"]]
    nb_before = df["problem_id"].nunique()
    df = df.merge(prim, on="problem_id", how="inner")
    print(f"skill join: {nb_before:,} -> {df['problem_id'].nunique():,} problems with a primary skill")

    df = df.rename(columns={"student_id":"learner_id","problem_id":"item_id",
                            "time_on_task":"response_time","fraction_of_hints_used":"hint_frac"})
    return df[["learner_id","item_id","skill_id","start_time","correct",
               "response_time","hint_frac","attempt_count"]], using_fixture


interactions, USING_FIXTURE = load_and_clean()
print(f"\ninteractions: {len(interactions):,} | learners: {interactions.learner_id.nunique():,} "
      f"| skills: {interactions.skill_id.nunique():,}")
print(f"overall correctness rate: {interactions.correct.mean():.3f}")
```

## 6. Leakage-Safe Splits and Sequence Construction

### Splitting unit

Splits are **by learner**, never by interaction. Splitting interactions randomly would place a learner's
later interactions in train and earlier ones in test, leaking that learner's trajectory across the
boundary and inflating every model's score. This is the most common silent error in knowledge-tracing
evaluation.

### Feature scaling

Behavioural features are standardised using statistics computed on the **training split only**, then
applied unchanged to validation and test. Computing them over the full dataset would leak test
distribution information into training.

### Sequence encoding

Each learner's history is ordered by time and windowed to `max_seq_len`. At position $t$ a model sees
interactions $1..t-1$ and the identity of item $t$, and predicts $a_t$. Positions beyond a learner's
history are padded and **masked out of both the loss and every metric**.

$\Delta t$ (elapsed seconds since the previous interaction) is retained as a per-position feature. Only the
PCDT model consumes it; baselines are indexed by interaction position alone, which is exactly the
representational difference under test.


```python
BEHAV_COLS = ["response_time", "hint_frac", "attempt_count"]

def build_sequences(df: pd.DataFrame, cfg: Config, rng):
    df = df.sort_values(["learner_id", "start_time"]).copy()

    # --- keep learners with enough history, then subsample for tractability ---
    counts = df.groupby("learner_id").size()
    keep = counts[counts >= cfg.min_seq_len].index
    if len(keep) > cfg.max_learners:
        keep = rng.choice(np.asarray(keep), cfg.max_learners, replace=False)
    df = df[df.learner_id.isin(set(keep))].copy()

    # --- learner-level split (NO interaction-level splitting) ---
    learners = np.sort(df.learner_id.unique()); rng.shuffle(learners)
    n = len(learners)
    n_tr, n_va = int(cfg.train_frac*n), int(cfg.val_frac*n)
    split_of = {}
    for lid in learners[:n_tr]:            split_of[lid] = "train"
    for lid in learners[n_tr:n_tr+n_va]:   split_of[lid] = "val"
    for lid in learners[n_tr+n_va:]:       split_of[lid] = "test"
    df["split"] = df.learner_id.map(split_of)

    # --- skill vocabulary from TRAIN only; unseen skills -> index 0 (UNK) ---
    train_skills = df.loc[df.split=="train", "skill_id"].unique()
    skill2idx = {s: i+1 for i, s in enumerate(sorted(train_skills))}
    df["skill_idx"] = df.skill_id.map(skill2idx).fillna(0).astype(int)
    n_skills = len(skill2idx) + 1

    # --- behavioural features: impute + scale using TRAIN statistics only ---
    tr = df[df.split=="train"]
    med = tr[BEHAV_COLS].median()
    df[BEHAV_COLS] = df[BEHAV_COLS].fillna(med)
    df["response_time"] = np.log1p(df["response_time"].clip(lower=0))   # heavy right skew
    tr = df[df.split=="train"]
    mu, sd = tr[BEHAV_COLS].mean(), tr[BEHAV_COLS].std().replace(0, 1.0)
    df[BEHAV_COLS] = (df[BEHAV_COLS] - mu) / sd

    # --- elapsed time since previous interaction (PCDT only) ---
    df["dt_sec"] = df.groupby("learner_id")["start_time"].diff().dt.total_seconds().fillna(0).clip(lower=0)

    # --- pack into fixed-width padded arrays ---
    packs = {}
    for sp in ("train","val","test"):
        sub = df[df.split==sp]
        gids = sub.learner_id.unique()
        L = cfg.max_seq_len
        S  = np.zeros((len(gids), L), dtype=np.int64)   # skill index
        A  = np.zeros((len(gids), L), dtype=np.float32) # correctness
        Fb = np.zeros((len(gids), L, len(BEHAV_COLS)), dtype=np.float32)
        DT = np.zeros((len(gids), L), dtype=np.float32)
        M  = np.zeros((len(gids), L), dtype=np.float32) # validity mask
        for i, lid in enumerate(gids):
            g = sub[sub.learner_id==lid].tail(L)        # most recent window
            k = len(g)
            S[i,:k]  = g.skill_idx.values
            A[i,:k]  = g.correct.values
            Fb[i,:k] = g[BEHAV_COLS].values
            DT[i,:k] = g.dt_sec.values
            M[i,:k]  = 1.0
        packs[sp] = dict(skill=S, correct=A, behav=Fb, dt=DT, mask=M, learner_id=gids)
        print(f"  {sp:5s}: {len(gids):,} learners | {int(M.sum()):,} interactions")
    return packs, n_skills, skill2idx


print("Building sequences (learner-level split):")
PACKS, N_SKILLS, SKILL2IDX = build_sequences(interactions, CFG, RNG)
print(f"\nskill vocabulary (train-derived): {N_SKILLS:,} (index 0 = unseen)")
```

```python
class KTDataset(Dataset):
    """Positions 0..L-2 are inputs; targets are positions 1..L-1."""
    def __init__(self, pack):
        self.s  = torch.from_numpy(pack["skill"])
        self.a  = torch.from_numpy(pack["correct"])
        self.f  = torch.from_numpy(pack["behav"])
        self.dt = torch.from_numpy(pack["dt"])
        self.m  = torch.from_numpy(pack["mask"])
        self.lid = pack["learner_id"]

    def __len__(self):  return self.s.shape[0]

    def __getitem__(self, i):
        return dict(skill=self.s[i], correct=self.a[i], behav=self.f[i],
                    dt=self.dt[i], mask=self.m[i], idx=i)

def make_loader(pack, cfg, shuffle):
    return DataLoader(KTDataset(pack), batch_size=cfg.batch_size, shuffle=shuffle,
                      num_workers=0, drop_last=False)

LOADERS = {sp: make_loader(PACKS[sp], CFG, shuffle=(sp=="train")) for sp in ("train","val","test")}
print("dataloaders ready:", {k: len(v.dataset) for k, v in LOADERS.items()})
```

## 7. Model Definitions

All neural models share an identical interface — `forward(batch) -> logits[B, L-1]` predicting positions
$1..L-1$ — so the training and evaluation harness is literally the same code for every model. Only the
representation differs.

**Reimplementation disclosure.** These are standardised reimplementations under a common pipeline and a
matched parameter budget, not verbatim reproductions of the original papers. This is the methodologically
correct choice for a controlled comparison — original reported numbers are not comparable across papers
because preprocessing and splitting differ — but it must be stated explicitly in the manuscript, and
absolute values should not be compared against published figures.


```python
class DKT(nn.Module):
    """Recurrent knowledge tracing: LSTM over (skill, correctness) interaction embeddings."""
    def __init__(self, n_skills, hidden, emb=None, dropout=0.2):
        super().__init__()
        emb = emb or hidden
        self.inter_emb = nn.Embedding(2*n_skills + 1, emb, padding_idx=0)
        self.lstm = nn.LSTM(emb, hidden, batch_first=True)
        self.drop = nn.Dropout(dropout)
        self.out  = nn.Linear(hidden, n_skills)

    def forward(self, b):
        s, a = b["skill"], b["correct"]
        inter = (s + a.long()*self.out.out_features).clamp(max=self.inter_emb.num_embeddings-1)
        h, _ = self.lstm(self.inter_emb(inter[:, :-1]))
        logits_all = self.out(self.drop(h))                       # [B, L-1, K]
        nxt = s[:, 1:].unsqueeze(-1)                              # query = next skill
        return logits_all.gather(-1, nxt).squeeze(-1)


class SAKT(nn.Module):
    """Self-attentive KT: query = target exercise, key/value = past interactions, causal mask."""
    def __init__(self, n_skills, hidden, n_heads=4, dropout=0.2, max_len=200):
        super().__init__()
        # nn.MultiheadAttention requires embed_dim % num_heads == 0. Without this rounding the
        # capacity search hits invalid widths, the constructor raises, and the search collapses
        # downward -- silently leaving SAKT far below the parameter budget.
        hidden = max(n_heads, (hidden // n_heads) * n_heads)
        self.hidden = hidden
        self.inter_emb = nn.Embedding(2*n_skills + 1, hidden, padding_idx=0)
        self.ex_emb    = nn.Embedding(n_skills, hidden, padding_idx=0)
        self.pos_emb   = nn.Embedding(max_len, hidden)
        self.attn = nn.MultiheadAttention(hidden, n_heads, dropout=dropout, batch_first=True)
        self.ln1, self.ln2 = nn.LayerNorm(hidden), nn.LayerNorm(hidden)
        self.ffn  = nn.Sequential(nn.Linear(hidden, hidden*2), nn.ReLU(),
                                  nn.Dropout(dropout), nn.Linear(hidden*2, hidden))
        self.drop = nn.Dropout(dropout)
        self.out  = nn.Linear(hidden, 1)
        self.n_skills = n_skills

    def forward(self, b):
        s, a = b["skill"], b["correct"]
        inter = (s + a.long()*self.n_skills).clamp(max=self.inter_emb.num_embeddings-1)
        L = s.shape[1] - 1
        pos = torch.arange(L, device=s.device).unsqueeze(0)
        k = self.inter_emb(inter[:, :-1]) + self.pos_emb(pos)      # past interactions
        q = self.ex_emb(s[:, 1:])                                  # target exercises
        causal = torch.triu(torch.ones(L, L, device=s.device, dtype=torch.bool), diagonal=1)
        att, _ = self.attn(q, k, k, attn_mask=causal)
        h = self.ln1(q + self.drop(att))
        h = self.ln2(h + self.drop(self.ffn(h)))
        return self.out(h).squeeze(-1)


class AKT(nn.Module):
    """Attentive KT: Rasch-style embeddings + monotonic distance-decayed attention."""
    def __init__(self, n_skills, hidden, n_heads=4, dropout=0.2, max_len=200):
        super().__init__()
        self.c_emb    = nn.Embedding(n_skills, hidden, padding_idx=0)   # concept
        self.d_emb    = nn.Embedding(n_skills, hidden, padding_idx=0)   # concept variation
        self.diff     = nn.Embedding(n_skills, 1, padding_idx=0)        # scalar item difficulty
        self.resp_emb = nn.Embedding(3, hidden, padding_idx=2)
        self.q_proj, self.k_proj, self.v_proj = (nn.Linear(hidden, hidden) for _ in range(3))
        self.theta = nn.Parameter(torch.tensor(0.5))                    # learned decay
        self.ln    = nn.LayerNorm(hidden)
        self.ffn   = nn.Sequential(nn.Linear(2*hidden, hidden), nn.ReLU(),
                                   nn.Dropout(dropout), nn.Linear(hidden, 1))
        self.drop  = nn.Dropout(dropout)
        self.scale = hidden ** 0.5

    def _rasch(self, s):
        return self.c_emb(s) + self.diff(s) * self.d_emb(s)

    def forward(self, b):
        s, a = b["skill"], b["correct"]
        L = s.shape[1] - 1
        x_past = self._rasch(s[:, :-1]) + self.resp_emb(a[:, :-1].long())
        q = self.q_proj(self._rasch(s[:, 1:]))
        k, v = self.k_proj(x_past), self.v_proj(x_past)

        scores = (q @ k.transpose(-2, -1)) / self.scale
        pos = torch.arange(L, device=s.device)
        dist = (pos.view(-1,1) - pos.view(1,-1)).float().clamp(min=0)   # temporal distance
        scores = scores - F.softplus(self.theta) * dist                 # monotonic decay
        causal = torch.triu(torch.ones(L, L, device=s.device, dtype=torch.bool), diagonal=1)
        scores = scores.masked_fill(causal, float("-inf"))
        ctx = self.drop(torch.softmax(scores, dim=-1)) @ v
        return self.ffn(torch.cat([self.ln(ctx), q], dim=-1)).squeeze(-1)


class DKVMN(nn.Module):
    """Memory-augmented KT: static key memory + dynamic value memory with erase-add writes."""
    def __init__(self, n_skills, hidden, n_concepts=32, dropout=0.2):
        super().__init__()
        self.N, self.dk, self.dv = n_concepts, hidden, hidden
        self.k_emb = nn.Embedding(n_skills, self.dk, padding_idx=0)
        self.v_emb = nn.Embedding(2*n_skills + 1, self.dv, padding_idx=0)
        self.Mk    = nn.Parameter(torch.randn(self.N, self.dk) * 0.1)
        self.Mv0   = nn.Parameter(torch.randn(self.N, self.dv) * 0.1)
        self.erase = nn.Linear(self.dv, self.dv)
        self.add   = nn.Linear(self.dv, self.dv)
        self.ffn   = nn.Sequential(nn.Linear(self.dv + self.dk, hidden), nn.Tanh(),
                                   nn.Dropout(dropout), nn.Linear(hidden, 1))
        self.n_skills = n_skills

    def forward(self, b):
        s, a = b["skill"], b["correct"]
        B, L = s.shape
        inter = (s + a.long()*self.n_skills).clamp(max=self.v_emb.num_embeddings-1)
        Mv = self.Mv0.unsqueeze(0).expand(B, -1, -1).contiguous()
        outs = []
        for t in range(L - 1):
            kq = self.k_emb(s[:, t+1])                              # query = NEXT item
            w  = torch.softmax(kq @ self.Mk.t(), dim=-1)            # [B, N]
            r  = torch.einsum("bn,bnd->bd", w, Mv)                  # read
            outs.append(self.ffn(torch.cat([r, kq], dim=-1)))
            vt = self.v_emb(inter[:, t])                            # write PAST interaction
            e, ad = torch.sigmoid(self.erase(vt)), torch.tanh(self.add(vt))
            Mv = Mv * (1 - w.unsqueeze(-1)*e.unsqueeze(1)) + w.unsqueeze(-1)*ad.unsqueeze(1)
        return torch.cat(outs, dim=1)


class PCDT(nn.Module):
    """Persistent Cognitive Digital Twin: structured state over the RQ2.1-audited dimensions,
    with an explicit time-gap persistence operator Psi."""
    def __init__(self, n_skills, hidden, dims=("C_t","B_t"), use_persistence=True,
                 n_behav=3, tau0=3600.0, dropout=0.2):
        super().__init__()
        self.dims, self.use_persistence, self.tau0 = list(dims), use_persistence, tau0
        self.n_dims = len(self.dims)
        self.dim_size = max(8, hidden // self.n_dims)
        self.state_size = self.dim_size * self.n_dims

        self.inter_emb = nn.Embedding(2*n_skills + 1, hidden, padding_idx=0)
        self.ex_emb    = nn.Embedding(n_skills, hidden, padding_idx=0)
        self.behav_proj = nn.Linear(n_behav, hidden) if "B_t" in self.dims else None

        in_dim = hidden + (hidden if self.behav_proj is not None else 0)
        self.cell = nn.GRUCell(in_dim, self.state_size)
        # per-dimension learned decay rate  lambda >= 0  (softplus-parameterised)
        self.log_lambda = nn.Parameter(torch.full((self.n_dims,), -1.0))
        self.drop = nn.Dropout(dropout)
        self.head = nn.Sequential(nn.Linear(self.state_size + hidden, hidden), nn.ReLU(),
                                  nn.Dropout(dropout), nn.Linear(hidden, 1))
        self.n_skills = n_skills

    def psi(self, h, dt):
        """Persistence operator: per-dimension exponential decay in log-elapsed time."""
        if not self.use_persistence:
            return h
        lam = F.softplus(self.log_lambda)                              # [n_dims]
        dt_scaled = torch.log1p(dt.clamp(min=0) / self.tau0).unsqueeze(-1)   # [B,1]
        decay = torch.exp(-lam.unsqueeze(0) * dt_scaled)               # [B, n_dims]
        decay = decay.repeat_interleave(self.dim_size, dim=-1)         # [B, state_size]
        return h * decay

    def forward(self, b):
        s, a, f, dt = b["skill"], b["correct"], b["behav"], b["dt"]
        B, L = s.shape
        inter = (s + a.long()*self.n_skills).clamp(max=self.inter_emb.num_embeddings-1)
        h = torch.zeros(B, self.state_size, device=s.device)
        outs = []
        for t in range(L - 1):
            x = self.inter_emb(inter[:, t])
            if self.behav_proj is not None:
                x = torch.cat([x, self.behav_proj(f[:, t])], dim=-1)
            h = self.psi(h, dt[:, t])                                  # decay across the gap
            h = self.cell(x, h)                                        # then absorb evidence
            outs.append(self.head(torch.cat([self.drop(h), self.ex_emb(s[:, t+1])], dim=-1)))
        return torch.cat(outs, dim=1)

print("Neural model classes defined:", ["DKT","SAKT","AKT","DKVMN","PCDT"])
```

### 7.1 Bayesian Knowledge Tracing (per-skill, EM-fitted)

BKT is fitted per skill by Expectation–Maximisation on the training split only, then applied to test
sequences by forward filtering. It is the classical baseline the manuscript requires, and "tuned" here
means genuine per-skill parameter estimation rather than fixed default values — a properly-fitted BKT is a
substantially stronger baseline than the version deep models are usually compared against.

Its parameter count is $4K$ by construction and cannot be scaled to the neural budget, so it is **excluded
from capacity matching**. Reported for reference, not as a capacity-matched competitor.


```python
class BKTPerSkill:
    """Per-skill BKT fitted by EM on the training split."""
    def __init__(self, n_skills, n_iter=25, seed=SEED):
        self.n_skills, self.n_iter = n_skills, n_iter
        self.rng = np.random.default_rng(seed)
        self.params = {}                       # skill -> (L0, T, G, S)
        self.default = (0.30, 0.10, 0.20, 0.10)

    @staticmethod
    def _seq_ll_and_post(seq, L0, T, G, S):
        """Forward filter; returns log-lik and per-step P(mastered before response)."""
        L, ll, posts = L0, 0.0, []
        for a in seq:
            p_corr = L*(1-S) + (1-L)*G
            p_corr = min(max(p_corr, 1e-6), 1-1e-6)
            ll += np.log(p_corr if a == 1 else 1-p_corr)
            posts.append(L)
            L_post = (L*(1-S))/p_corr if a == 1 else (L*S)/(1-p_corr)
            L = L_post + (1-L_post)*T
        return ll, posts

    def fit(self, seqs_by_skill):
        for k, seqs in seqs_by_skill.items():
            if not seqs:
                self.params[k] = self.default; continue
            L0, T, G, S = self.default
            for _ in range(self.n_iter):
                num_L0 = den_L0 = 0.0
                cg = cg_n = cs = cs_n = 0.0
                trans_num = trans_den = 0.0
                for seq in seqs:
                    _, posts = self._seq_ll_and_post(seq, L0, T, G, S)
                    num_L0 += posts[0]; den_L0 += 1
                    for a, pm in zip(seq, posts):
                        cg   += (1-pm)*a;     cg_n += (1-pm)     # guess: unmastered but correct
                        cs   += pm*(1-a);     cs_n += pm         # slip: mastered but incorrect
                    for i in range(len(seq)-1):
                        trans_num += (1-posts[i])*posts[i+1]
                        trans_den += (1-posts[i])
                L0 = float(np.clip(num_L0/max(den_L0,1e-9), .01, .99))
                G  = float(np.clip(cg/max(cg_n,1e-9),      .01, .40))   # standard identifiability bounds
                S  = float(np.clip(cs/max(cs_n,1e-9),      .01, .40))
                T  = float(np.clip(trans_num/max(trans_den,1e-9), .001, .60))
            self.params[k] = (L0, T, G, S)
        return self

    def predict_pack(self, pack):
        """Forward-filter each learner sequence; returns preds/targets aligned to positions 1..L-1."""
        S_, A_, M_ = pack["skill"], pack["correct"], pack["mask"]
        n, L = S_.shape
        preds = np.zeros((n, L-1), np.float32); targs = np.zeros((n, L-1), np.float32)
        msk   = np.zeros((n, L-1), np.float32)
        for i in range(n):
            state = {}
            for t in range(L-1):
                if M_[i, t] == 0: break
                k_now = int(S_[i, t]); a_now = int(A_[i, t])
                L0, T, G, Sl = self.params.get(k_now, self.default)
                Lm = state.get(k_now, L0)
                p = Lm*(1-Sl) + (1-Lm)*G
                p = min(max(p, 1e-6), 1-1e-6)
                Lp = (Lm*(1-Sl))/p if a_now == 1 else (Lm*Sl)/(1-p)
                state[k_now] = Lp + (1-Lp)*T                          # update with observed response
                if M_[i, t+1] == 0: break
                k_nx = int(S_[i, t+1])
                L0n, Tn, Gn, Sn = self.params.get(k_nx, self.default)
                Ln = state.get(k_nx, L0n)
                preds[i, t] = Ln*(1-Sn) + (1-Ln)*Gn                   # predict NEXT
                targs[i, t] = A_[i, t+1]; msk[i, t] = 1.0
        return preds, targs, msk

    @property
    def n_params(self): return 4 * self.n_skills


def bkt_training_sequences(pack):
    """Per-skill list of response sequences from the training pack."""
    by_skill = {}
    S_, A_, M_ = pack["skill"], pack["correct"], pack["mask"]
    for i in range(S_.shape[0]):
        valid = M_[i] > 0
        per = {}
        for k, a in zip(S_[i][valid], A_[i][valid]):
            per.setdefault(int(k), []).append(int(a))
        for k, seq in per.items():
            by_skill.setdefault(k, []).append(seq)
    return by_skill

print("BKT implementation ready.")
```

## 8. Capacity Matching

Every neural model's hidden width is searched so that its trainable-parameter count lands within
`param_tolerance` of `param_budget`. Without this, any observed difference could simply reflect that one
model is larger — the single most common confound in architecture comparisons, and one that standardised
benchmarking has shown accounts for much of the apparent progress in this literature.

$$\text{select } d^\star = \arg\min_{d} \bigl| \, |\Theta_{\text{model}}(d)| - \Theta_{\text{budget}} \bigr|$$

Embedding tables dominate parameter counts and scale with the skill vocabulary, so the attainable width
differs across architectures — that is expected, and matching is on *total parameters*, which is the
quantity that governs capacity.


```python
def count_params(m): return sum(p.numel() for p in m.parameters() if p.requires_grad)

def match_capacity(builder, budget, tol, lo=8, hi=512):
    """Binary-search hidden width so total parameters land closest to the budget."""
    best, best_gap = None, float("inf")
    while lo <= hi:
        mid = (lo + hi) // 2
        try:
            n = count_params(builder(mid))
        except Exception:
            hi = mid - 1; continue
        gap = abs(n - budget)
        if gap < best_gap: best, best_gap = (mid, n), gap
        if n < budget: lo = mid + 1
        else:          hi = mid - 1
    hidden, n = best
    ok = abs(n - budget) / budget <= tol
    return hidden, n, ok

BUILDERS = {
    "DKT":   lambda h: DKT(N_SKILLS, h),
    "SAKT":  lambda h: SAKT(N_SKILLS, h, max_len=CFG.max_seq_len),
    "AKT":   lambda h: AKT(N_SKILLS, h, max_len=CFG.max_seq_len),
    "DKVMN": lambda h: DKVMN(N_SKILLS, h),
    "PCDT":  lambda h: PCDT(N_SKILLS, h, dims=CFG.pcdt_dimensions,
                            use_persistence=CFG.use_persistence,
                            n_behav=len(BEHAV_COLS), tau0=CFG.tau0_seconds),
}

CAPACITY = {}
print(f"Capacity matching to {CFG.param_budget:,} parameters (+/- {CFG.param_tolerance:.0%})\n")
for name, b in BUILDERS.items():
    h, n, ok = match_capacity(b, CFG.param_budget, CFG.param_tolerance)
    CAPACITY[name] = dict(hidden=h, n_params=n, within_tol=ok)
    print(f"  {name:6s} hidden={h:4d}  params={n:9,}  "
          f"{'OK' if ok else 'OUTSIDE TOLERANCE -- report explicitly'}")

# Fail loudly rather than silently comparing models of wildly different size.
_failed = [k for k, v in CAPACITY.items() if not v["within_tol"]]
if _failed:
    print(f"\n*** WARNING: capacity matching failed for {_failed}. ***")
    print("*** Any comparison involving these models confounds representation with capacity. ***")
    print("*** Inspect the builder's width constraints before interpreting Table 2.        ***")

_bkt_params = 4 * N_SKILLS
CAPACITY["BKT"] = dict(hidden=None, n_params=_bkt_params, within_tol=False)
print(f"  {'BKT':6s} params={_bkt_params:9,}  (4 per skill; excluded from matching by construction)")
```

## 9. Training Harness

One harness, used identically by every neural model: same optimiser (AdamW), same schedule, same gradient
clipping, same early-stopping criterion (validation AUC, `patience` epochs), same learning-rate search
grid. The only thing that varies between runs is the model class.

Loss is masked binary cross-entropy — padded positions contribute nothing to gradients or metrics.


```python
def masked_bce(logits, targets, mask):
    loss = F.binary_cross_entropy_with_logits(logits, targets, reduction="none")
    return (loss * mask).sum() / mask.sum().clamp(min=1.0)


@torch.no_grad()
def collect_predictions(model, loader, device):
    """Returns (probs, targets, learner_index) over all valid positions."""
    model.eval()
    P, T, Lidx = [], [], []
    for b in loader:
        b = {k: (v.to(device) if torch.is_tensor(v) else v) for k, v in b.items()}
        logits = model(b)
        tgt, msk = b["correct"][:, 1:], b["mask"][:, 1:]
        p = torch.sigmoid(logits)
        sel = msk > 0
        P.append(p[sel].cpu().numpy()); T.append(tgt[sel].cpu().numpy())
        rows = b["idx"].unsqueeze(1).expand_as(msk)[sel]
        Lidx.append(rows.cpu().numpy())
    return np.concatenate(P), np.concatenate(T), np.concatenate(Lidx)


def train_model(name, builder, hidden, cfg, loaders, device, lr):
    seed_everything(SEED)                                # identical init seed for every model
    model = builder(hidden).to(device)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=cfg.weight_decay)
    best_auc, best_state, bad, history = -np.inf, None, 0, []

    for ep in range(cfg.max_epochs):
        model.train()
        tot = 0.0
        for b in loaders["train"]:
            b = {k: (v.to(device) if torch.is_tensor(v) else v) for k, v in b.items()}
            logits = model(b)
            loss = masked_bce(logits, b["correct"][:, 1:], b["mask"][:, 1:])
            opt.zero_grad(); loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), cfg.grad_clip)
            opt.step(); tot += loss.item()

        p, t, _ = collect_predictions(model, loaders["val"], device)
        auc = roc_auc_score(t, p) if len(np.unique(t)) > 1 else float("nan")
        history.append(dict(epoch=ep, train_loss=tot/max(len(loaders["train"]),1), val_auc=auc))
        if auc > best_auc + 1e-5:
            best_auc, bad = auc, 0
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        else:
            bad += 1
            if bad >= cfg.patience: break

    if best_state is not None: model.load_state_dict(best_state)
    return model, best_auc, history

print("Training harness ready.")
```

## 10. Evaluation Metrics

All five metrics are computed by one function applied identically to every model's test predictions.

**Expected Calibration Error** is computed for **every** model. The original Table 2 template marked ECE as
unavailable for baselines; that would have made the proposed model's calibration claim unfalsifiable, since
there would be nothing to compare against. Every model here emits a probability, so every model has a
measurable calibration error. This deviation is deliberate and is documented in the manuscript.


```python
def expected_calibration_error(probs, targets, n_bins=15):
    edges = np.linspace(0.0, 1.0, n_bins + 1)
    ece, n = 0.0, len(probs)
    for lo, hi in zip(edges[:-1], edges[1:]):
        m = (probs > lo) & (probs <= hi) if lo > 0 else (probs >= lo) & (probs <= hi)
        if m.sum() == 0: continue
        ece += (m.sum()/n) * abs(targets[m].mean() - probs[m].mean())
    return float(ece)


def evaluate(probs, targets, n_bins=15):
    eps = 1e-7
    p = np.clip(probs, eps, 1-eps)
    return dict(
        auc      = float(roc_auc_score(targets, p)) if len(np.unique(targets)) > 1 else float("nan"),
        accuracy = float(((p >= 0.5).astype(float) == targets).mean()),
        rmse     = float(np.sqrt(np.mean((p - targets)**2))),
        nll      = float(-np.mean(targets*np.log(p) + (1-targets)*np.log(1-p))),
        ece      = expected_calibration_error(p, targets, n_bins),
        n        = int(len(targets)),
    )

print("Metric functions ready:", ["auc","accuracy","rmse","nll","ece"])
```

## 11. Running the Experiment

For each neural model: search the learning rate over the identical grid (selecting on validation AUC),
retrain at the best setting, then evaluate once on the held-out test split. **The test split is touched
exactly once per model**, after all selection is complete.

BKT is fitted by EM on the training split and evaluated on the same test learners.


```python
RESULTS, PREDICTIONS, HISTORIES = {}, {}, {}
t_start = time.time()

# ---------------- BKT ----------------
print("=" * 68); print("BKT (per-skill, EM)")
bkt = BKTPerSkill(N_SKILLS).fit(bkt_training_sequences(PACKS["train"]))
p_bkt, t_bkt, m_bkt = bkt.predict_pack(PACKS["test"])
sel = m_bkt > 0
probs, targs = p_bkt[sel], t_bkt[sel]
rows = np.repeat(np.arange(p_bkt.shape[0])[:, None], p_bkt.shape[1], axis=1)[sel]
RESULTS["BKT"] = evaluate(probs, targs, CFG.ece_bins)
PREDICTIONS["BKT"] = (probs, targs, rows)
print(f"  test AUC={RESULTS['BKT']['auc']:.4f}  ECE={RESULTS['BKT']['ece']:.4f}  n={RESULTS['BKT']['n']:,}")

# ---------------- neural models ----------------
for name in ["DKT", "SAKT", "AKT", "DKVMN", "PCDT"]:
    print("=" * 68); print(name)
    hidden = CAPACITY[name]["hidden"]
    best = dict(auc=-np.inf)
    for lr in CFG.search_lrs:                       # identical search budget for every model
        _, val_auc, _ = train_model(name, BUILDERS[name], hidden, CFG, LOADERS, DEVICE, lr)
        print(f"    lr={lr:<7g} val AUC={val_auc:.4f}")
        if val_auc > best["auc"]: best = dict(auc=val_auc, lr=lr)

    model, val_auc, hist = train_model(name, BUILDERS[name], hidden, CFG, LOADERS, DEVICE, best["lr"])
    probs, targs, rows = collect_predictions(model, LOADERS["test"], DEVICE)
    RESULTS[name] = evaluate(probs, targs, CFG.ece_bins)
    RESULTS[name]["best_lr"] = best["lr"]
    PREDICTIONS[name] = (probs, targs, rows)
    HISTORIES[name] = hist
    print(f"  selected lr={best['lr']}  val AUC={val_auc:.4f}")
    print(f"  test AUC={RESULTS[name]['auc']:.4f}  ECE={RESULTS[name]['ece']:.4f}  n={RESULTS[name]['n']:,}")

print("=" * 68)
print(f"total runtime: {(time.time()-t_start)/60:.1f} min")
```

## 12. Results — Manuscript Table 2

Best value per column is marked. Note the direction of each metric: **AUC and Accuracy are
higher-is-better; RMSE, NLL and ECE are lower-is-better.**


```python
LOWER_IS_BETTER = {"rmse", "nll", "ece"}

rows = []
for name in MODELS:
    r = RESULTS[name]
    rows.append({
        "Model":    {"BKT":"Tuned Bayesian KT (per-skill)", "DKT":"Deep KT (recurrent)",
                     "SAKT":"Self-attentive KT", "AKT":"Attentive KT",
                     "DKVMN":"Memory-augmented KT", "PCDT":"PCDT (audited subset)"}[name],
        "Params":   CAPACITY[name]["n_params"],
        "AUC-ROC":  r["auc"], "Accuracy": r["accuracy"],
        "RMSE":     r["rmse"], "NLL": r["nll"], "ECE": r["ece"],
    })
table2 = pd.DataFrame(rows)

def mark_best(df):
    out = df.copy()
    for col in ["AUC-ROC","Accuracy","RMSE","NLL","ECE"]:
        vals = df[col].astype(float)
        best = vals.min() if col.lower() in LOWER_IS_BETTER else vals.max()
        out[col] = [f"**{v:.4f}**" if np.isclose(v, best) else f"{v:.4f}" for v in vals]
    out["Params"] = [f"{p:,}" for p in df["Params"]]
    return out

table2_fmt = mark_best(table2)
print("Table 2. Next-interaction predictive performance "
      "(standardised pipeline, capacity-matched neural models)\n")
print(table2_fmt.to_markdown(index=False))

table2.to_csv("table2_raw.csv", index=False)
table2_fmt.to_csv("table2_formatted.csv", index=False)
print("\nsaved: table2_raw.csv, table2_formatted.csv")
```

## 13. Statistical Comparison

Paired bootstrap resampling **over test learners** — the independent sampling unit. Resampling individual
interactions would violate independence (interactions within a learner are correlated) and produce
misleadingly narrow intervals.

Reported per baseline: $\Delta\text{AUC}$ (PCDT − baseline), its 95% percentile interval, a two-sided
bootstrap $p$-value, and the Holm–Bonferroni corrected $p$ across the family of comparisons.

The final column applies the **pre-registered** practical-significance rule: a difference below
$\Delta_{\min} = 0.010$ is reported as negligible regardless of its $p$-value.


```python
def paired_bootstrap_auc(pred_a, pred_b, n_boot, rng):
    """Paired bootstrap over learners. Each pred is (probs, targets, learner_row_index)."""
    pa, ta, ra = pred_a
    pb, tb, rb = pred_b
    learners = np.intersect1d(np.unique(ra), np.unique(rb))
    idx_a = {l: np.where(ra == l)[0] for l in learners}
    idx_b = {l: np.where(rb == l)[0] for l in learners}

    deltas = []
    for _ in range(n_boot):
        samp = rng.choice(learners, size=len(learners), replace=True)
        ia = np.concatenate([idx_a[l] for l in samp])
        ib = np.concatenate([idx_b[l] for l in samp])
        if len(np.unique(ta[ia])) < 2 or len(np.unique(tb[ib])) < 2: continue
        deltas.append(roc_auc_score(ta[ia], pa[ia]) - roc_auc_score(tb[ib], pb[ib]))
    d = np.asarray(deltas)
    lo, hi = np.percentile(d, [2.5, 97.5])
    p = 2 * min((d <= 0).mean(), (d >= 0).mean())
    return float(d.mean()), float(lo), float(hi), float(min(p, 1.0))


def holm_bonferroni(pvals):
    order = np.argsort(pvals); m = len(pvals)
    adj = np.empty(m); running = 0.0
    for rank, i in enumerate(order):
        running = max(running, (m - rank) * pvals[i])
        adj[i] = min(running, 1.0)
    return adj

baselines = [m for m in MODELS if m != "PCDT"]
raw = []
for b in baselines:
    d, lo, hi, p = paired_bootstrap_auc(PREDICTIONS["PCDT"], PREDICTIONS[b], CFG.n_bootstrap, RNG)
    raw.append(dict(baseline=b, delta=d, lo=lo, hi=hi, p=p))

adj = holm_bonferroni(np.array([r["p"] for r in raw]))
for r, a in zip(raw, adj): r["p_holm"] = float(a)

stats_rows = []
for r in raw:
    sig = r["p_holm"] < 0.05
    meaningful = abs(r["delta"]) >= CFG.min_meaningful_auc_delta
    if sig and meaningful:   verdict = "significant AND practically meaningful"
    elif sig:                verdict = f"significant but < {CFG.min_meaningful_auc_delta} (negligible)"
    else:                    verdict = "not significant"
    stats_rows.append({"Baseline": r["baseline"],
                       "dAUC (PCDT - baseline)": f"{r['delta']:+.4f}",
                       "95% CI": f"[{r['lo']:+.4f}, {r['hi']:+.4f}]",
                       "p (raw)": f"{r['p']:.4f}", "p (Holm)": f"{r['p_holm']:.4f}",
                       "Interpretation": verdict})

stats_table = pd.DataFrame(stats_rows)
print("Paired bootstrap over test learners "
      f"({CFG.n_bootstrap:,} resamples, Holm-Bonferroni corrected)\n")
print(stats_table.to_markdown(index=False))
stats_table.to_csv("table2_statistics.csv", index=False)

best_base = max(baselines, key=lambda b: RESULTS[b]["auc"])
delta_best = RESULTS["PCDT"]["auc"] - RESULTS[best_base]["auc"]
print(f"\nStrongest baseline: {best_base} (AUC={RESULTS[best_base]['auc']:.4f})")
print(f"PCDT - strongest baseline: {delta_best:+.4f} AUC")
print(f"H1 {'SUPPORTED' if delta_best >= CFG.min_meaningful_auc_delta else 'NOT SUPPORTED'} "
      f"at the pre-registered threshold of {CFG.min_meaningful_auc_delta}")
```

## 14. Ablation — Which Component Carries the Effect?

Two ablations isolate the PCDT model's two claimed differences from the baselines:

1. **− persistence ($\Psi$)** — identical architecture with the decay operator disabled. Isolates the
   contribution of modelling elapsed time between interactions.
2. **− behavioural dimension ($B_t$)** — state restricted to $C_t$ only. Isolates the contribution of the
   multidimensionality itself.

If neither ablation degrades performance, the PCDT model's advantage (if any) does not come from the
mechanisms it claims, and the manuscript must say so.


```python
ABLATIONS = {
    "PCDT (full)":        dict(dims=CFG.pcdt_dimensions, persistence=CFG.use_persistence),
    "PCDT - persistence": dict(dims=CFG.pcdt_dimensions, persistence=False),
    "PCDT - B_t (C only)":dict(dims=("C_t",),            persistence=CFG.use_persistence),
}

abl_rows = []
for label, spec in ABLATIONS.items():
    if label == "PCDT (full)":
        r = RESULTS["PCDT"]; nparam = CAPACITY["PCDT"]["n_params"]
    else:
        builder = lambda h, s=spec: PCDT(N_SKILLS, h, dims=s["dims"],
                                         use_persistence=s["persistence"],
                                         n_behav=len(BEHAV_COLS), tau0=CFG.tau0_seconds)
        h, nparam, _ = match_capacity(builder, CFG.param_budget, CFG.param_tolerance)
        model, _, _ = train_model(label, builder, h, CFG, LOADERS, DEVICE,
                                  RESULTS["PCDT"].get("best_lr", CFG.lr))
        p, t, _ = collect_predictions(model, LOADERS["test"], DEVICE)
        r = evaluate(p, t, CFG.ece_bins)
    abl_rows.append({"Variant": label, "Params": f"{nparam:,}",
                     "AUC-ROC": f"{r['auc']:.4f}", "RMSE": f"{r['rmse']:.4f}",
                     "NLL": f"{r['nll']:.4f}", "ECE": f"{r['ece']:.4f}",
                     "dAUC vs full": f"{r['auc']-RESULTS['PCDT']['auc']:+.4f}"})

ablation_table = pd.DataFrame(abl_rows)
print("Ablation of the PCDT model's two claimed mechanisms\n")
print(ablation_table.to_markdown(index=False))
ablation_table.to_csv("table2_ablation.csv", index=False)
```

## 15. Visualisation

Discrimination and calibration are shown side by side. A model can win on AUC while being poorly
calibrated, and reliability diagrams make that visible in a way scalar metrics do not.


```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(17, 4.6))
order = sorted(MODELS, key=lambda m: RESULTS[m]["auc"])
cols = ["#c0392b" if m == "PCDT" else "#34495e" for m in order]

axes[0].barh(order, [RESULTS[m]["auc"] for m in order], color=cols, edgecolor="black", lw=.6)
axes[0].set_xlim(0.5, max(RESULTS[m]["auc"] for m in MODELS) + 0.03)
axes[0].axvline(0.5, ls=":", c="grey"); axes[0].set_xlabel("Test AUC-ROC")
axes[0].set_title("Discrimination (higher is better)")

order_e = sorted(MODELS, key=lambda m: -RESULTS[m]["ece"])
cols_e = ["#c0392b" if m == "PCDT" else "#34495e" for m in order_e]
axes[1].barh(order_e, [RESULTS[m]["ece"] for m in order_e], color=cols_e, edgecolor="black", lw=.6)
axes[1].set_xlabel("Expected Calibration Error"); axes[1].set_title("Calibration (lower is better)")

for m in MODELS:
    p, t, _ = PREDICTIONS[m]
    bins = np.linspace(0, 1, 11); mids, accs = [], []
    for lo, hi in zip(bins[:-1], bins[1:]):
        sel = (p >= lo) & (p < hi)
        if sel.sum() > 30:
            mids.append(p[sel].mean()); accs.append(t[sel].mean())
    axes[2].plot(mids, accs, "o-", lw=2.2 if m == "PCDT" else 1.2,
                 color="#c0392b" if m == "PCDT" else None,
                 alpha=1.0 if m == "PCDT" else .65, label=m)
axes[2].plot([0,1], [0,1], "k--", lw=1, label="perfect")
axes[2].set_xlabel("mean predicted probability"); axes[2].set_ylabel("observed frequency")
axes[2].set_title("Reliability diagram"); axes[2].legend(fontsize=8)

plt.tight_layout(); plt.show()
```

## 16. Interpretation

> Written before the run, so the interpretation is not fitted to whatever appeared. Read the tables above
> for the actual numbers.

### If $H_1$ is supported ($\Delta\mathrm{AUC} \ge 0.010$ over the strongest tuned baseline)

The claim is licensed *only* because the comparison was capacity-matched and the baselines were tuned on an
identical grid. Report the effect **with its confidence interval**, and attribute it to a mechanism only if
the ablation in §14 supports that attribution. If removing $\Psi$ does not degrade performance, the gain
did not come from persistence, whatever the model is called.

### If $H_1$ is not supported (the more likely outcome)

This is a publishable result and should not be buried. It says: *a construct-validity-audited
multidimensional representation does not improve next-step correctness prediction over well-tuned
knowledge-only baselines.* Three things follow.

1. **It is consistent with RQ2.1.** The audit found essentially one dimension ($C_t$) with real-data
   validity, and $C_t$ is precisely what the baselines already model. A large gain would have been
   *surprising*, and a large gain in the absence of construct validity would have warranted suspicion of a
   pipeline artifact rather than celebration.

2. **Redirect the contribution, do not inflate it.** The representation's justification moves to
   persistence (RQ2.3), stability (RQ2.4), and downstream decision quality — where the baselines have no
   mechanism at all — rather than to benchmark accuracy.

3. **Check calibration before conceding.** AUC parity with materially better ECE is a real and defensible
   contribution: a model that knows when it is uncertain is more useful for instructional decision-making
   than one that is marginally better at ranking. This is the strongest available fallback claim, and it is
   only checkable because ECE was computed for every model.

### On the negative $B_t$ signal

The RQ2.1 audit found $B_t$'s convergent validity to be negative ($r = -0.233$): longer response times and
more attempts indicate *struggle*, not proficiency. If including $B_t$ helps here, it is because struggle is
predictive of failure — a legitimate signal. Do not describe it as "behavioural engagement improves
prediction"; describe it as what it is.

### On BKT

If per-skill EM-fitted BKT lands close to the deep models, that is a finding worth stating plainly rather
than an embarrassment. Standardised benchmarking has repeatedly shown the deep-vs-classical gap narrows
under fair tuning, and reproducing that here strengthens the credibility of the whole comparison.


## 17. Limitations and Threats to Validity

### Construct and measurement
1. **Predicting an observable, not cognition.** High AUC demonstrates predictive validity of a
   representation, never that its internal state corresponds to $S_t$.
2. **Inherited attrition.** Rows with unusable correctness and problems without skill tags are removed
   upstream; the population is not the full ASSISTments population.
3. **Primary-skill-only encoding.** Multi-skill items are represented by their first tag, discarding
   secondary skill information that the RQ2.1 aggregate analysis retained.

### Internal
4. **Reimplementations, not reproductions.** Baselines are standardised reimplementations under a matched
   budget; absolute values are not comparable to published figures, only to each other within this pipeline.
5. **Search budget is small.** Three learning rates per model is an identical but *modest* budget; a larger
   search could shift the ordering, and this bounds the strength of any "we beat X" claim.
6. **Single split.** One learner-level partition. Repeated splits or $k$-fold would give a more robust
   estimate of variance across partitions than the within-split bootstrap used here.
7. **Subsampling.** `max_learners` limits the training population for tractability; results at full scale
   may differ, and the final manuscript run should use the complete dataset.

### Statistical
8. **Bootstrap over learners captures sampling variance, not initialisation variance.** Seeds are fixed, so
   run-to-run architectural variance is not estimated; multiple seeds per model would be stronger.
9. **Large $n$ makes trivial differences significant** — which is exactly why a pre-registered minimum
   meaningful effect is applied alongside every $p$-value.

### Scope
10. **Log-only features.** No multimodal signals; conclusions do not transfer to instrumented tutoring
    contexts with gaze, affect, or physiological input.
11. **One dataset.** Cross-dataset replication (EdNet, Junyi) is required before any generalisation claim,
    and is the single most valuable next addition to this experiment.

**Reproducibility guarantees in this notebook**

- **One seed** (`SEED`) re-applied before *every* model's initialisation, so architectures are compared
  under identical initial conditions rather than different random draws.
- **No leakage:** splits are by learner; the skill vocabulary and all feature-scaling statistics are
  derived from the training split only; the test split is evaluated exactly once per model after selection.
- **Identical treatment:** same optimiser, schedule, clipping, early-stopping rule, and learning-rate grid
  for every neural model; capacity matched to a common budget.
- **Pre-registered** thresholds and minimum meaningful effect fixed in §4 before training.
- **Artifacts:** `table2_raw.csv`, `table2_formatted.csv`, `table2_statistics.csv`,
  `table2_ablation.csv`, `table2_provenance.csv`.
- **Determinism caveat:** exact reproduction requires the same library versions and hardware; cuDNN kernel
  selection can vary GPU results in the last decimal places.


## 19. Conclusion

This experiment produces manuscript **Table 2** under conditions designed to make a favourable result hard
to obtain: capacity-matched neural models, an identically-tuned baseline family including a properly
EM-fitted per-skill BKT, a single standardised pipeline, learner-level splits, and a pre-registered minimum
meaningful effect that prevents a statistically-significant-but-trivial difference from being reported as a
contribution.

The PCDT model is constrained to the dimensions that **survived the RQ2.1 audit**, so this is a genuine test
of whether construct-valid multidimensionality plus explicit persistence buys predictive accuracy — not a
test of whether adding features helps.

### Feeding the rest of Paper 2

- **If the effect is null or negligible**, RQ2.2 is answered honestly in the negative, and the
  representation's contribution rests on RQ2.3 (persistence) and RQ2.4 (stability) — which is precisely why
  those research questions exist as separate claims rather than as corollaries of predictive performance.
- **If calibration improves at AUC parity**, that becomes the reportable contribution and should be carried
  into Paper 3a, where a well-calibrated belief $b_t$ is a prerequisite rather than a nicety.
- **Either way**, the ablation in §14 determines whether any observed effect is attributable to the
  mechanisms the paper claims, which is what a reviewer will actually test the argument on.

### Immediate next steps

1. Re-run with `max_learners` raised to the full population and the complete ASSISTments release.
2. Replicate on a second dataset (EdNet KT1 or Junyi) before making any generalisation claim.
3. Add multiple seeds per model to estimate initialisation variance alongside sampling variance.
4. Carry the calibration result into RQ2.3's persistence evaluation.
