# Experiment P2-03 — Cross-Session Persistence of the PCDT Representation

**Paper 2:** *A Persistent Multidimensional State Representation for Cognitive Digital Twins*
**Research question:** RQ2.3
**Produces:** Manuscript **Table 4** (data-availability prerequisite) and **Table 5** (cross-session
predictive validity at gap horizons)
**Depends on:** Experiment P2-01 (RQ2.1 audit) and Experiment P2-02 (trained models + cached
test-set predictions)

---

> **RQ2.3 —** Under what specification of a persistence mechanism does the representation preserve
> cross-session predictive validity — measured by degradation at 1-day, 1-week, and 4-week
> inter-session gaps — with greater retention than session-independent baselines?

**Table 4 can fail, and that is its purpose.** It is a prerequisite, not a result. If the target data
does not contain enough learners returning after four weeks, no four-week persistence claim can be made
from it, and the notebook will say so and refuse to emit that row of Table 5 rather than reporting an
AUC computed on 80 interactions. Section 4.3 of the manuscript already pre-commits to that fallback;
this notebook enforces it.


## 0. Three things to read before running

### 0.1 The persistence operator never sees the gap it is predicting across

This is a defect in the Experiment P2-02 model, and it lands precisely on RQ2.3. In that `forward` loop:

```
h = psi(h, dt[t])        # decay across the gap BEFORE interaction t
h = cell(x_t, h)         # absorb interaction t
predict a_{t+1}          # <-- dt[t+1], the gap being predicted across, is never applied
```

The state used to predict the first item after a four-week absence is the state as of the learner's
*last interaction before* the absence, undecayed. $\Psi$ is applied one step too late: the gap only
influences predictions from the *following* step onward, by which point fresh evidence has arrived and
the persistence question is moot.

This barely affects Table 2 — the state still carries gap-decayed history, which is consistent with the
$-0.0004$ AUC that removing $\Psi$ produced — but it guts the mechanism that RQ2.3 is about.

`CFG.psi_readout_decay = True` adds a readout-time decay $\tilde{h}_t = \Psi(h_t, \Delta t_{t+1})$ used
only for the prediction. This is legitimate rather than leakage: in a tutoring system the elapsed time
since the learner's last interaction is known the moment they return, before they answer anything. The
notebook reports **both** variants, because the as-run model is the one Tables 2 and 3 describe.

**If the corrected variant wins, it becomes the paper's model and Tables 2 and 3 must be regenerated
with it.** One model, all tables.

### 0.2 Table 4 names datasets this project has not used

Manuscript §5.1 specifies ASSISTments 2012 and EdNet KT1; the executed experiments used ASSISTments
2020–2021. The dataset registry below scans whatever is actually present and reports unscanned
manuscript datasets as `not scanned`, never as a number. This is the same divergence flagged for Table 3
and it has to be resolved in the manuscript, not in code — note in particular that ASSISTments 2020–2021
Feb–Apr covers roughly twelve weeks, so a four-week gap must fit inside a short window, and thin
four-week cells are the expected outcome rather than a surprise.

### 0.3 AUC is not comparable across gap buckets

Learners who return after four weeks are a self-selected population, the items they meet are not drawn
from the same difficulty distribution, and the base rate differs. So the *level* of AUC in the 4-week
bucket says little on its own.

What is comparable is the **paired difference between two models scored on identical prediction points**.
The retention claim is therefore operationalised as a difference-in-differences,

$$\mathrm{RA}(h) \;=\; \bigl[\mathrm{AUC}_{\mathrm{PCDT}}(h) - \mathrm{AUC}_{\mathrm{base}}(h)\bigr]
\;-\; \bigl[\mathrm{AUC}_{\mathrm{PCDT}}(\mathrm{ref}) - \mathrm{AUC}_{\mathrm{base}}(\mathrm{ref})\bigr],$$

with the within-session bucket as reference: *how much less does the PCDT degrade than a
session-independent model, over and above any advantage it already had within session.* Reporting the
raw per-bucket AUC gap instead would let a stable baseline advantage masquerade as retention.

## 2. Configuration and Pre-Registered Decisions

**The split parameters must match Experiment P2-02 exactly.** They are what makes the cached
predictions from that experiment reusable here: same subsample, same learner-level split, same
sequence packing, so a cached probability vector still lines up with the interaction it was produced
for. The notebook verifies the alignment rather than assuming it, and refuses to proceed if it breaks.

**Gap buckets are fixed before looking at the data**, left-closed and right-open so that every
prediction falls in exactly one bucket and nothing is double-counted:

| Bucket | Range | Role |
|---|---|---|
| `within` | $[0,\ 20\ \mathrm{min})$ | reference for the difference-in-differences |
| `sameday` | $[20\ \mathrm{min},\ 1\ \mathrm{d})$ | reported in the extended table only |
| `d1` | $[1\ \mathrm{d},\ 7\ \mathrm{d})$ | manuscript row "1 day" |
| `d7` | $[7\ \mathrm{d},\ 28\ \mathrm{d})$ | manuscript row "1 week" |
| `d28` | $[28\ \mathrm{d},\ \infty)$ | manuscript row "4 weeks" |

**Adequacy thresholds are pre-registered too.** Table 4's "adequate for evaluation?" column is a
judgement that becomes unfalsifiable if it is made after seeing the counts. A bucket is adequate only
if it clears `min_learners_bucket` *and* `min_points_bucket` *and* contains both outcome classes. The
achieved bootstrap precision is reported alongside as supporting evidence, but it is not what decides
the verdict.


````python
@dataclass(frozen=True)
class Config:
    # ---------- data / split: MUST match Experiment P2-02 or cached predictions will not align ----
    max_learners: int = 20_000
    min_seq_len: int = 10
    max_seq_len: int = 200
    train_frac: float = 0.70
    val_frac: float = 0.10

    # ---------- gap buckets (pre-registered, seconds) ----------
    within_session_max_s: float = 20 * 60
    day_s: float = 24 * 3600.0

    # ---------- adequacy thresholds (pre-registered BEFORE inspecting the data) ----------
    min_learners_dataset: int = 200      # Table 4: learners in the whole dataset with such a gap
    min_learners_bucket: int = 300       # Table 5: distinct test learners contributing to the bucket
    min_points_bucket: int = 2_000       # Table 5: prediction points in the bucket

    # ---------- prediction unit ----------
    first_post_gap_only: bool = True     # score only the FIRST item after the gap

    # ---------- models ----------
    psi_readout_decay: bool = False      # corrected variant; True adds ~1 training run (see 0.1)
    train_if_missing: bool = True        # train when a cached prediction set is absent
    param_budget: int = 350_000
    param_tolerance: float = 0.05
    pcdt_dimensions: Sequence[str] = ("C_t", "B_t")
    use_persistence: bool = True
    tau0_seconds: float = 3600.0
    batch_size: int = 64
    lr: float = 1e-3
    weight_decay: float = 1e-5
    max_epochs: int = 30
    patience: int = 5
    grad_clip: float = 5.0
    search_lrs: Sequence[float] = (5e-4, 1e-3, 2e-3)

    # ---------- evaluation ----------
    n_bootstrap: int = 2_000
    min_meaningful_auc_delta: float = 0.010     # pre-registered in Experiment P2-02
    ece_bins: int = 15

    # ---------- synthetic fallback (used only if the 4-week cell is inadequate) ----------
    run_synthetic_fallback: bool = True
    sim_learners: int = 4_000
    sim_skills: int = 40
    sim_decay_d: float = 0.5             # ACT-R base-level decay exponent

    # ---------- caching (must match Experiment P2-02) ----------
    cache_dir: Optional[str] = "PCDT_P2_cache"

CFG = Config()

BUCKETS = [
    ("within",  0.0,                       CFG.within_session_max_s, "within session (<20 min)"),
    ("sameday", CFG.within_session_max_s,  CFG.day_s,                "same day (>=20 min)"),
    ("d1",      CFG.day_s,                 7  * CFG.day_s,           "1 day"),
    ("d7",      7  * CFG.day_s,            28 * CFG.day_s,           "1 week"),
    ("d28",     28 * CFG.day_s,            np.inf,                   "4 weeks"),
]
MANUSCRIPT_BUCKETS = ["d1", "d7", "d28"]
REF_BUCKET = "within"

if CFG.cache_dir:
    try: os.makedirs(CFG.cache_dir, exist_ok=True)
    except OSError as e: print(f"cache directory unavailable ({e}); caching disabled")

print("Pre-registered before any analysis:")
for k, lo, hi, lab in BUCKETS:
    hi_s = "inf" if not np.isfinite(hi) else f"{hi/CFG.day_s:.3g}d"
    print(f"  bucket {k:8s} [{lo/CFG.day_s:.4g}d, {hi_s})  '{lab}'")
print(f"  adequacy      : >= {CFG.min_learners_bucket} test learners AND "
      f">= {CFG.min_points_bucket:,} prediction points per bucket")
print(f"  prediction unit: {'first item after the gap only' if CFG.first_post_gap_only else 'all items in the returning session'}")
print(f"  psi readout decay (see 0.1): {CFG.psi_readout_decay}")
````

## 3. Dataset Registry

Each dataset is registered with a loader that returns a frame of
`learner_id | item_id | skill_id | start_time | correct | response_time | hint_frac | attempt_count`.
Only datasets whose files are actually present are scanned; the rest appear in Table 4 as
`not scanned`, which is the honest entry for a dataset nobody has looked at.

The ASSISTments 2020–2021 loader is the cleaning pipeline from Experiments P2-01 and P2-02, unchanged.
The ASSISTments 2012 and EdNet KT1 loaders are best-effort: they auto-detect the usual column names and
fail loudly with the columns they did find, rather than silently producing a frame with the wrong
timestamp semantics. **Verify the timestamp units before trusting any gap statistic** — EdNet stores
milliseconds since epoch, and an unnoticed factor of 1000 turns every four-week gap into forty minutes.


````python
def _parse_skills(v):
    if isinstance(v, list): return v
    if isinstance(v, str) and v.strip().startswith("["):
        try: return ast.literal_eval(v)
        except (ValueError, SyntaxError): return []
    return [] if pd.isna(v) else [v]


STD_COLS = ["learner_id", "item_id", "skill_id", "start_time", "correct",
            "response_time", "hint_frac", "attempt_count"]


def load_assistments_2021(paths):
    """ASSISTments 2020-2021 -- identical cleaning to Experiments P2-01 / P2-02."""
    plogs = pd.read_csv(paths["plogs"],
                        usecols=["student_id","problem_id","start_time","time_on_task",
                                 "attempt_count","fraction_of_hints_used","correct"])
    pdets = pd.read_csv(paths["pdets"], usecols=["problem_id","skills"])
    pdets["skills"] = pdets["skills"].map(_parse_skills)

    df = plogs
    df["correct"] = pd.to_numeric(
        df["correct"].map({True:1.0, False:0.0, "True":1.0, "False":0.0}).fillna(df["correct"]),
        errors="coerce")
    for c in ("time_on_task","attempt_count","fraction_of_hints_used"):
        df[c] = pd.to_numeric(df[c], errors="coerce")
    df = df.dropna(subset=["correct"])
    df["start_time"] = pd.to_datetime(df["start_time"], format="ISO8601", utc=True)

    prim = pdets.copy()
    prim["skill_id"] = prim["skills"].map(lambda s: s[0] if isinstance(s, list) and len(s) else np.nan)
    prim = prim.dropna(subset=["skill_id"])[["problem_id","skill_id"]]
    df = df.merge(prim, on="problem_id", how="inner")
    df = df.rename(columns={"student_id":"learner_id","problem_id":"item_id",
                            "time_on_task":"response_time","fraction_of_hints_used":"hint_frac"})
    return df[STD_COLS]


def _pick(cols, *cands):
    for c in cands:
        if c in cols: return c
    return None


def load_assistments_2012(paths):
    """ASSISTments 2012-2013. Column names vary by release; detected, not assumed."""
    df = pd.read_csv(paths["csv"], low_memory=False)
    cols = set(df.columns)
    m = dict(learner=_pick(cols, "user_id", "student_id"),
             item=_pick(cols, "problem_id", "assistment_id"),
             skill=_pick(cols, "skill_id", "skill", "skill_name"),
             time=_pick(cols, "start_time", "startTime", "timestamp"),
             correct=_pick(cols, "correct"),
             rt=_pick(cols, "ms_first_response", "overlap_time", "time_on_task"),
             hint=_pick(cols, "hint_count", "hint_total"),
             att=_pick(cols, "attempt_count",))
    missing = [k for k in ("learner","item","time","correct") if m[k] is None]
    if missing:
        raise KeyError(f"ASSISTments 2012: could not find {missing}. Columns present: {sorted(cols)[:40]}")
    out = pd.DataFrame({
        "learner_id": df[m["learner"]], "item_id": df[m["item"]],
        "skill_id": df[m["skill"]] if m["skill"] else "UNK",
        "start_time": pd.to_datetime(df[m["time"]], utc=True, errors="coerce"),
        "correct": pd.to_numeric(df[m["correct"]], errors="coerce"),
        "response_time": pd.to_numeric(df[m["rt"]], errors="coerce")/1000.0 if m["rt"] else np.nan,
        "hint_frac": pd.to_numeric(df[m["hint"]], errors="coerce") if m["hint"] else np.nan,
        "attempt_count": pd.to_numeric(df[m["att"]], errors="coerce") if m["att"] else np.nan,
    })
    return out.dropna(subset=["correct","start_time"])[STD_COLS]


def load_ednet_kt1(paths):
    """EdNet KT1. Timestamps are MILLISECONDS since epoch -- verify before trusting any gap."""
    import glob
    files = sorted(glob.glob(os.path.join(paths["dir"], "u*.csv")))[: paths.get("max_files", 20000)]
    if not files:
        raise FileNotFoundError(f"no u*.csv under {paths['dir']}")
    q = pd.read_csv(paths["questions"])          # question_id, correct_answer, tags, ...
    key = dict(zip(q["question_id"], q["correct_answer"]))
    tags = dict(zip(q["question_id"], q.get("tags", pd.Series(["UNK"]*len(q)))))
    frames = []
    for f in files:
        d = pd.read_csv(f)
        uid = os.path.splitext(os.path.basename(f))[0]
        frames.append(pd.DataFrame({
            "learner_id": uid, "item_id": d["question_id"],
            "skill_id": d["question_id"].map(tags).fillna("UNK").astype(str).str.split(";").str[0],
            "start_time": pd.to_datetime(d["timestamp"], unit="ms", utc=True),
            "correct": (d["user_answer"] == d["question_id"].map(key)).astype(float),
            "response_time": pd.to_numeric(d.get("elapsed_time"), errors="coerce")/1000.0,
            "hint_frac": np.nan, "attempt_count": 1.0}))
    return pd.concat(frames, ignore_index=True)[STD_COLS]


DATA_ROOT = "dataset"
REGISTRY = {
    "ASSISTments 2020-2021": dict(
        loader=load_assistments_2021,
        paths=dict(plogs=f"{DATA_ROOT}/feb_to_apr/plogs.csv",
                   pdets=f"{DATA_ROOT}/feb_to_apr/pdets.csv"),
        primary=True),
    "ASSISTments 2012": dict(
        loader=load_assistments_2012,
        paths=dict(csv=f"{DATA_ROOT}/2012-2013-data-with-predictions-4-final.csv"),
        primary=False),
    "EdNet KT1": dict(
        loader=load_ednet_kt1,
        paths=dict(dir=f"{DATA_ROOT}/EdNet/KT1", questions=f"{DATA_ROOT}/EdNet/contents/questions.csv",
                   max_files=20000),
        primary=False),
}

def registry_status():
    rows = []
    for name, spec in REGISTRY.items():
        present = all(os.path.exists(p) or os.path.isdir(p)
                      for k, p in spec["paths"].items() if isinstance(p, str))
        rows.append(dict(Dataset=name, Primary=spec["primary"], Files_present=present,
                         Path=list(spec["paths"].values())[0]))
    return pd.DataFrame(rows)

print(registry_status().to_markdown(index=False))
print("\nDatasets whose files are absent are reported as 'not scanned' in Table 4, never as a number.")
````

## 4. Gap Analysis and Table 4

For every learner the inter-interaction gaps $\Delta t = t_i - t_{i-1}$ are computed from sorted
timestamps and assigned to a bucket. Table 4 asks a *dataset-level* question — does this dataset
contain enough learners who ever return after each horizon — so a learner counts toward a horizon if
they have **at least one** gap that reaches it. That is deliberately more permissive than Table 5's
per-bucket prediction-point requirement, because a dataset can clear Table 4 (enough returning
learners exist) yet still fail Table 5 for a given horizon (too few of them land in the test split).

The verdict column applies the pre-registered `min_learners_dataset` threshold. Datasets that were not
scanned are carried through as `not scanned` so the manuscript's specified-but-unused datasets remain
visible instead of vanishing.


````python
def gap_profile(df):
    """Per-learner gap buckets. Returns (counts_dict, n_learners, median_span_days)."""
    df = df.dropna(subset=["start_time"]).sort_values(["learner_id", "start_time"])
    g = df.groupby("learner_id")["start_time"]
    dt = g.diff().dt.total_seconds()
    lid = df["learner_id"].values
    reach = {k: set() for k, *_ in BUCKETS}
    valid = ~dt.isna()
    dt_v, lid_v = dt.values[valid], lid[valid]
    for k, lo, hi, _ in BUCKETS:
        sel = (dt_v >= lo) & (dt_v < hi)
        reach[k].update(np.unique(lid_v[sel]).tolist())
    span = g.agg(lambda s: (s.max() - s.min()).total_seconds() / CFG.day_s)
    return {k: len(v) for k, v in reach.items()}, df["learner_id"].nunique(), float(span.median())


SCAN = {}
print("Scanning available datasets (this reads the raw files; may take a few minutes)\n")
for name, spec in REGISTRY.items():
    present = all(os.path.exists(p) or os.path.isdir(p)
                  for k, p in spec["paths"].items() if isinstance(p, str))
    if not present:
        SCAN[name] = dict(scanned=False)
        print(f"  {name:24s} not scanned (files absent)")
        continue
    try:
        t0 = time.time()
        df = spec["loader"](spec["paths"])
        counts, n_learn, med_span = gap_profile(df)
        SCAN[name] = dict(scanned=True, df=df, counts=counts, n_learners=n_learn,
                          median_span_days=med_span, n_interactions=len(df))
        print(f"  {name:24s} {len(df):>12,} interactions | {n_learn:>8,} learners | "
              f"median span {med_span:6.1f} d | {time.time()-t0:5.1f}s")
    except Exception as e:
        SCAN[name] = dict(scanned=False, error=str(e))
        print(f"  {name:24s} FAILED: {e}")
````

````python
def _verdict4(counts):
    d1  = counts.get("d1", 0)  + counts.get("d7", 0) + counts.get("d28", 0)   # >=1 day
    d7  = counts.get("d7", 0)  + counts.get("d28", 0)                          # >=1 week
    d28 = counts.get("d28", 0)                                                 # >=4 weeks
    ok = d28 >= CFG.min_learners_dataset
    partial = (d7 >= CFG.min_learners_dataset) and not ok
    verdict = ("yes" if ok else
               "1d/1wk only -- 4wk inadequate" if partial else
               "no -- insufficient returning learners")
    return d1, d7, d28, verdict

rows4 = []
for name in REGISTRY:
    s = SCAN.get(name, {})
    if not s.get("scanned"):
        why = "files absent" if "error" not in s else f"load failed"
        rows4.append({"Dataset": name, "Learners with >=1-day gaps": "not scanned",
                      "with >=1-week gaps": "not scanned", "with >=4-week gaps": "not scanned",
                      "Adequate for evaluation?": f"not scanned ({why})"})
        continue
    d1, d7, d28, verdict = _verdict4(s["counts"])
    rows4.append({"Dataset": name,
                  "Learners with >=1-day gaps": f"{d1:,}",
                  "with >=1-week gaps": f"{d7:,}",
                  "with >=4-week gaps": f"{d28:,}",
                  "Adequate for evaluation?": verdict})

table4 = pd.DataFrame(rows4)
print("Table 4. Data-availability check (prerequisite).\n")
print(table4.to_markdown(index=False))
table4.to_csv("table4.csv", index=False)
with open("table4.md", "w") as fh: fh.write(table4.to_markdown(index=False))

t4_notes = [
    f"A learner counts toward a horizon if they have at least one inter-interaction gap reaching it; "
    f"horizons are nested (>=4wk implies >=1wk implies >=1d).",
    f"Pre-registered adequacy threshold: >= {CFG.min_learners_dataset} learners in the "
    f">=4-week column. This is a dataset-level screen; Table 5 additionally requires enough of those "
    f"learners to fall in the held-out test split.",
    "Datasets marked 'not scanned' are specified in the manuscript (Section 5.1) but their files were "
    "not present in this run. They must be scanned, or the manuscript's dataset list amended, before "
    "submission.",
    "EdNet timestamps are milliseconds since epoch; the loader converts with unit='ms'. Confirm the "
    "median-span figure is plausible (weeks, not minutes) before trusting the >=4-week count.",
]
print("\nCaption notes:")
for i, n in enumerate(t4_notes, 1): print(f"  ({i}) {n}")
with open("table4_caption_notes.txt", "w") as fh:
    fh.write("\n".join(f"({i}) {n}" for i, n in enumerate(t4_notes, 1)))

primary = next(n for n, sp in REGISTRY.items() if sp["primary"])
if SCAN.get(primary, {}).get("scanned"):
    _, _, d28p, _ = _verdict4(SCAN[primary]["counts"])
    if d28p < CFG.min_learners_dataset:
        print(f"\n*** {primary}: only {d28p:,} learners with a >=4-week gap "
              f"(threshold {CFG.min_learners_dataset}). ***")
        print("*** The 4-week row of Table 5 will fall back to the synthetic simulator, and the "
              "external-validity limitation must be stated prominently (Section 4.3). ***")
````

## 9. ACT-R Synthetic Simulator (Fallback for Inadequate Horizons)

Used only where Table 4 shows a horizon is inadequate in the real data — on ASSISTments 2020–2021, that
is expected at four weeks because the release spans roughly twelve weeks. The simulator gives known
latent mastery with an ACT-R base-level decay,
$m \leftarrow m \cdot (1 + \Delta t/\tau)^{-d}$, so genuine multi-week gaps can be injected and the
persistence mechanism tested against a ground truth that real logs cannot provide.

Synthetic results demonstrate that the mechanism behaves correctly under controlled decay; they do
**not** establish real-world retention. Every synthetic row of Table 5 is tagged `(sim)` and the
simulator-to-reality gap is stated with it.


````python
def simulate_learners(n_learners=None, rng=None):
    """ACT-R-flavoured simulator with real multi-week gaps and known latent mastery."""
    n_learners = n_learners or CFG.sim_learners
    rng = rng or np.random.default_rng(SEED)
    skills = [f"S{i}" for i in range(CFG.sim_skills)]
    diff = {s: rng.normal(0, 0.8) for s in skills}
    t0 = pd.Timestamp("2024-01-01", tz="UTC")
    rows = []
    for lid in range(n_learners):
        ability = rng.normal(0, 1.0)
        mastery = {s: rng.normal(-0.5, 0.3) for s in skills}
        last_t = {s: None for s in skills}
        t = t0 + pd.Timedelta(days=int(rng.integers(0, 30)))
        n_sessions = int(rng.integers(4, 12))
        for _ in range(n_sessions):
            # inter-session gap: mixture giving days, weeks, and multi-week returns
            u = rng.random()
            gap_d = (rng.uniform(0, 1) if u < 0.4 else
                     rng.uniform(1, 7) if u < 0.7 else
                     rng.uniform(7, 28) if u < 0.9 else
                     rng.uniform(28, 70))
            t = t + pd.Timedelta(days=float(gap_d))
            for _ in range(int(rng.integers(5, 25))):
                s = skills[int(rng.integers(0, len(skills)))]
                if last_t[s] is not None:
                    dt_d = (t - last_t[s]).total_seconds()/CFG.day_s
                    mastery[s] *= (1.0 + dt_d/1.0) ** (-CFG.sim_decay_d)     # ACT-R base-level decay
                logit = ability + mastery[s] - diff[s]
                p = 1/(1+np.exp(-logit)); corr = int(rng.random() < np.clip(p, .02, .98))
                mastery[s] += 0.20 if corr else 0.06                         # learning
                last_t[s] = t
                rows.append(dict(learner_id=lid, item_id=int(rng.integers(0, 500)), skill_id=s,
                                 start_time=t, correct=float(corr),
                                 response_time=float(np.clip(rng.lognormal(3, .4), 2, 900)),
                                 hint_frac=float(np.clip(rng.beta(2,5), 0, 1)),
                                 attempt_count=float(1+rng.poisson(0.5))))
                t = t + pd.Timedelta(seconds=int(rng.integers(20, 240)))
    return pd.DataFrame(rows)

# only build/train on the simulator if a manuscript horizon is inadequate in the real data
def horizon_adequate_realdata(bucket_key):
    mflat = PACKS["test"]["mask"][:,1:] > 0
    bflat = TEST_BUCKET[:,:-1]
    idx = [i for i,(k,*_ ) in enumerate(BUCKETS) if k == bucket_key][0]
    if CFG.first_post_gap_only:
        sel = _first_post_gap_mask() & (bflat == idx)
    else:
        sel = mflat & (bflat == idx)
    n_pts = int(sel.sum())
    n_learn = int(np.unique(np.where(sel)[0]).size)
    return (n_learn >= CFG.min_learners_bucket) and (n_pts >= CFG.min_points_bucket), n_learn, n_pts

print("Simulator defined. It is invoked only for horizons Table 4 flags as inadequate.")
````

## 5. Sequence Packing, Aligned to the Cached Predictions

The persistence question is about the model as trained, so this notebook does not retrain the baselines
if it can avoid it. It rebuilds the **exact** sequence packing of Experiment P2-02 — same subsample under
the same seed, same learner-level split, same most-recent-window truncation — and then reloads that
experiment's cached test-set predictions and checks, position by position, that the reloaded
probabilities line up with the interactions the packing produced.

If the alignment check fails (different data, different split parameters, a stale cache), the notebook
stops rather than silently scoring one model's predictions against another model's interactions. A
per-position gap array `dt` is attached to every test interaction so each prediction can be routed to
its bucket. Only the **first prediction after a qualifying gap** is scored when
`first_post_gap_only=True`, which is the honest operationalisation of "predicting across the gap": later
items in the returning session are once again ordinary within-session predictions.


````python
BEHAV_COLS = ["response_time", "hint_frac", "attempt_count"]

def build_packs(df, cfg, rng):
    """Rebuild Experiment P2-02's packing. dt is retained per position for bucket routing."""
    df = df.sort_values(["learner_id", "start_time"]).copy()
    counts = df.groupby("learner_id").size()
    keep = counts[counts >= cfg.min_seq_len].index
    if len(keep) > cfg.max_learners:
        keep = rng.choice(np.asarray(keep), cfg.max_learners, replace=False)
    df = df[df.learner_id.isin(set(keep))].copy()

    learners = np.sort(df.learner_id.unique()); rng.shuffle(learners)
    n = len(learners); n_tr, n_va = int(cfg.train_frac*n), int(cfg.val_frac*n)
    split_of = {}
    for lid in learners[:n_tr]:            split_of[lid] = "train"
    for lid in learners[n_tr:n_tr+n_va]:   split_of[lid] = "val"
    for lid in learners[n_tr+n_va:]:       split_of[lid] = "test"
    df["split"] = df.learner_id.map(split_of)

    train_skills = df.loc[df.split=="train", "skill_id"].unique()
    skill2idx = {s: i+1 for i, s in enumerate(sorted(train_skills, key=str))}
    df["skill_idx"] = df.skill_id.map(skill2idx).fillna(0).astype(int)
    n_skills = len(skill2idx) + 1

    med = df[df.split=="train"][BEHAV_COLS].median()
    df[BEHAV_COLS] = df[BEHAV_COLS].fillna(med)
    df["response_time"] = np.log1p(df["response_time"].clip(lower=0))
    tr = df[df.split=="train"]
    mu, sd = tr[BEHAV_COLS].mean(), tr[BEHAV_COLS].std().replace(0, 1.0)
    df[BEHAV_COLS] = (df[BEHAV_COLS] - mu) / sd
    df["dt_sec"] = df.groupby("learner_id")["start_time"].diff().dt.total_seconds().fillna(0).clip(lower=0)

    packs = {}
    for sp in ("train","val","test"):
        sub = df[df.split==sp]; gids = sub.learner_id.unique(); L = cfg.max_seq_len
        S  = np.zeros((len(gids), L), np.int64);   A  = np.zeros((len(gids), L), np.float32)
        Fb = np.zeros((len(gids), L, len(BEHAV_COLS)), np.float32)
        DT = np.zeros((len(gids), L), np.float32); M  = np.zeros((len(gids), L), np.float32)
        for i, lid in enumerate(gids):
            g = sub[sub.learner_id==lid].tail(L); k = len(g)
            S[i,:k]=g.skill_idx.values; A[i,:k]=g.correct.values
            Fb[i,:k]=g[BEHAV_COLS].values; DT[i,:k]=g.dt_sec.values; M[i,:k]=1.0
        packs[sp] = dict(skill=S, correct=A, behav=Fb, dt=DT, mask=M, learner_id=gids)
    return packs, n_skills, skill2idx


PRIMARY = next(n for n, sp in REGISTRY.items() if sp["primary"])
USING_FIXTURE = not SCAN.get(PRIMARY, {}).get("scanned", False)

if USING_FIXTURE:
    print("!"*76)
    print(f"! {PRIMARY} not scanned -> building a synthetic fixture for pipeline verification only.")
    print("! Numbers produced are NOT manuscript results.")
    print("!"*76)
    # reuse the simulator defined in Section 9 so the fixture has realistic gaps
    _sim = simulate_learners(n_learners=1500, rng=RNG)
    PACKS, N_SKILLS, SKILL2IDX = build_packs(_sim, CFG, RNG)
else:
    PACKS, N_SKILLS, SKILL2IDX = build_packs(SCAN[PRIMARY]["df"], CFG, RNG)

for sp in ("train","val","test"):
    print(f"  {sp:5s}: {len(PACKS[sp]['learner_id']):,} learners | "
          f"{int(PACKS[sp]['mask'].sum()):,} interactions")
print(f"  skills (train-derived): {N_SKILLS:,}")

# per-position bucket id for the TEST pack (position t predicts t+1; gap is dt[:, t+1])
def bucket_of_seconds(x):
    out = np.full(x.shape, -1, np.int64)
    for bi, (k, lo, hi, _) in enumerate(BUCKETS):
        out[(x >= lo) & (x < hi)] = bi
    return out

_dt_pred = np.zeros_like(PACKS["test"]["dt"])
_dt_pred[:, :-1] = PACKS["test"]["dt"][:, 1:]     # gap being predicted ACROSS at each prediction pt
TEST_BUCKET = bucket_of_seconds(_dt_pred)         # [n_test, L], aligned to positions 0..L-2 predictions
print("test prediction points per bucket:",
      {BUCKETS[b][0]: int(((TEST_BUCKET[:, :-1] == b) & (PACKS['test']['mask'][:, 1:] > 0)).sum())
       for b in range(len(BUCKETS))})
````

## 6. Models

Two models are compared, differing only in whether the state persists across the gap.

**PCDT (with $\Psi$)** is the Experiment P2-02 model. `psi_readout_decay` optionally adds the
readout-time gap decay described in Section 0.1; with it `False` the class is byte-identical to the
model Tables 2 and 3 describe, and its cached predictions are reused directly.

**Session-independent baseline** is the honest control for a persistence claim. It is the *same*
architecture with the same capacity, but its state is **reset to zero at the start of every session** (a
gap $\ge$ 20 minutes marks a new session), and $\Psi$ is disabled. It therefore has no mechanism to
carry information across a gap — which is exactly the capability RQ2.3 tests. Comparing $\Psi$ against a
model that resets isolates the contribution of persistence, whereas comparing it against a plain DKT
(which carries a hidden state across gaps implicitly) would understate the baseline and inflate the
claim.


````python
class PCDT(nn.Module):
    """Experiment P2-02 PCDT with an optional readout-time gap decay (see Section 0.1).

    reset_sessions=True turns this into the session-independent control: the state is zeroed at the
    start of every session and Psi is disabled, so nothing crosses an inter-session gap."""
    def __init__(self, n_skills, hidden, dims=("C_t","B_t"), use_persistence=True,
                 n_behav=3, tau0=3600.0, dropout=0.2,
                 psi_readout_decay=False, reset_sessions=False,
                 session_gap_s=20*60):
        super().__init__()
        self.dims, self.use_persistence, self.tau0 = list(dims), use_persistence, tau0
        self.psi_readout_decay, self.reset_sessions = psi_readout_decay, reset_sessions
        self.session_gap_s = session_gap_s
        self.n_dims = len(self.dims)
        self.dim_size = max(8, hidden // self.n_dims)
        self.state_size = self.dim_size * self.n_dims
        self.inter_emb = nn.Embedding(2*n_skills + 1, hidden, padding_idx=0)
        self.ex_emb    = nn.Embedding(n_skills, hidden, padding_idx=0)
        self.behav_proj = nn.Linear(n_behav, hidden) if "B_t" in self.dims else None
        in_dim = hidden + (hidden if self.behav_proj is not None else 0)
        self.cell = nn.GRUCell(in_dim, self.state_size)
        self.log_lambda = nn.Parameter(torch.full((self.n_dims,), -1.0))
        self.drop = nn.Dropout(dropout)
        self.head = nn.Sequential(nn.Linear(self.state_size + hidden, hidden), nn.ReLU(),
                                  nn.Dropout(dropout), nn.Linear(hidden, 1))
        self.n_skills = n_skills

    def _decay(self, h, dt):
        lam = F.softplus(self.log_lambda)
        dt_scaled = torch.log1p(dt.clamp(min=0) / self.tau0).unsqueeze(-1)
        decay = torch.exp(-lam.unsqueeze(0) * dt_scaled).repeat_interleave(self.dim_size, dim=-1)
        return h * decay

    def forward(self, b):
        s, a, f, dt = b["skill"], b["correct"], b["behav"], b["dt"]
        B, L = s.shape
        inter = (s + a.long()*self.n_skills).clamp(max=self.inter_emb.num_embeddings-1)
        h = torch.zeros(B, self.state_size, device=s.device)
        outs = []
        for t in range(L - 1):
            x = self.inter_emb(inter[:, t])
            if self.behav_proj is not None:
                x = torch.cat([x, self.behav_proj(f[:, t])], dim=-1)

            if self.reset_sessions:
                new_sess = (dt[:, t] >= self.session_gap_s).float().unsqueeze(-1)
                h = h * (1.0 - new_sess)                       # zero the state at a session boundary
            elif self.use_persistence:
                h = self._decay(h, dt[:, t])                  # P2-02 behaviour: decay before evidence

            h = self.cell(x, h)

            h_read = h
            if self.psi_readout_decay and self.use_persistence and not self.reset_sessions:
                h_read = self._decay(h, dt[:, t+1])           # decay across the gap being predicted
            outs.append(self.head(torch.cat([self.drop(h_read), self.ex_emb(s[:, t+1])], dim=-1)))
        return torch.cat(outs, dim=1)


def count_params(m): return sum(p.numel() for p in m.parameters() if p.requires_grad)

def match_capacity(builder, budget, tol, lo=8, hi=512):
    best, best_gap = None, float("inf")
    while lo <= hi:
        mid = (lo + hi)//2
        try: n = count_params(builder(mid))
        except Exception: hi = mid-1; continue
        gap = abs(n-budget)
        if gap < best_gap: best, best_gap = (mid, n), gap
        if n < budget: lo = mid+1
        else: hi = mid-1
    h, n = best
    return h, n, abs(n-budget)/budget <= tol

def build_pcdt(h, **kw):
    return PCDT(N_SKILLS, h, dims=CFG.pcdt_dimensions, use_persistence=CFG.use_persistence,
                n_behav=len(BEHAV_COLS), tau0=CFG.tau0_seconds, **kw)

VARIANTS = {
    "PCDT (Psi)":              dict(psi_readout_decay=CFG.psi_readout_decay, reset_sessions=False),
    "Session-independent":     dict(psi_readout_decay=False, reset_sessions=True, use_persistence=False),
}
if CFG.psi_readout_decay:
    VARIANTS["PCDT (Psi, as-run)"] = dict(psi_readout_decay=False, reset_sessions=False)

CAPACITY = {}
for name, kw in VARIANTS.items():
    kw2 = {k: v for k, v in kw.items() if k != "use_persistence"}
    builder = lambda h, kk=kw2: build_pcdt(h, **kk)
    hh, nn_, ok = match_capacity(builder, CFG.param_budget, CFG.param_tolerance)
    CAPACITY[name] = dict(hidden=hh, n_params=nn_, within_tol=ok, kw=kw)
    print(f"  {name:22s} hidden={hh:4d}  params={nn_:9,}  {'OK' if ok else 'OUTSIDE TOL'}")
````

## 7. Training Harness and Prediction Reuse

`PCDT (Psi)` with `psi_readout_decay=False` is the Experiment P2-02 model, so its test predictions are
loaded from that experiment's cache when present and the alignment is verified against the freshly built
test pack. Everything genuinely new — the session-independent control, and the readout-decay variant if
enabled — is trained here under the identical P2-02 harness (same optimiser, schedule, early stopping,
learning-rate grid) and cached.


````python
def masked_bce(logits, targets, mask):
    loss = F.binary_cross_entropy_with_logits(logits, targets, reduction="none")
    return (loss * mask).sum() / mask.sum().clamp(min=1.0)

@torch.no_grad()
def collect_predictions(model, pack, device):
    model.eval()
    loader = DataLoader(_PackDS(pack), batch_size=CFG.batch_size, shuffle=False)
    P = np.full((pack["skill"].shape[0], pack["skill"].shape[1]-1), np.nan, np.float32)
    for b in loader:
        idx = b["idx"].numpy()
        b = {k: (v.to(device) if torch.is_tensor(v) else v) for k, v in b.items()}
        p = torch.sigmoid(model(b)).cpu().numpy()
        P[idx] = p
    return P                                        # [n, L-1], row-aligned to the pack

class _PackDS(Dataset):
    def __init__(self, pack): self.p = pack
    def __len__(self): return self.p["skill"].shape[0]
    def __getitem__(self, i):
        return dict(skill=torch.from_numpy(self.p["skill"][i]),
                    correct=torch.from_numpy(self.p["correct"][i]),
                    behav=torch.from_numpy(self.p["behav"][i]),
                    dt=torch.from_numpy(self.p["dt"][i]),
                    mask=torch.from_numpy(self.p["mask"][i]), idx=i)

def train_variant(name, kw, cfg, device):
    kw2 = {k: v for k, v in kw.items() if k != "use_persistence"}
    builder = lambda h: build_pcdt(h, **kw2)
    hidden = CAPACITY[name]["hidden"]
    tr = DataLoader(_PackDS(PACKS["train"]), batch_size=cfg.batch_size, shuffle=True)
    va = DataLoader(_PackDS(PACKS["val"]),   batch_size=cfg.batch_size, shuffle=False)

    best = dict(auc=-np.inf)
    for lr in cfg.search_lrs:
        seed_everything(SEED)
        model = builder(hidden).to(device)
        opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=cfg.weight_decay)
        b_auc, b_state, bad = -np.inf, None, 0
        for ep in range(cfg.max_epochs):
            model.train()
            for b in tr:
                b = {k:(v.to(device) if torch.is_tensor(v) else v) for k,v in b.items()}
                loss = masked_bce(model(b), b["correct"][:,1:], b["mask"][:,1:])
                opt.zero_grad(); loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), cfg.grad_clip); opt.step()
            P = np.full((PACKS["val"]["skill"].shape[0], PACKS["val"]["skill"].shape[1]-1), np.nan, np.float32)
            model.eval()
            with torch.no_grad():
                for b in va:
                    idx=b["idx"].numpy(); b={k:(v.to(device) if torch.is_tensor(v) else v) for k,v in b.items()}
                    P[idx]=torch.sigmoid(model(b)).cpu().numpy()
            m = PACKS["val"]["mask"][:,1:] > 0
            auc = roc_auc_score(PACKS["val"]["correct"][:,1:][m], P[m])
            if auc > b_auc + 1e-5: b_auc, bad, b_state = auc, 0, {k:v.detach().cpu().clone() for k,v in model.state_dict().items()}
            else:
                bad += 1
                if bad >= cfg.patience: break
        if b_auc > best["auc"]: best = dict(auc=b_auc, lr=lr, state=b_state, hidden=hidden)
    seed_everything(SEED); model = builder(best["hidden"]).to(device)
    model.load_state_dict(best["state"])
    return model, best["lr"], best["auc"]

# ---------- cache plumbing (shared key scheme with Experiment P2-02) ----------
DATA_FINGERPRINT = (f"L{CFG.max_learners}_m{CFG.min_seq_len}_S{N_SKILLS}"
                    f"_n{sum(int(PACKS[s]['mask'].sum()) for s in PACKS)}")
def _key(t): return "".join(c if (c.isalnum() or c in "-_.") else "_" for c in t)
def _p02_pcdt_tag():
    spec = (f"dims{'-'.join(CFG.pcdt_dimensions)}_psi{int(CFG.use_persistence)}_mem0_bel0")
    return _key(f"PCDT_{spec}_b{CFG.param_budget}_{DATA_FINGERPRINT}_seed{SEED}")
def _variant_tag(name):
    return _key(f"P03_{name}_ro{int(CAPACITY[name]['kw'].get('psi_readout_decay', False))}"
                f"_rs{int(CAPACITY[name]['kw'].get('reset_sessions', False))}"
                f"_b{CFG.param_budget}_{DATA_FINGERPRINT}_seed{SEED}")

def load_p02_pcdt_predictions():
    """Reload P2-02's PCDT test predictions and verify alignment to the rebuilt test pack."""
    if not CFG.cache_dir: return None
    f = os.path.join(CFG.cache_dir, _p02_pcdt_tag() + ".npz")
    if not os.path.exists(f): return None
    z = np.load(f); probs, targs, rows = z["probs"], z["targs"], z["rows"]
    # rebuild the same flat (prob,target,learner_row) stream from our pack and compare targets
    m = PACKS["test"]["mask"][:,1:] > 0
    tgt_here = PACKS["test"]["correct"][:,1:][m]
    if len(targs) != len(tgt_here) or not np.array_equal(targs.astype(np.float32), tgt_here.astype(np.float32)):
        print("  *** cached P2-02 predictions do NOT align with the rebuilt test pack "
              f"(cached n={len(targs)}, here n={len(tgt_here)}). Retraining PCDT. ***")
        return None
    # scatter the flat probs back into [n, L-1] positions
    P = np.full((PACKS["test"]["skill"].shape[0], PACKS["test"]["skill"].shape[1]-1), np.nan, np.float32)
    P[m] = probs
    print("  reused Experiment P2-02 PCDT predictions (alignment verified)")
    return P

print("Harness ready. Fingerprint:", DATA_FINGERPRINT)
````

## 8. Producing the Two Prediction Sets

The PCDT predictions are reused from Experiment P2-02 when the cache aligns; otherwise the PCDT is
retrained here. The session-independent control is trained here in every case. Each model yields a
`[n_test, L-1]` array of test probabilities aligned to the packed interactions, ready for bucket
routing.


````python
PRED = {}
t0 = time.time()
for name, kw in VARIANTS.items():
    print("="*68); print(name)
    reused = None
    if name == "PCDT (Psi)" and not CFG.psi_readout_decay:
        reused = load_p02_pcdt_predictions()
    if reused is not None:
        PRED[name] = reused; continue

    hit = None
    if CFG.cache_dir:
        f = os.path.join(CFG.cache_dir, _variant_tag(name) + ".npz")
        if os.path.exists(f):
            z = np.load(f)
            m = PACKS["test"]["mask"][:,1:] > 0
            if np.array_equal(z["targs"].astype(np.float32),
                              PACKS["test"]["correct"][:,1:][m].astype(np.float32)):
                P = np.full((PACKS["test"]["skill"].shape[0], PACKS["test"]["skill"].shape[1]-1),
                            np.nan, np.float32); P[m] = z["probs"]
                PRED[name] = P; print("  loaded from cache"); hit = True
    if hit: continue

    if not CFG.train_if_missing:
        print("  no cache and train_if_missing=False -> skipped"); continue
    model, lr, va = train_variant(name, kw, CFG, DEVICE)
    P = collect_predictions(model, PACKS["test"], DEVICE)
    PRED[name] = P
    if CFG.cache_dir:
        m = PACKS["test"]["mask"][:,1:] > 0
        np.savez_compressed(os.path.join(CFG.cache_dir, _variant_tag(name)+".npz"),
                            probs=P[m], targs=PACKS["test"]["correct"][:,1:][m],
                            rows=np.repeat(np.arange(P.shape[0])[:,None], P.shape[1], 1)[m])
    print(f"  trained (lr={lr}, val AUC={va:.4f})")
print("="*68); print(f"runtime: {(time.time()-t0)/60:.1f} min")
````

## 10. Bucketed Evaluation and Table 5

Each test prediction is routed to its gap bucket, and within every bucket the two models are scored on
the **identical** set of prediction points. For the manuscript horizons the reported quantities are the
two models' AUC and RMSE, and the **retention advantage** — the difference-in-differences of Section
0.3, with a learner-level paired-bootstrap 95% interval, Holm-corrected across the three horizons.

The evaluation functions take the pack and its bucket array as arguments, so the real-data and
simulator-fallback horizons are scored by the same code rather than two parallel implementations. A
horizon whose cell fails the pre-registered point/learner floor is reported as `inadequate` (real data)
or filled from the simulator and tagged `(sim)`; no AUC is printed for a bucket that fails the floor.


````python
def _mask_targets(pack):
    return pack["mask"][:,1:] > 0, pack["correct"][:,1:]

def bucket_eval(pred, sel, pack):
    m, tgt = _mask_targets(pack)
    ok = sel & m & ~np.isnan(pred)
    y, p = tgt[ok], pred[ok]
    if len(np.unique(y)) < 2 or len(y) < 5:
        return dict(auc=np.nan, rmse=np.nan, n=int(len(y)),
                    n_learn=int(np.unique(np.where(ok)[0]).size))
    return dict(auc=float(roc_auc_score(y, p)), rmse=float(np.sqrt(np.mean((p-y)**2))),
                n=int(len(y)), n_learn=int(np.unique(np.where(ok)[0]).size))

def _auc(y, x): return roc_auc_score(y, x) if len(np.unique(y)) > 1 else np.nan

def paired_did_bootstrap(pcdt, base, sel_h, sel_ref, pack, n_boot, rng):
    """Difference-in-differences retention advantage, learner-level paired bootstrap.
    RA = [AUC_pcdt(h) - AUC_base(h)] - [AUC_pcdt(ref) - AUC_base(ref)]."""
    m, tgt = _mask_targets(pack)
    def rows_for(sel):
        ok = sel & m & ~np.isnan(pcdt) & ~np.isnan(base)
        r, _ = np.where(ok)
        return r, tgt[ok], pcdt[ok], base[ok]
    rh, yh, ph, bh = rows_for(sel_h)
    rr, yr, pr, br = rows_for(sel_ref)
    if len(yh) < 5 or len(yr) < 5:
        return np.nan, np.nan, np.nan, np.nan
    lh = {l: np.where(rh==l)[0] for l in np.unique(rh)}
    lr = {l: np.where(rr==l)[0] for l in np.unique(rr)}
    point = (_auc(yh, ph) - _auc(yh, bh)) - (_auc(yr, pr) - _auc(yr, br))
    common = np.array(sorted(set(lh) | set(lr)))
    ds = []
    for _ in range(n_boot):
        samp = rng.choice(common, size=len(common), replace=True)
        ih = np.concatenate([lh[l] for l in samp if l in lh]) if any(l in lh for l in samp) else np.array([], int)
        ir = np.concatenate([lr[l] for l in samp if l in lr]) if any(l in lr for l in samp) else np.array([], int)
        if len(ih) < 5 or len(ir) < 5: continue
        v = (_auc(yh[ih], ph[ih]) - _auc(yh[ih], bh[ih])) - (_auc(yr[ir], pr[ir]) - _auc(yr[ir], br[ir]))
        if np.isfinite(v): ds.append(v)
    d = np.asarray(ds)
    if len(d) == 0: return float(point), np.nan, np.nan, np.nan
    lo, hi = np.percentile(d, [2.5, 97.5]); pv = 2*min((d<=0).mean(), (d>=0).mean())
    return float(point), float(lo), float(hi), float(min(pv, 1.0))

def holm(pvals):
    pv = np.array([p if np.isfinite(p) else 1.0 for p in pvals])
    order = np.argsort(pv); adj = np.empty(len(pv)); run = 0.0
    for rank, i in enumerate(order):
        run = max(run, (len(pv)-rank)*pv[i]); adj[i] = min(run, 1.0)
    return adj

def first_post_gap_mask(pack, bucket_arr):
    m = pack["mask"][:,1:] > 0
    gap_ids = [i for i,(k,*_ ) in enumerate(BUCKETS) if k in ("d1","d7","d28")]
    return m & np.isin(bucket_arr[:,:-1], gap_ids)

def ref_mask(bucket_arr):
    ri = [i for i,(k,*_ ) in enumerate(BUCKETS) if k==REF_BUCKET][0]
    return bucket_arr[:,:-1] == ri

def horizon_adequate(pack, bucket_arr, key, first_only):
    bi = [i for i,(k,*_ ) in enumerate(BUCKETS) if k==key][0]
    use = first_post_gap_mask(pack, bucket_arr) if first_only else (pack["mask"][:,1:] > 0)
    sel = use & (bucket_arr[:,:-1] == bi)
    n_pts = int(sel.sum()); n_learn = int(np.unique(np.where(sel)[0]).size)
    return (n_learn >= CFG.min_learners_bucket and n_pts >= CFG.min_points_bucket), n_learn, n_pts

def evaluate_horizons(pack, bucket_arr, pred_pcdt, pred_base, first_only):
    """Returns {bucket_key: row dict} over the manuscript horizons for one pack."""
    use = first_post_gap_mask(pack, bucket_arr) if first_only else (pack["mask"][:,1:] > 0)
    sref = ref_mask(bucket_arr)
    out, ps = {}, []
    for key in MANUSCRIPT_BUCKETS:
        bi = [i for i,(k,*_ ) in enumerate(BUCKETS) if k==key][0]
        sh = use & (bucket_arr[:,:-1] == bi)
        ep = bucket_eval(pred_pcdt, sh, pack); eb = bucket_eval(pred_base, sh, pack)
        d, lo, hi, pv = paired_did_bootstrap(pred_pcdt, pred_base, sh, sref, pack, CFG.n_bootstrap, RNG)
        out[key] = dict(label=BUCKETS[bi][3], pcdt_auc=ep["auc"], base_auc=eb["auc"],
                        pcdt_rmse=ep["rmse"], base_rmse=eb["rmse"], ra=d, lo=lo, hi=hi,
                        p=pv, n=ep["n"], n_learn=ep["n_learn"])
        ps.append(pv)
    for key, pa in zip(MANUSCRIPT_BUCKETS, holm(ps)):
        out[key]["p_holm"] = float(pa)
    return out

print("Unified evaluation helpers ready.")
````

````python
PCDT_KEY, BASE_KEY = "PCDT (Psi)", "Session-independent"

# --- real-data horizons ---
adeq_flags = {k: horizon_adequate(PACKS["test"], TEST_BUCKET, k, CFG.first_post_gap_only)
              for k in MANUSCRIPT_BUCKETS}
real_rows = evaluate_horizons(PACKS["test"], TEST_BUCKET, PRED[PCDT_KEY], PRED[BASE_KEY],
                              CFG.first_post_gap_only)
pend = [k for k in MANUSCRIPT_BUCKETS if not adeq_flags[k][0]]

print("Real-data horizon adequacy:")
for k in MANUSCRIPT_BUCKETS:
    ok, nl, npt = adeq_flags[k]
    print(f"  {real_rows[k]['label']:8s}  {'adequate' if ok else 'INADEQUATE'} "
          f"(learners={nl:,}, points={npt:,})")

# --- simulator fallback for inadequate horizons ---
SIM_ROWS = {}
if pend and CFG.run_synthetic_fallback:
    print(f"\nInadequate horizons {[real_rows[k]['label'] for k in pend]} -> simulator fallback.")
    _stash = (PACKS, N_SKILLS, SKILL2IDX, TEST_BUCKET)
    sim_df = simulate_learners(rng=RNG)
    PACKS, N_SKILLS, SKILL2IDX = build_packs(sim_df, CFG, RNG)
    _dt = np.zeros_like(PACKS["test"]["dt"]); _dt[:,:-1] = PACKS["test"]["dt"][:,1:]
    SIM_BUCKET = bucket_of_seconds(_dt)
    sim_pred = {}
    for name, kw in VARIANTS.items():
        if name.endswith("as-run)"): continue
        model, lr, va = train_variant(name, kw, CFG, DEVICE)
        sim_pred[name] = collect_predictions(model, PACKS["test"], DEVICE)
        print(f"  (sim) {name}: val AUC={va:.4f}")
    sim_all = evaluate_horizons(PACKS["test"], SIM_BUCKET, sim_pred[PCDT_KEY], sim_pred[BASE_KEY],
                                CFG.first_post_gap_only)
    SIM_ROWS = {k: sim_all[k] for k in pend}
    PACKS, N_SKILLS, SKILL2IDX, TEST_BUCKET = _stash
    print("  simulator horizons complete.")
elif pend:
    print(f"\nInadequate horizons {[real_rows[k]['label'] for k in pend]} left as 'inadequate' "
          f"(run_synthetic_fallback=False).")
````

````python
def _fmt(v, nd=4): return "--" if (v is None or (isinstance(v,float) and not np.isfinite(v))) else f"{v:.{nd}f}"

def _ra_cell(r):
    if not np.isfinite(r.get("ra", np.nan)): return "--"
    star = ""
    if np.isfinite(r.get("p_holm", np.nan)):
        if r["p_holm"] < 0.05 and abs(r["ra"]) >= CFG.min_meaningful_auc_delta: star = ""
        elif r["p_holm"] < 0.05: star = " (n.s. by size)"
        else: star = " (n.s.)"
    ci = f" [{r['lo']:+.4f}, {r['hi']:+.4f}]" if np.isfinite(r.get("lo", np.nan)) else ""
    return f"{r['ra']:+.4f}{ci}{star}"

rows = []
for k in MANUSCRIPT_BUCKETS:
    label = real_rows[k]["label"]
    if adeq_flags[k][0]:
        r = real_rows[k]; tag = ""
    elif k in SIM_ROWS:
        r = SIM_ROWS[k]; tag = " (sim)"
    else:
        rows.append({"Gap horizon": label, "PCDT AUC": "inadequate", "Baseline AUC": "inadequate",
                     "PCDT RMSE": "inadequate", "Baseline RMSE": "inadequate",
                     "Retention advantage": f"inadequate (learners={adeq_flags[k][1]:,}, "
                                            f"points={adeq_flags[k][2]:,})"})
        continue
    rows.append({"Gap horizon": label + tag,
                 "PCDT AUC": _fmt(r["pcdt_auc"]), "Baseline AUC": _fmt(r["base_auc"]),
                 "PCDT RMSE": _fmt(r["pcdt_rmse"]), "Baseline RMSE": _fmt(r["base_rmse"]),
                 "Retention advantage": _ra_cell(r)})

table5 = pd.DataFrame(rows)
print("Table 5. Cross-session predictive validity at inter-session gap horizons, "
      "PCDT (with persistence operator Psi) versus session-independent baseline.\n")
print(table5.to_markdown(index=False))
table5.to_csv("table5.csv", index=False)
with open("table5.md", "w") as fh: fh.write(table5.to_markdown(index=False))

# extended table with n and per-horizon precision, for the appendix
ext = []
for k in MANUSCRIPT_BUCKETS:
    src = "real" if adeq_flags[k][0] else ("sim" if k in SIM_ROWS else "inadequate")
    r = real_rows[k] if adeq_flags[k][0] else SIM_ROWS.get(k, {})
    ext.append({"Horizon": real_rows[k]["label"], "Source": src,
                "PCDT AUC": _fmt(r.get("pcdt_auc")), "Base AUC": _fmt(r.get("base_auc")),
                "RA (DiD)": _fmt(r.get("ra")),
                "95% CI": (f"[{r['lo']:+.4f}, {r['hi']:+.4f}]"
                           if np.isfinite(r.get("lo", np.nan)) else "--"),
                "p (Holm)": _fmt(r.get("p_holm"), 3),
                "n points": f"{r.get('n', 0):,}", "n learners": f"{r.get('n_learn', 0):,}"})
table5_ext = pd.DataFrame(ext)
print("\nExtended (appendix / reviewer response):\n")
print(table5_ext.to_markdown(index=False))
table5_ext.to_csv("table5_extended.csv", index=False)

t5_notes = [
    "Baseline is the SAME architecture and capacity as the PCDT with its state reset to zero at every "
    "session boundary (gap >= 20 min) and Psi disabled; it has no mechanism to cross an inter-session "
    "gap, which is the capability under test.",
    "Both models are scored on the IDENTICAL prediction points within each bucket. Only the first item "
    "after a qualifying gap is scored (first_post_gap_only), so later within-session items do not dilute "
    "the gap effect.",
    "Retention advantage is a difference-in-differences: (PCDT - baseline) AUC at the horizon minus "
    "(PCDT - baseline) AUC within session. It removes any standing within-session advantage so that "
    "only differential degradation is credited. 95% CI is a learner-level paired bootstrap; p is "
    "Holm-corrected across the three horizons; the pre-registered minimum meaningful effect is "
    f"{CFG.min_meaningful_auc_delta:.3f}.",
    "Per-bucket AUC levels are NOT comparable across horizons (returning learners are self-selected and "
    "meet a different item mix); read the retention-advantage column, not the raw AUC gap.",
    "Rows tagged '(sim)' are from the ACT-R simulator because the real-data cell failed the "
    f"pre-registered floor (>= {CFG.min_learners_bucket} learners and >= {CFG.min_points_bucket:,} "
    "points). Simulator results show the mechanism behaves under controlled decay; they do not "
    "establish real-world retention.",
    f"psi_readout_decay={CFG.psi_readout_decay}. With it False the PCDT is the Experiment P2-02 model "
    "(Section 0.1); if the corrected variant is adopted, Tables 2 and 3 must be regenerated to match.",
]
print("\nCaption notes:")
for i, n in enumerate(t5_notes, 1): print(f"  ({i}) {n}")
with open("table5_caption_notes.txt", "w") as fh:
    fh.write("\n".join(f"({i}) {n}" for i, n in enumerate(t5_notes, 1)))
````

## 11. Interpretation

> Written before the run. Read the tables for the actual numbers.

**If the retention advantage is positive and clears the threshold at one or more horizons** — the PCDT
degrades less than a session-resetting model across genuine gaps — the contribution is *the
cross-session evaluation protocol and the demonstration that $\Psi$ retains predictive validity*, not
the invention of forgetting. Section 4.3's framing already concedes the mechanism family to prior work;
the novel object is the evaluation at real gap horizons against a session-independent control.

**If it is null** — $\Psi$ retains no better than a resetting baseline — that is a clean, reportable
result and it reframes the paper: persistence, as currently specified, does not buy cross-session
predictive validity, and the representation's justification rests on whatever RQ2.4 (stability) and the
downstream tasks establish. Given that removing $\Psi$ cost only $-0.0004$ AUC within session
(Experiment P2-02), a null here would be *consistent*, not surprising — and if the readout-decay variant
also fails to help, that is strong evidence the effect is genuinely absent rather than an artefact of
the one-step-late application.

**If four weeks is simulator-only**, say so in the abstract, not just the limitations. A persistence
claim resting on synthetic four-week gaps is weaker than one resting on real returns, and the honest
framing states the horizon at which real-data evidence runs out.

**Read the two Tables together.** If Table 4 shows the primary dataset thin at four weeks and Table 5's
four-week row is `(sim)`, the paper's persistence claim is real-data-backed at one week and
simulator-backed at four. That is a defensible position stated plainly and an indefensible one stated by
omission.
